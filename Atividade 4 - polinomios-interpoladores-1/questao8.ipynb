# 8. Os dados do Exercício 6 foram gerados pelas funções:

# (a) f(x) = e^x
# (b) f(x) = x^4 - x^3 + x^2 - x + 1
# (c) f(x) = x^2*cos(x) - 3x
# (d) f(x) = ln(e^x + 2)

# Use a fórmula do erro para encontrar um limitante
# e compare com o erro real para n = 1 e n = 2.

```python
import numpy as np

# Função de interpolação de Lagrange

def lagrange(xp, yp, x):

    n = len(xp)  # número de pontos
    i = 0
    s = 0        # soma final (resultado do polinômio)

    # Loop externo: percorre cada termo do somatório
    while i < n:

        # Começa com o valor yi
        p = yp[i]

        j = 0
        # Loop interno: constrói o polinômio base Li(x)
        while j < n:
            if i != j:
                # Multiplica pelos fatores de Lagrange
                p *= (x - xp[j]) / (xp[i] - xp[j])
            j += 1

        # Soma o termo ao resultado final
        s += p
        i += 1

    return s

# Ponto onde queremos avaliar a interpolação
x = 0.43

# Nós de interpolação
xp = [0, 0.25, 0.5]


# ITEM (a)
# f(x) = e^x

def fa(x): return np.exp(x)

# Valores da função nos pontos conhecidos
yp = [fa(0), fa(0.25), fa(0.5)]

# Erro absoluto entre valor real e interpolado
print("8a:", abs(fa(x) - lagrange(xp, yp, x)))

# ITEM (b)
# f(x) = x^4 - x^3 + x^2 - x + 1

def fb(x): return x**4 - x**3 + x**2 - x + 1

yp = [fb(0), fb(0.25), fb(0.5)]

print("8b:", abs(fb(x) - lagrange(xp, yp, x)))


# ITEM (c)
# f(x) = x^2 cos(x) - 3x

def fc(x): return x**2 * np.cos(x) - 3*x

yp = [fc(0), fc(0.25), fc(0.5)]

print("8c:", abs(fc(x) - lagrange(xp, yp, x)))


# ITEM (d)
# f(x) = ln(e^x + 2)

def fd(x): return np.log(np.exp(x) + 2)

yp = [fd(0), fd(0.25), fd(0.5)]

print("8d:", abs(fd(x) - lagrange(xp, yp, x)))
```
