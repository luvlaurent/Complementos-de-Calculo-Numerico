# 5. Utilize polinômios interpoladores de Lagrange de graus 1, 2 e 3
# para aproximar:

# (a) f(8.4), dados:
# f(8.1)=16.94410, f(8.3)=17.56492, f(8.6)=18.50515, f(8.7)=18.82091

# (b) f(-1/3), dados:
# f(-0.75)=-0.07181250, f(-0.5)=-0.02475000,
# f(-0.25)=0.33493750, f(0)=1.10100000

# (c) f(0.25), dados:
# f(0.1)=0.62049958, f(0.2)=-0.28398668,
# f(0.3)=0.00660095, f(0.4)=0.24842440

# (d) f(0.9), dados:
# f(0.6)=-0.17694460, f(0.7)=0.01375227,
# f(0.8)=0.22363362, f(1.0)=0.65809197

```python
def lagrange(xi, yi, x):
    soma = 0
    for i in range(len(xi)):
        termo = yi[i]
        for j in range(len(xi)):
            if i != j:
                termo *= (x-xi[j])/(xi[i]-xi[j])
        soma += termo
    return soma


# -------- letra a --------
x = [8.1,8.3,8.6,8.7]
y = [16.94410,17.56492,18.50515,18.82091]
xp = 8.4

print("a)")
print("Grau1:",lagrange(x[:2],y[:2],xp))
print("Grau2:",lagrange(x[:3],y[:3],xp))
print("Grau3:",lagrange(x,y,xp))
```
