## Questão 2

Use o **método de Runge-Kutta de quarta ordem (RK4)** para obter uma aproximação para a solução do problema de valor inicial.

### (a)

$$
y' = e^{\,t-y},
\qquad
0 \le t \le 1,
$$

com condição inicial

$$
y(0)=1.
$$

Utilize

$$
h=0.2.
$$

### (b)

$$
y'=\frac{1+t}{1+y},
\qquad
1 \le t \le 2,
$$

com condição inicial

$$
y(1)=2.
$$

Utilize

$$
h=0.1.
$$


```python
import numpy as np

# RK4

def rk4(f, t0, y0, h, tf):

    n = int((tf - t0)/h)

    t = [t0]
    y = [y0]

    for i in range(n):

        k1 = h*f(t[i], y[i])

        k2 = h*f(
            t[i] + h/2,
            y[i] + k1/2
        )

        k3 = h*f(
            t[i] + h/2,
            y[i] + k2/2
        )

        k4 = h*f(
            t[i] + h,
            y[i] + k3
        )

        y_next = y[i] + (k1 + 2*k2 + 2*k3 + k4)/6

        t.append(t[i] + h)
        y.append(y_next)

    return np.array(t), np.array(y)

# ITEM A
# y' = exp(t-y)

def f1(t, y):
    return np.exp(t - y)

t1, y1 = rk4(
    f1,
    t0=0,
    y0=1,
    h=0.2,
    tf=1
)

print("\nITEM A")
print(" t\t\t y(RK4)")
for ti, yi in zip(t1, y1):
    print(f"{ti:.2f}\t {yi:.10f}")

# ITEM B
# y' = (1+t)/(1+y)

def f2(t, y):
    return (1+t)/(1+y)

t2, y2 = rk4(
    f2,
    t0=1,
    y0=2,
    h=0.1,
    tf=2
)

print("\nITEM B")
print(" t\t\t y(RK4)")
for ti, yi in zip(t2, y2):
    print(f"{ti:.2f}\t {yi:.10f}")
```
