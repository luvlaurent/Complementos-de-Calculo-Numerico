## Questão 1

Use o **método de Euler** para obter uma aproximação para a solução do problema de valor inicial

### (a)

$$
y' = 1 + (t-y)^2,
\qquad
2 \le t \le 3,
$$

com condição inicial

$$
y(2)=1.
$$

Utilize

$$
h=0.5.
$$

### (b)

$$
y' = \cos(2t)+\sin(3t),
\qquad
0 \le t \le 1,
$$

com condição inicial

$$
y(0)=1.
$$

Utilize

$$
h=0.25.
$$

```python
import numpy as np

# MÉTODO DE EULER

def euler(f, t0, y0, h, tf):
    n = int((tf - t0)/h)

    t = [t0]
    y = [y0]

    for i in range(n):
        y_next = y[i] + h*f(t[i], y[i])

        t.append(t[i] + h)
        y.append(y_next)

    return np.array(t), np.array(y)

# ITEM A
# y' = 1 + (t - y)^2

def f1(t, y):
    return 1 + (t - y)**2

t1, y1 = euler(
    f1,
    t0=2,
    y0=1,
    h=0.5,
    tf=3
)

print("\nITEM A")
print(" t\t\t y(Euler)")
for ti, yi in zip(t1, y1):
    print(f"{ti:.2f}\t {yi:.8f}")

# ITEM B
# y' = cos(2t) + sin(3t)


def f2(t, y):
    return np.cos(2*t) + np.sin(3*t)

t2, y2 = euler(
    f2,
    t0=0,
    y0=1,
    h=0.25,
    tf=1
)

print("\nITEM B")
print(" t\t\t y(Euler)")
for ti, yi in zip(t2, y2):
    print(f"{ti:.2f}\t {yi:.8f}")
```
