## Questão 3

Use o **método de Runge-Kutta para sistemas de equações diferenciais** para encontrar aproximações das soluções dos sistemas abaixo e compare os resultados com as soluções exatas.

### (a)

Considere o sistema

$$
\begin{cases}
u_1' = 3u_1 + 2u_2 - (2t^2+1)e^{2t},\\[2mm]
u_2' = 4u_1 + u_2 + (t^2+2t-4)e^{2t},
\end{cases}
$$

com condições iniciais

$$
u_1(0)=1,
\qquad
u_2(0)=1.
$$

Utilize

$$
0\le t\le1,
\qquad
h=0.1.
$$

Compare os resultados com as soluções exatas

$$
u_1(t)=
\frac13e^{5t}
-\frac13e^{-t}
+e^{2t},
$$

$$
u_2(t)=
\frac13e^{5t}
+\frac23e^{-t}
+t^2e^{2t}.
$$

### (b)

Considere o sistema

$$
\begin{cases}
u_1'=-4u_1-2u_2+\cos t+4\sin t,\\[2mm]
u_2'=3u_1+u_2-3\sin t,
\end{cases}
$$

com condições iniciais

$$
u_1(0)=0,
\qquad
u_2(0)=-1.
$$

Utilize

$$
0\le t\le2,
\qquad
h=0.1.
$$

Compare os resultados com as soluções exatas

$$
u_1(t)=
2e^{-t}
-2e^{-2t}
+\sin t,
$$

$$
u_2(t)=
-3e^{-t}
+2e^{-2t}.
$$

```python
import numpy as np
import pandas as pd

# RK4 PARA SISTEMAS

def rk4_system(F, t0, U0, h, tf):

    n = int((tf - t0)/h)

    t = [t0]
    U = [np.array(U0, dtype=float)]

    for i in range(n):

        k1 = h * F(t[i], U[i])

        k2 = h * F(
            t[i] + h/2,
            U[i] + k1/2
        )

        k3 = h * F(
            t[i] + h/2,
            U[i] + k2/2
        )

        k4 = h * F(
            t[i] + h,
            U[i] + k3
        )

        U_next = U[i] + (k1 + 2*k2 + 2*k3 + k4)/6

        t.append(t[i] + h)
        U.append(U_next)

    return np.array(t), np.array(U)

# QUESTÃO 3

print("QUESTÃO 3 - RUNGE-KUTTA PARA SISTEMAS")

# ITEM A

def F1(t, U):

    u1, u2 = U

    du1 = 3*u1 + 2*u2 - (2*t**2 + 1)*np.exp(2*t)
    du2 = 4*u1 + u2 + (t**2 + 2*t - 4)*np.exp(2*t)

    return np.array([du1, du2])

def u1_exata(t):
    return (1/3)*np.exp(5*t) - (1/3)*np.exp(-t) + np.exp(2*t)

def u2_exata(t):
    return (1/3)*np.exp(5*t) + (2/3)*np.exp(-t) + t**2*np.exp(2*t)

tA, UA = rk4_system(
    F1,
    t0=0,
    U0=[1, 1],
    h=0.1,
    tf=1
)

dfA = pd.DataFrame({
    't': tA,
    'u1_RK4': UA[:,0],
    'u1_exata': [u1_exata(x) for x in tA],
    'erro_u1': np.abs(
        UA[:,0] -
        np.array([u1_exata(x) for x in tA])
    ),
    'u2_RK4': UA[:,1],
    'u2_exata': [u2_exata(x) for x in tA],
    'erro_u2': np.abs(
        UA[:,1] -
        np.array([u2_exata(x) for x in tA])
    )
})

print("\n")
print("ITEM A")
print(dfA.round(8))

# ITEM B

def F2(t, U):

    u1, u2 = U

    du1 = -4*u1 - 2*u2 + np.cos(t) + 4*np.sin(t)
    du2 = 3*u1 + u2 - 3*np.sin(t)

    return np.array([du1, du2])

def u1b_exata(t):
    return 2*np.exp(-t) - 2*np.exp(-2*t) + np.sin(t)

def u2b_exata(t):
    return -3*np.exp(-t) + 2*np.exp(-2*t)

tB, UB = rk4_system(
    F2,
    t0=0,
    U0=[0, -1],
    h=0.1,
    tf=2
)

dfB = pd.DataFrame({
    't': tB,
    'u1_RK4': UB[:,0],
    'u1_exata': [u1b_exata(x) for x in tB],
    'erro_u1': np.abs(
        UB[:,0] -
        np.array([u1b_exata(x) for x in tB])
    ),
    'u2_RK4': UB[:,1],
    'u2_exata': [u2b_exata(x) for x in tB],
    'erro_u2': np.abs(
        UB[:,1] -
        np.array([u2b_exata(x) for x in tB])
    )
})

print("\n")
print("ITEM B")
print(dfB.round(8))

# ERRO MÁXIMO

print("\n")
print("RESUMO DOS ERROS")


print("\nITEM A")
print("Erro máximo u1 =", dfA["erro_u1"].max())
print("Erro máximo u2 =", dfA["erro_u2"].max())

print("\nITEM B")
print("Erro máximo u1 =", dfB["erro_u1"].max())
print("Erro máximo u2 =", dfB["erro_u2"].max())
```
