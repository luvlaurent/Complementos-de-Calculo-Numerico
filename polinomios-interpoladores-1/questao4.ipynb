# 4. Use o Teorema 3.3 para encontrar um limitante do erro
# para as aproximações do Exercício 2.

```python
import math

# Ponto onde queremos aproximar a função
xp = 1.4

# Nós (pontos conhecidos usados na interpolação)
x = [1, 1.25, 1.6]


# Para estimar o erro de interpolação, usamos:
# |Erro(x)| ≤ (M / (n+1)!) * |(x - x0)(x - x1)...(x - xn)|

# A derivada terceira de f(x) = e^(2x) - x é:
# f'''(x) = 8e^(2x)

M = 100  # estimativa segura (valor alto para garantir o limite)

# Fórmula do erro para n = 2:
# |Erro(xp)| ≤ (M / 3!) * |(xp - x0)(xp - x1)(xp - x2)|

erro = (M / 6) * abs((xp - x[0]) * (xp - x[1]) * (xp - x[2]))

print("Limitante aproximado:", erro)
```
