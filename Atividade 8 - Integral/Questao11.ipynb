Repita o Exercício 3 utilizando a regra do ponto médio e os resultados do Exercício 9.

```python
import numpy as np
import sympy as sp

def ponto_medio(f, a, b):

    xm = (a+b)/2

    return (b-a)*f(xm)

x = sp.symbols('x')

# a)

f_expr = x**4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0.5
b = 1

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("a)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# b)

f_expr = 2**x - 4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 0.5

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nb)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# c)

f_expr = x**2*sp.log(x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 1
b = 1.5

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nc)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# d)

f_expr = x**2*sp.exp(-x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 1

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nd)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)
```
