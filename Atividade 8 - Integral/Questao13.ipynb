# Questão 13

A regra do trapézio aplicada a ∫ᵇₐ f(x)dx fornece o valor 4, e a regra de Simpson fornece o valor 2.

Quanto é f'(1)?

```python
import sympy as sp

# Definições simbólicas
fa, fb, fm, h = sp.symbols('fa fb fm h', real=True)

# Fórmulas exatas
T = (h/2)*(fa + fb)
S = (h/6)*(fa + 4*fm + fb)

# Valores dados
eq1 = sp.Eq(T, 4)
eq2 = sp.Eq(S, 2)

# Resolver sistema (em termos de h)
sol = sp.solve([eq1, eq2], (fa + fb, fm), dict=True)

# Derivada via diferença central
# f'(1) ≈ (f(b) - f(a)) / (2h)
f_prime = (fb - fa)/(2*h)

print("Sistema resolvido:", sol)
print("Expressão geral de f'(1):", f_prime)
```
