**25.** Suponha que \( f(0) = 1, f(0,5) = 2,5, f(1) = 2 \) e \( f(0,25) = f(0,75) = \alpha \). Encontre \( \alpha \) se a regra do trapézio composta com \( n = 4 \) resultar o valor \( 1,75 \) para \( \int_{0}^{1} f(x) \, dx \).

```python
# Questão 25 - Regra do trapézio composta
# Dados do problema
n = 4
a, b = 0, 1
h = (b - a) / n  # h = 0.25

# Valores conhecidos
f0 = 1      # f(0)
f05 = 2.5   # f(0.5)
f1 = 2      # f(1)

# Valores desconhecidos (mas iguais)
# f025 = alpha
# f075 = alpha

# Regra do trapézio composta:
# I ≈ (h/2) * [f(x0) + 2*f(x1) + 2*f(x2) + 2*f(x3) + f(x4)]

# Substituindo:
# I = (0.25/2) * [f(0) + 2*f(0.25) + 2*f(0.5) + 2*f(0.75) + f(1)]
# I = 0.125 * [1 + 2α + 2*2.5 + 2α + 2]
# I = 0.125 * [1 + 2α + 5 + 2α + 2]
# I = 0.125 * [8 + 4α]

# Sabemos que I = 1.75
# Então:
# 1.75 = 0.125 * (8 + 4α)

# Resolvendo a equação:
# 1.75 / 0.125 = 8 + 4α
# 14 = 8 + 4α
# 4α = 14 - 8
# 4α = 6
# α = 6/4 = 1.5

alpha = 1.5

# Verificação
integral = 0.125 * (8 + 4*alpha)

print("QUESTÃO 25")

print(f"α = {alpha}")
print("\nVerificação:")
print(f"I = 0.125 * [f(0) + 2α + 2*f(0.5) + 2α + f(1)]")
print(f"I = 0.125 * [1 + 2*{alpha} + 2*2.5 + 2*{alpha} + 2]")
print(f"I = 0.125 * [1 + {2*alpha} + 5 + {2*alpha} + 2]")
print(f"I = 0.125 * [{1 + 2*alpha + 5 + 2*alpha + 2}]")
print(f"I = 0.125 * [{8 + 4*alpha}]")
print(f"I = 0.125 * [{8 + 4*alpha}]")
print(f"I = 0.125 * {8 + 4*alpha}")
print(f"I = {integral}")
print(f"\n✓ Aproximação = {integral}, que é igual a 1.75")
```
