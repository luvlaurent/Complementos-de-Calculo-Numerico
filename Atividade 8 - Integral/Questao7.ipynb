Repita o Exercício 3 utilizando a regra de Simpson e os resultados do Exercício 5.

```python
import numpy as np
import sympy as sp

def simpson(f, a, b):

    xm = (a+b)/2

    return (b-a)/6 * (f(a) + 4*f(xm) + f(b))

x = sp.symbols('x')

# a)

f_expr = x**4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0.5
b = 1

I_simpson = simpson(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f4 = sp.diff(f_expr, x, 4)

M4 = abs(float(f4))

erro_limite = ((b-a)**5 / 2880) * M4

erro_real = abs(I_exato - I_simpson)

print("a)")
print("Simpson =", I_simpson)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# b)

f_expr = 2**x - 4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 0.5

I_simpson = simpson(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f4 = sp.diff(f_expr, x, 4)

M4 = max(abs(float(f4.subs(x,a))),
         abs(float(f4.subs(x,b))))

erro_limite = ((b-a)**5 / 2880) * M4

erro_real = abs(I_exato - I_simpson)

print("\nb)")
print("Simpson =", I_simpson)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# c)

f_expr = x**2*sp.log(x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 1
b = 1.5

I_simpson = simpson(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f4 = sp.diff(f_expr, x, 4)

M4 = max(abs(float(f4.subs(x,a))),
         abs(float(f4.subs(x,b))))

erro_limite = ((b-a)**5 / 2880) * M4

erro_real = abs(I_exato - I_simpson)

print("\nc)")
print("Simpson =", I_simpson)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# d)

f_expr = x**2*sp.exp(-x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 1

I_simpson = simpson(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f4 = sp.diff(f_expr, x, 4)

M4 = max(abs(float(f4.subs(x,a))),
         abs(float(f4.subs(x,b))))

erro_limite = ((b-a)**5 / 2880) * M4

erro_real = abs(I_exato - I_simpson)

print("\nd)")
print("Simpson =", I_simpson)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)
```
