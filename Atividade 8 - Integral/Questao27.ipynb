**27.** Determine os valores de \( n \) e \( h \) necessários para obter uma aproximação de  
\[
\int_{0}^{\pi} e^{2x} \sin 3x \, dx
\]  
com precisão de \( 10^{-4} \). Use a:  
a. regra do trapézio composta  
b. regra de Simpson composta  
c. regra do ponto médio composta

```python
import numpy as np

def f(x):
    return np.exp(2*x) * np.sin(3*x)

def f2linhas(x):
    """Segunda derivada de f"""
    return 5*np.exp(2*x)*np.sin(3*x) + 12*np.exp(2*x)*np.cos(3*x)

def f4linhas(x):
    """Quarta derivada de f (aproximada numericamente)"""
    h = 1e-6
    return (f2linhas(x + 2*h) - 4*f2linhas(x + h) + 6*f2linhas(x) 
            - 4*f2linhas(x - h) + f2linhas(x - 2*h)) / (h**4)

a, b = 0, np.pi
tolerancia = 1e-4

# Encontrar máximos das derivadas
x_vals = np.linspace(a, b, 1000)
max_f2 = max(abs(f2linhas(x)) for x in x_vals)
max_f4 = max(abs(f4linhas(x)) for x in x_vals)


print("QUESTÃO 27 - ∫₀^π e^(2x) sen(3x) dx, precisão = 1e-4\n")


# a) Trapézio: |E| ≤ (b-a) * h²/12 * max|f''| ≤ 1e-4
h_trap = np.sqrt((12 * tolerancia) / ((b - a) * max_f2))
n_trap = max(4, int(np.ceil((b - a) / h_trap)))
h_trap = (b - a) / n_trap
print(f"\na) Trapézio composta:")
print(f"   n = {n_trap}, h = {h_trap:.6f}")

# b) Simpson: |E| ≤ (b-a) * h⁴/180 * max|f⁽⁴⁾| ≤ 1e-4
h_simp = ((180 * tolerancia) / ((b - a) * max_f4)) ** 0.25
n_simp = int(np.ceil((b - a) / h_simp))
if n_simp % 2 != 0:
    n_simp += 1
n_simp = max(4, n_simp)
h_simp = (b - a) / n_simp
print(f"\nb) Simpson composta:")
print(f"   n = {n_simp}, h = {h_simp:.6f}")

# c) Ponto médio: |E| ≤ (b-a) * h²/24 * max|f''| ≤ 1e-4
h_pm = np.sqrt((24 * tolerancia) / ((b - a) * max_f2))
n_pm = max(4, int(np.ceil((b - a) / h_pm)))
h_pm = (b - a) / n_pm
print(f"\nc) Ponto médio composta:")
print(f"   n = {n_pm}, h = {h_pm:.6f}")
```
