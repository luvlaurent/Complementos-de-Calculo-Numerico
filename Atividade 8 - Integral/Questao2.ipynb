# Questão 2

Obtenha uma aproximação das integrais a seguir utilizando a regra do trapézio.

**a)** ∫₋₀․₂₅⁰․²⁵ (cos x)² dx

**b)** ∫₋₀․⁵⁰ x ln(x+1) dx

**c)** ∫₀․⁷⁵¹․³ [(sen x)² − 2x sen(x) + 1] dx

**d)** ∫ₑ^(e+1) 1/(x ln x) dx

```python
import numpy as np

# Regra do Trapézio Simples
def trapezio(f, a, b):
    return (b - a)/2 * (f(a) + f(b))

# a) ∫₋₀․₂₅⁰․²⁵ (cos x)² dx

def fa(x):
    return np.cos(x)**2

Ia = trapezio(fa, -0.25, 0.25)

# b) ∫₋₀․⁵⁰ xln(x+1) dx

def fb(x):
    return x*np.log(x + 1)

Ib = trapezio(fb, -0.5, 0)

# c) ∫₀․⁷⁵¹․³[(sen x)²−2xsen(x)+1]dx

def fc(x):
    return np.sin(x)**2 - 2*x*np.sin(x) + 1

Ic = trapezio(fc, 0.75, 1.3)

# d) ∫ₑ^(e+1) 1/(xln(x)) dx

def fd(x):
    return 1/(x*np.log(x))

Id = trapezio(fd, np.e, np.e + 1)


# Resultados

print("a) =", Ia)
print("b) =", Ib)
print("c) =", Ic)
print("d) =", Id)
```
