17a. Aproxime \(\int_{1}^{2} x \ln x \, dx\) usando a regra do trapézio composta com \(n = 4\).

```python
import numpy as np

# 17a
def f17a(x):
    return x * np.log(x)

a, b, n = 1, 2, 4
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17a = (h / 2) * (f17a(x_vals[0]) + 2 * np.sum(f17a(x_vals[1:n])) + f17a(x_vals[n]))
print(f"17a. ∫₁² x ln x dx, n=4 -> {integral_17a:.6f}")

# 17b
def f17b(x):
    return x**3 * np.exp(x)

a, b, n = -2, 2, 4
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17b = (h / 2) * (f17b(x_vals[0]) + 2 * np.sum(f17b(x_vals[1:n])) + f17b(x_vals[n]))
print(f"17b. ∫₋₂² x³ eˣ dx, n=4 -> {integral_17b:.6f}")

# 17c
def f17c(x):
    return x**2 / (x**2 + 4)

a, b, n = 0, np.pi / 2, 6
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17c = (h / 2) * (f17c(x_vals[0]) + 2 * np.sum(f17c(x_vals[1:n])) + f17c(x_vals[n]))
print(f"17c. ∫₀^{np.pi/2:.2f} x²/(x²+4) dx, n=6 -> {integral_17c:.6f}")

# 17d
def f17d(x):
    return x**2 * np.cos(x)

a, b, n = 0, np.pi, 6
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17d = (h / 2) * (f17d(x_vals[0]) + 2 * np.sum(f17d(x_vals[1:n])) + f17d(x_vals[n]))
print(f"17d. ∫₀^π x² cos x dx, n=6 -> {integral_17d:.6f}")

# 17e
def f17e(x):
    return np.exp(2 * x) * np.sin(3 * x)

a, b, n = 0, np.pi, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17e = (h / 2) * (f17e(x_vals[0]) + 2 * np.sum(f17e(x_vals[1:n])) + f17e(x_vals[n]))
print(f"17e. ∫₀^π e²ˣ sen(3x) dx, n=8 -> {integral_17e:.6f}")

# 17f
def f17f(x):
    return x / (x**2 + 4)

a, b, n = 1, 3, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17f = (h / 2) * (f17f(x_vals[0]) + 2 * np.sum(f17f(x_vals[1:n])) + f17f(x_vals[n]))
print(f"17f. ∫₁³ x/(x²+4) dx, n=8 -> {integral_17f:.6f}")

# 17g
def f17g(x):
    return 1 / np.sqrt(x**2 - 4)

a, b, n = 3, 5, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17g = (h / 2) * (f17g(x_vals[0]) + 2 * np.sum(f17g(x_vals[1:n])) + f17g(x_vals[n]))
print(f"17g. ∫₃⁵ 1/√(x²-4) dx, n=8 -> {integral_17g:.6f}")

# 17h
def f17h(x):
    return np.tan(x)

a, b, n = 0, 3 * np.pi / 8, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_17h = (h / 2) * (f17h(x_vals[0]) + 2 * np.sum(f17h(x_vals[1:n])) + f17h(x_vals[n]))
print(f"17h. ∫₀^{3*np.pi/8:.3f} tg x dx, n=8 -> {integral_17h:.6f}")
```
