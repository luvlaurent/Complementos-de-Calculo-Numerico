# Questão 14

A regra do trapézio aplicada a ∫₁² f(x)dx fornece o valor 5, e a regra do ponto médio fornece o valor 4.

Que valor resulta da regra de Simpson?

```python
def questao14(T, M):
    """
    Calcula a regra de Simpson a partir das regras do trapézio e do ponto médio.

    Parâmetros:
    T : float - valor da regra do trapézio (simples)
    M : float - valor da regra do ponto médio (simples)

    Retorna:
    S : float - valor da regra de Simpson (1/3 simples)
    """
    S = (2/3)*M + (1/3)*T
    return S

if __name__ == "__main__":
    T = 5   # dado
    M = 4   # dado
    S = questao14(T, M)
    print(f"Regra do trapézio = {T}, Regra do ponto médio = {M}")
    print(f"Regra de Simpson = {S:.6f}  (valor exato: 13/3 ≈ 4.333333)")
```
