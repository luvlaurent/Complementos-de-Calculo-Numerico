# Questão 1

Obtenha uma aproximação das seguintes integrais utilizando a regra do trapézio.

**a)** ∫₀․₅¹ x⁴ dx

**b)** ∫₀⁰․⁵ (2ˣ − 4) dx

**c)** ∫₁¹․⁵ x² ln(x) dx

**d)** ∫₀¹ x²e⁻ˣ dx

**e)** ∫₁¹․⁶ (2x)/(x²−4) dx

**f)** ∫₀⁰․³⁵ 2/(x²−4) dx

**g)** ∫₀^(π/4) x·sen(x) dx

**h)** ∫₀³ e³ˣ·sen(2x) dx

```python
import numpy as np

# Regra do Trapézio Simples
def trapezio(f, a, b):
    return (b - a)/2 * (f(a) + f(b))

# a) ∫₀․₅¹ x⁴ dx

def fa(x):
    return x**4

Ia = trapezio(fa, 0.5, 1)

# b) ∫₀⁰․⁵ (2^x − 4) dx

def fb(x):
    return 2**x - 4

Ib = trapezio(fb, 0, 0.5)

# c) ∫₁¹․⁵ x²ln(x) dx

def fc(x):
    return x**2 * np.log(x)

Ic = trapezio(fc, 1, 1.5)

# d) ∫₀¹ x²e⁻ˣ dx

def fd(x):
    return x**2 * np.exp(-x)

Id = trapezio(fd, 0, 1)

# e) ∫₁¹․⁶ (2x)/(x²−4) dx

def fe(x):
    return (2*x)/(x**2 - 4)

Ie = trapezio(fe, 1, 1.6)

# f) ∫₀⁰․³⁵ 2/(x²−4) dx

def ff(x):
    return 2/(x**2 - 4)

If = trapezio(ff, 0, 0.35)

# g) ∫₀^(π/4) x sen(x) dx

def fg(x):
    return x*np.sin(x)

Ig = trapezio(fg, 0, np.pi/4)

# h) ∫₀³ e^(3x) sen(2x) dx

def fh(x):
    return np.exp(3*x)*np.sin(2*x)

Ih = trapezio(fh, 0, 3)

# Resultados

print("a) =", Ia)
print("b) =", Ib)
print("c) =", Ic)
print("d) =", Id)
print("e) =", Ie)
print("f) =", If)
print("g) =", Ig)
print("h) =", Ih)
```
