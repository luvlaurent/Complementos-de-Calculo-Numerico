**28.** Repita o Exercício 27 para a integral  
\[
\int_{0}^{\pi} x^2 \cos x \, dx.
\]

```python
import numpy as np

# Função e suas derivadas
def f(x):
    return x**2 * np.cos(x)

# Derivadas
def f_linha(x):
    return 2*x*np.cos(x) - x**2 * np.sin(x)

def f_2linhas(x):
    return 2*np.cos(x) - 2*x*np.sin(x) - 2*x*np.sin(x) - x**2 * np.cos(x)
    # = 2cos(x) - 4x sen(x) - x² cos(x)

def f_4linhas(x):
    h = 1e-5
    return (f_2linhas(x + 2*h) - 4*f_2linhas(x + h) + 6*f_2linhas(x)
            - 4*f_2linhas(x - h) + f_2linhas(x - 2*h)) / (h**4)

# Encontrar máximos no intervalo [0, π]
x_vals = np.linspace(0, np.pi, 1000)

# a) Trapézio
max_f2 = max(abs(f_2linhas(x)) for x in x_vals)
h_trap = np.sqrt((12 * 1e-4) / (np.pi * max_f2))
n_trap = int(np.ceil((np.pi - 0) / h_trap))
h_trap = np.pi / n_trap

# b) Simpson
max_f4 = max(abs(f_4linhas(x)) for x in x_vals)
h_simp = ( (180 * 1e-4) / (np.pi * max_f4) ) ** 0.25
n_simp = int(np.ceil((np.pi - 0) / h_simp))
if n_simp % 2 != 0:
    n_simp += 1
h_simp = np.pi / n_simp

# c) Ponto médio
max_f2_pm = max_f2
h_pm = np.sqrt((24 * 1e-4) / (np.pi * max_f2_pm))
n_pm = int(np.ceil((np.pi - 0) / h_pm))
h_pm = np.pi / n_pm


print("QUESTÃO 28 - ∫₀^π x² cos x dx, precisão = 1e-4")

print(f"a. Trapézio composta: n = {n_trap}, h = {h_trap:.6f}")
print(f"b. Simpson composta:   n = {n_simp}, h = {h_simp:.6f}")
print(f"c. Ponto médio composta: n = {n_pm}, h = {h_pm:.6f}")
```
