**31.** Seja \( f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-(1/2)(x/\sigma)^2} \).  
Obtenha uma aproximação, com precisão de \( 10^{-5} \), da probabilidade de um valor escolhido aleatoriamente estar em:

a. \([- \sigma, \sigma]\)  
b. \([-2\sigma, 2\sigma]\)  
c. \([-3\sigma, 3\sigma]\)

```python
import numpy as np

def normal_pdf(x, sigma=1):
    """Função densidade da normal N(0, σ)"""
    return (1/(sigma * np.sqrt(2*np.pi))) * np.exp(-0.5*(x/sigma)**2)

def simpson_adaptativa(f, a, b, tol, max_recursao=20):
    """Regra de Simpson adaptativa para garantir precisão"""
    def _simpson(a, b):
        return (b-a)/6 * (f(a) + 4*f((a+b)/2) + f(b))
    
    def _adaptativa(a, b, tol, nivel):
        m = (a + b) / 2
        S = _simpson(a, b)
        S_esq = _simpson(a, m)
        S_dir = _simpson(m, b)
        
        if nivel >= max_recursao or abs(S_esq + S_dir - S) < 15 * tol:
            return S_esq + S_dir
        return (_adaptativa(a, m, tol/2, nivel+1) + 
                _adaptativa(m, b, tol/2, nivel+1))
    
    return _adaptativa(a, b, tol, 0)

sigma = 1
tolerancia = 1e-5


print("QUESTÃO 31 - Probabilidades da distribuição normal N(0,σ)\n")

print(f"Precisão requerida: {tolerancia}\n")


# a) P(-σ ≤ X ≤ σ)
a, b = -1, 1
prob_a = simpson_adaptativa(lambda x: normal_pdf(x, sigma), a, b, tolerancia)
print(f"\na) P(-σ ≤ X ≤ σ) = P(-1 ≤ Z ≤ 1)")
print(f"   Aproximação = {prob_a:.10f}")

# b) P(-2σ ≤ X ≤ 2σ)
a, b = -2, 2
prob_b = simpson_adaptativa(lambda x: normal_pdf(x, sigma), a, b, tolerancia)
print(f"\nb) P(-2σ ≤ X ≤ 2σ) = P(-2 ≤ Z ≤ 2)")
print(f"   Aproximação = {prob_b:.10f}")

# c) P(-3σ ≤ X ≤ 3σ)
a, b = -3, 3
prob_c = simpson_adaptativa(lambda x: normal_pdf(x, sigma), a, b, tolerancia)
print(f"\nc) P(-3σ ≤ X ≤ 3σ) = P(-3 ≤ Z ≤ 3)")
print(f"   Aproximação = {prob_c:.10f}")

print("\n" + "-" * 70)
print("Valores teóricos conhecidos (para verificação):")
print("P(-1 ≤ Z ≤ 1) ≈ 0.6826894921")
print("P(-2 ≤ Z ≤ 2) ≈ 0.9544997361")
print("P(-3 ≤ Z ≤ 3) ≈ 0.9973002039")

# Calculando os erros aproximados
erro_a = abs(prob_a - 0.6826894921)
erro_b = abs(prob_b - 0.9544997361)
erro_c = abs(prob_c - 0.9973002039)

print(f"\nErros aproximados:")
print(f"a) {erro_a:.2e}")
print(f"b) {erro_b:.2e}")
print(f"c) {erro_c:.2e}")
```
