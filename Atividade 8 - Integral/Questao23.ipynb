**23.** Obtenha uma aproximação de \(\int_{1}^{2} x^2 \ln(x^2 + 1) \, dx\) usando \(h = 0,25\) com:

a. regra do trapézio composta  
b. regra de Simpson composta  
c. regra do ponto médio composta

```python
import numpy as np

f = lambda x: x**2 * np.log(x**2 + 1)

a, b = 1, 2
h = 0.25
n = int((b - a) / h)  # n = 4

print("QUESTÃO 23 - ∫₁² x² ln(x²+1) dx, h = 0.25\n")


# a) Trapézio composta
x = np.linspace(a, b, n + 1)
trap = (h/2) * (f(x[0]) + 2 * np.sum(f(x[1:n])) + f(x[n]))
print(f"a) Trapézio composta: {trap:.6f}")

# b) Simpson composta (n precisa ser par)
if n % 2 != 0:
    n_simp = n + 1
    h_simp = (b - a) / n_simp
    x_simp = np.linspace(a, b, n_simp + 1)
else:
    n_simp = n
    h_simp = h
    x_simp = x

simpson = (h_simp/3) * (f(x_simp[0]) + f(x_simp[n_simp]) + 
                        4 * np.sum(f(x_simp[1:n_simp:2])) + 
                        2 * np.sum(f(x_simp[2:n_simp-1:2])))
print(f"b) Simpson composta: {simpson:.6f}")

# c) Ponto médio composta
ponto_medio = 0
for i in range(n):
    xm = a + (i + 0.5) * h
    ponto_medio += f(xm)
ponto_medio *= h
print(f"c) Ponto médio composta: {ponto_medio:.6f}")
```
