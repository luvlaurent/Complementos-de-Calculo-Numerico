Repita o Exercício 1 utilizando a regra de Simpson.

```python
import numpy as np

# Regra de Simpson Simples
def simpson(f, a, b):

    xm = (a + b)/2

    return (b - a)/6 * (f(a) + 4*f(xm) + f(b))

# a) ∫₀․₅¹ x⁴ dx

def fa(x):
    return x**4

Ia = simpson(fa, 0.5, 1)

# b) ∫₀⁰․⁵ (2^x−4) dx

def fb(x):
    return 2**x - 4

Ib = simpson(fb, 0, 0.5)

# c) ∫₁¹․⁵ x²ln(x) dx

def fc(x):
    return x**2*np.log(x)

Ic = simpson(fc, 1, 1.5)

# d) ∫₀¹ x²e⁻ˣ dx

def fd(x):
    return x**2*np.exp(-x)

Id = simpson(fd, 0, 1)

# e) ∫₁¹․⁶ (2x)/(x²−4) dx

def fe(x):
    return (2*x)/(x**2 - 4)

Ie = simpson(fe, 1, 1.6)

# f) ∫₀⁰․³⁵ 2/(x²−4) dx

def ff(x):
    return 2/(x**2 - 4)

If = simpson(ff, 0, 0.35)

# g) ∫₀^(π/4) xsen(x) dx

def fg(x):
    return x*np.sin(x)

Ig = simpson(fg, 0, np.pi/4)

# h) ∫₀³ e^(3x)sen(2x) dx

def fh(x):
    return np.exp(3*x)*np.sin(2*x)

Ih = simpson(fh, 0, 3)

print("a) =", Ia)
print("b) =", Ib)
print("c) =", Ic)
print("d) =", Id)
print("e) =", Ie)
print("f) =", If)
print("g) =", Ig)
print("h) =", Ih)
```
