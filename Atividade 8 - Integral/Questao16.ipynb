# Questão 16

Aproxime as seguintes integrais usando as fórmulas (4.25) a (4.32). Qual das partes (c) e (d) fornece a melhor aproximação?

**a)** ∫₂²․⁵ (ln x)³/(3x) dx

**b)** ∫₀․⁵¹ 5xe^(x²) dx

**c)** ∫₁¹⁰ 1/x dx

**d)** ∫₁⁵․⁵ 1/x dx + ∫₅․⁵¹⁰ 1/x dx

```python
# QUESTÃO 16

import numpy as np

# MÉTODO DE SIMPSON

def simpson(f, a, b):
    h = b - a
    m = (a + b)/2
    return (h/6)*(f(a) + 4*f(m) + f(b))

# FUNÇÕES

f_a = lambda x: (np.log(x)**3)/(3*x)
f_b = lambda x: 5*x*np.exp(x**2)
f_c = lambda x: 1/x

# CÁLCULOS

# Itens a e b
Ia = simpson(f_a, 2, 2.5)
Ib = simpson(f_b, 0.5, 1)

# Comparação pedida
Ic = simpson(f_c, 1, 10)

Id = simpson(f_c, 1, 5.5) + simpson(f_c, 5.5, 10)

# RESULTADOS

print("Item a:", Ia)
print("Item b:", Ib)

print("\nItem c (intervalo único):", Ic)
print("Item d (intervalo dividido):", Id)

print("\nDiferença |d - c|:", abs(Id - Ic))
```
