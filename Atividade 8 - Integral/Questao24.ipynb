**24.** Obtenha uma aproximação de \(\int_{0}^{2} x^2 e^{-x^2} \, dx\) usando \(h = 0,25\) com:

a. regra do trapézio composta

b. regra de Simpson composta

c. regra do ponto médio composta

```python
import numpy as np

f = lambda x: x**2 * np.exp(-x**2)
a, b, h = 0, 2, 0.25
n = int((b - a) / h)  # n = 8

x = np.linspace(a, b, n + 1)

# Trapézio
trap = (h/2) * (f(x[0]) + 2*np.sum(f(x[1:n])) + f(x[n]))

# Simpson
if n % 2 == 0:
    simpson = (h/3) * (f(x[0]) + f(x[n]) + 4*np.sum(f(x[1:n:2])) + 2*np.sum(f(x[2:n-1:2])))
else:
    simpson = None

# Ponto médio
ponto_medio = 0
for i in range(n):
    xm = a + (i + 0.5) * h
    ponto_medio += f(xm)
ponto_medio *= h

print("QUESTÃO 24")
print(f"a. Trapézio composta (h=0.25): {trap:.6f}")
print(f"b. Simpson composta (h=0.25): {simpson:.6f}")
print(f"c. Ponto médio composta (h=0.25): {ponto_medio:.6f}")
```
