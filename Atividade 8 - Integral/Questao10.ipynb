Repita o Exercício 2 utilizando a regra do ponto médio.

```python
import numpy as np

def ponto_medio(f, a, b):

    xm = (a + b)/2

    return (b-a)*f(xm)


def fa(x):
    return np.cos(x)**2

Ia = ponto_medio(fa, -0.25, 0.25)


def fb(x):
    return x*np.log(x+1)

Ib = ponto_medio(fb, -0.5, 0)


def fc(x):
    return np.sin(x)**2 - 2*x*np.sin(x) + 1

Ic = ponto_medio(fc, 0.75, 1.3)


def fd(x):
    return 1/(x*np.log(x))

Id = ponto_medio(fd, np.e, np.e + 1)


print("a) =", Ia)
print("b) =", Ib)
print("c) =", Ic)
print("d) =", Id)
```
