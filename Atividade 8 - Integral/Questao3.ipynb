# Questão 3

Encontre um limitante para o erro no Exercício 1 utilizando a fórmula de erro e compare-o ao erro real.

```python
import numpy as np
import sympy as sp

def trapezio(f, a, b):
    return (b-a)/2 * (f(a) + f(b))

x = sp.symbols('x')

# a)

f_expr = x**4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0.5
b = 1

I_trap = trapezio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato - I_trap)

print("a)")
print("Trapézio =", I_trap)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# b)

f_expr = 2**x - 4

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 0.5

I_trap = trapezio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato - I_trap)

print("\nb)")
print("Trapézio =", I_trap)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# c)

f_expr = x**2*sp.log(x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 1
b = 1.5

I_trap = trapezio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato - I_trap)

print("\nc)")
print("Trapézio =", I_trap)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# d)

f_expr = x**2*sp.exp(-x)

f = sp.lambdify(x, f_expr, 'numpy')

a = 0
b = 1

I_trap = trapezio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato - I_trap)

print("\nd)")
print("Trapézio =", I_trap)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)
```
