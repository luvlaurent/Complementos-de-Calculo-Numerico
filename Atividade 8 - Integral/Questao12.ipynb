Repita o Exercício 4 utilizando a regra do ponto médio e os resultados do Exercício 10.

```python
import numpy as np
import sympy as sp

def ponto_medio(f, a, b):

    xm = (a+b)/2

    return (b-a)*f(xm)

x = sp.symbols('x')

# a)

f_expr = sp.cos(x)**2

f = sp.lambdify(x, f_expr, 'numpy')

a = -0.25
b = 0.25

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("a)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# b)

f_expr = x*sp.log(x+1)

f = sp.lambdify(x, f_expr, 'numpy')

a = -0.5
b = 0

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nb)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# c)

f_expr = sp.sin(x)**2 - 2*x*sp.sin(x) + 1

f = sp.lambdify(x, f_expr, 'numpy')

a = 0.75
b = 1.3

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nc)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)


# d)

f_expr = 1/(x*sp.log(x))

f = sp.lambdify(x, f_expr, 'numpy')

a = np.e
b = np.e + 1

I_pm = ponto_medio(f, a, b)

I_exato = float(sp.integrate(f_expr, (x,a,b)))

f2 = sp.diff(f_expr, x, 2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/24)*M2

erro_real = abs(I_exato - I_pm)

print("\nd)")
print("Ponto Médio =", I_pm)
print("Exato =", I_exato)
print("Erro real =", erro_real)
print("Limitante =", erro_limite)
```
