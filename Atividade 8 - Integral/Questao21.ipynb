**21.** Utilize a regra do ponto médio composta com \( n + 2 \) subintervalos para obter uma aproximação das integrais do Exercício 1.

**Exercício 1 (original):**
a) \(\int_{0.5}^{1} x^4 \, dx\)
b) \(\int_{0}^{0.5} \frac{2}{x - 4} \, dx\)
c) \(\int_{1}^{1.5} x^2 \ln x \, dx\)
d) \(\int_{0}^{1} x^2 e^{-x} \, dx\)
e) \(\int_{1}^{1.6} \frac{2x}{x^2 - 4} \, dx\)
f) \(\int_{0}^{0.35} \frac{2}{x^2 - 4} \, dx\)
g) \(\int_{0}^{\pi/4} x \sin x \, dx\)
h) \(\int_{0}^{3} e^{3x} \sin(2x) \, dx\)

Para cada integral, use \( n \) conforme indicado no Exercício 17 original (da primeira imagem do dia). Como o Exercício 17 original não foi totalmente repetido aqui, usarei \( n = 4 \) para todas, exceto onde outro \( n \) for natural.

```python
import numpy as np

def ponto_medio_composta(f, a, b, n):
    """Regra do ponto médio composta com n subintervalos"""
    h = (b - a) / n
    soma = 0
    for i in range(n):
        x_medio = a + (i + 0.5) * h
        soma += f(x_medio)
    return h * soma

print("QUESTÃO 21 - Regra do ponto médio composta (n+2 subintervalos)")

# Usando n = 4 para todas (baseado no padrão do Exercício 17)
n_base = 4
n_medio = n_base + 2  # n + 2 subintervalos

# 21a
f = lambda x: x**4
a, b = 0.5, 1
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21a. ∫₀.₅¹ x⁴ dx, n={n_medio} -> {resultado:.6f}")

# 21b
f = lambda x: 2 / (x - 4)
a, b = 0, 0.5
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21b. ∫₀⁰·⁵ 2/(x-4) dx, n={n_medio} -> {resultado:.6f}")

# 21c
f = lambda x: x**2 * np.log(x)
a, b = 1, 1.5
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21c. ∫₁¹·⁵ x² ln x dx, n={n_medio} -> {resultado:.6f}")

# 21d
f = lambda x: x**2 * np.exp(-x)
a, b = 0, 1
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21d. ∫₀¹ x² e⁻ˣ dx, n={n_medio} -> {resultado:.6f}")

# 21e
f = lambda x: (2*x) / (x**2 - 4)
a, b = 1, 1.6
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21e. ∫₁¹·⁶ 2x/(x²-4) dx, n={n_medio} -> {resultado:.6f}")

# 21f
f = lambda x: 2 / (x**2 - 4)
a, b = 0, 0.35
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21f. ∫₀⁰·³⁵ 2/(x²-4) dx, n={n_medio} -> {resultado:.6f}")

# 21g
f = lambda x: x * np.sin(x)
a, b = 0, np.pi/4
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21g. ∫₀^{np.pi/4:.3f} x sen x dx, n={n_medio} -> {resultado:.6f}")

# 21h
f = lambda x: np.exp(3*x) * np.sin(2*x)
a, b = 0, 3
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"21h. ∫₀³ e³ˣ sen(2x) dx, n={n_medio} -> {resultado:.6f}")
```
