# Questão 15

Obtenha uma aproximação das integrais a seguir usando as Equações (4.25) a (4.32). As precisões das aproximações são consistentes com as fórmulas de erro?

**a)** ∫₀¹ √(1+x) dx

**b)** ∫₀^(π/2) (sen x)² dx

**c)** ∫₁․₁¹․⁵ eˣ dx

**d)** ∫₀¹ x^(1/3) dx

```python
# QUESTÃO 15

import numpy as np

# MÉTODOS NUMÉRICOS

def trapezio_simples(f, a, b):
    h = b - a
    return (h/2)*(f(a) + f(b))

def ponto_medio(f, a, b):
    h = b - a
    m = (a + b)/2
    return h * f(m)

def simpson(f, a, b):
    h = b - a
    m = (a + b)/2
    return (h/6)*(f(a) + 4*f(m) + f(b))

# FUNÇÕES DO PROBLEMA

f_a = lambda x: np.sqrt(1 + x)
f_b = lambda x: np.sin(x)**2
f_c = lambda x: np.exp(x)
f_d = lambda x: x**(1/3)

# INTERVALOS

problemas = [
    ("a", f_a, 0, 1),
    ("b", f_b, 0, np.pi/2),
    ("c", f_c, 1.1, 1.5),
    ("d", f_d, 0, 1)
]

# EXECUÇÃO

for nome, f, a, b in problemas:
    T = trapezio_simples(f, a, b)
    M = ponto_medio(f, a, b)
    S = simpson(f, a, b)

    print(f"\nItem {nome}:")
    print("Trapézio:", T)
    print("Ponto médio:", M)
    print("Simpson:", S)
```
