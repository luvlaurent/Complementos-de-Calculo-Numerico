Repita o Exercício 1 utilizando a regra do ponto médio.

```python
import numpy as np

def ponto_medio(f, a, b):

    xm = (a + b)/2

    return (b-a)*f(xm)


def fa(x):
    return x**4

Ia = ponto_medio(fa, 0.5, 1)


def fb(x):
    return 2**x - 4

Ib = ponto_medio(fb, 0, 0.5)


def fc(x):
    return x**2*np.log(x)

Ic = ponto_medio(fc, 1, 1.5)


def fd(x):
    return x**2*np.exp(-x)

Id = ponto_medio(fd, 0, 1)


def fe(x):
    return (2*x)/(x**2 - 4)

Ie = ponto_medio(fe, 1, 1.6)


def ff(x):
    return 2/(x**2 - 4)

If = ponto_medio(ff, 0, 0.35)


def fg(x):
    return x*np.sin(x)

Ig = ponto_medio(fg, 0, np.pi/4)


def fh(x):
    return np.exp(3*x)*np.sin(2*x)

Ih = ponto_medio(fh, 0, 3)


print("a) =", Ia)
print("b) =", Ib)
print("c) =", Ic)
print("d) =", Id)
print("e) =", Ie)
print("f) =", If)
print("g) =", Ig)
print("h) =", Ih)
```
