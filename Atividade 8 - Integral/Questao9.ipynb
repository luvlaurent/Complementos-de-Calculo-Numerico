Repita o Exercício 1 utilizando a regra do ponto médio.

```python
"""
Regra do ponto médio simples: ∫_a^b f(x) dx ≈ (b - a) * f((a + b)/2)

Integrais do Exercício 1:
a) ∫_{0.5}^{1} x^4 dx
b) ∫_{0}^{0.5} 2/(x-4) dx
c) ∫_{1}^{1.5} x^2 ln(x) dx
d) ∫_{0}^{1} x^2 e^{-x} dx
e) ∫_{1}^{1.6} (2x)/(x^2-4) dx
f) ∫_{0}^{0.35} 2/(x^2-4) dx
g) ∫_{0}^{π/4} x sin(x) dx
h) ∫_{0}^{3} e^{3x} sin(2x) dx
"""

import numpy as np
import math

# Definição das funções integrandas
def f_a(x):
    return x**4

def f_b(x):
    return 2 / (x - 4)

def f_c(x):
    return x**2 * np.log(x)

def f_d(x):
    return x**2 * np.exp(-x)

def f_e(x):
    return (2*x) / (x**2 - 4)

def f_f(x):
    return 2 / (x**2 - 4)

def f_g(x):
    return x * np.sin(x)

def f_h(x):
    return np.exp(3*x) * np.sin(2*x)

# Intervalos de integração
intervals = [
    ('a', f_a, 0.5, 1.0),
    ('b', f_b, 0.0, 0.5),
    ('c', f_c, 1.0, 1.5),
    ('d', f_d, 0.0, 1.0),
    ('e', f_e, 1.0, 1.6),
    ('f', f_f, 0.0, 0.35),
    ('g', f_g, 0.0, np.pi/4),
    ('h', f_h, 0.0, 3.0)
]

def midpoint_rule(f, a, b):
    """Regra do ponto médio simples."""
    mid = (a + b) / 2.0
    return (b - a) * f(mid)

print("Questão 9 - Aproximações pela Regra do Ponto Médio Simples\n")
for label, func, a, b in intervals:
    approx = midpoint_rule(func, a, b)
    print(f"{label}) ∫_{a}^{b} f(x) dx ≈ {approx:.10f}")
```
