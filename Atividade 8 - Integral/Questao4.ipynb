# Questão 4

Encontre um limitante para o erro no Exercício 2 utilizando a fórmula de erro e compare-o ao erro real.

```python
import numpy as np
import sympy as sp

def trapezio(f, a, b):
    return (b-a)/2 * (f(a) + f(b))

x = sp.symbols('x')

# a)

f_expr = sp.cos(x)**2

f = sp.lambdify(x, f_expr, 'numpy')

a = -0.25
b = 0.25

I_trap = trapezio(f,a,b)

I_exato = float(sp.integrate(f_expr,(x,a,b)))

f2 = sp.diff(f_expr,x,2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato - I_trap)

print("a)")
print("Trapézio =",I_trap)
print("Exato =",I_exato)
print("Erro real =",erro_real)
print("Limitante =",erro_limite)


# b)

f_expr = x*sp.log(x+1)

f = sp.lambdify(x,f_expr,'numpy')

a = -0.5
b = 0

I_trap = trapezio(f,a,b)

I_exato = float(sp.integrate(f_expr,(x,a,b)))

f2 = sp.diff(f_expr,x,2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato-I_trap)

print("\nb)")
print("Trapézio =",I_trap)
print("Exato =",I_exato)
print("Erro real =",erro_real)
print("Limitante =",erro_limite)


# c)

f_expr = sp.sin(x)**2 - 2*x*sp.sin(x) + 1

f = sp.lambdify(x,f_expr,'numpy')

a = 0.75
b = 1.3

I_trap = trapezio(f,a,b)

I_exato = float(sp.integrate(f_expr,(x,a,b)))

f2 = sp.diff(f_expr,x,2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato-I_trap)

print("\nc)")
print("Trapézio =",I_trap)
print("Exato =",I_exato)
print("Erro real =",erro_real)
print("Limitante =",erro_limite)


# d)

f_expr = 1/(x*sp.log(x))

f = sp.lambdify(x,f_expr,'numpy')

a = np.e
b = np.e + 1

I_trap = trapezio(f,a,b)

I_exato = float(sp.integrate(f_expr,(x,a,b)))

f2 = sp.diff(f_expr,x,2)

M2 = max(abs(float(f2.subs(x,a))),
         abs(float(f2.subs(x,b))))

erro_limite = ((b-a)**3/12)*M2

erro_real = abs(I_exato-I_trap)

print("\nd)")
print("Trapézio =",I_trap)
print("Exato =",I_exato)
print("Erro real =",erro_real)
print("Limitante =",erro_limite)
```
