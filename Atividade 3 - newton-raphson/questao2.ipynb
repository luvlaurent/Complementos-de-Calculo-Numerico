# 2. Seja f(x) = -x^3 - cos(x) e p0 = -1.
# Utilize o método de Newton para determinar p2.
# Pergunta: Poderíamos utilizar p0 = 0?

```python
import math

def f(x):
    return -x**3 - math.cos(x)

def df(x):
    return -3*x**2 + math.sin(x)

def newton(p0, n_iter):
    
    p = p0
    
    for i in range(n_iter):
        
        if abs(df(p)) < 1e-12:
            print("Derivada zero -> método falha")
            return
        
        p = p - f(p)/df(p)
        print(f"p{i+1} =",p)


# -------- p0 = -1 --------
print("Com p0 = -1")
newton(-1,2)


# -------- p0 = 0 --------
print("\nTestando p0 = 0")
newton(0,2)
```
