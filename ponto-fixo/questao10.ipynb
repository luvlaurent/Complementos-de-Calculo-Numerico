# Aplique a iteração de ponto fixo para encontrar uma aproximação
# com precisão de 1e-4.

# Utilize o Corolário 2.5 para estimar o número de iterações
# necessárias e compare com o valor real obtido.

```python
import math

def g(x):
    return 2**(-x)

p = 0.5
tol = 1e-4
erro = 1
iteracao = 0

while erro > tol:
    p_new = g(p)
    erro = abs(p_new - p)

    p = p_new
    iteracao += 1

print(f"Ponto fixo: {p:.6f}")
print(f"Iterações: {iteracao}")
```
