Suponha que xⱼ = j, para j = 0, 1, 2, 3, e que seja sabido que

\[
P_{0,1}(x) = x + 1, \quad P_{1,2}(x) = 3x - 1, \quad \text{e} \quad P_{1,2,3}(1,5) = 4.
\]

Encontre P₀,₁,₂,₃(1,5).

```python


import numpy as np

x0 = 1.5

P01 = x0 + 1
P12 = 3*x0 - 1
P123 = 4

# interpolação final
P012 = ((x0 - 2)*P01 + (0 - x0)*P12)/(0 - 2)
P0123 = ((x0 - 3)*P012 + (0 - x0)*P123)/(0 - 3)

print(P0123)
```
