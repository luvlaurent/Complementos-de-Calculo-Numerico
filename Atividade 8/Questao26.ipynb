**26.** A regra do ponto médio simples para \(\int_{-1}^{1} f(x) \, dx\) resulta em 12, a regra do ponto médio composta com \(n = 2\) resulta em 5 e a regra de Simpson composta resulta em 6. Use \( f(-1) = f(1) \) e \( f(-0,5) = f(0,5) - 1 \) para determinar \( f(-1), f(-0,5), f(0), f(0,5), f(1) \).

```python
# Vamos resolver simbolicamente
# Seja:
# A = f(-1) = f(1)
# B = f(-0.5)
# C = f(0)
# D = f(0.5)
# Dado: B = D - 1

# Regra do ponto médio simples (n=1, ponto médio em x=0):
# I ≈ 2 * f(0) = 2C = 12 → C = 6

# Regra do ponto médio composta com n=2 (subintervalos: [-1,0] e [0,1])
# Pontos médios: x=-0.5 e x=0.5
# I ≈ 1*f(-0.5) + 1*f(0.5) = B + D = 5
# Como B = D - 1 → (D-1) + D = 5 → 2D -1 = 5 → 2D = 6 → D = 3, B = 2

# Regra de Simpson composta com n=2 (x: -1, 0, 1)
# I ≈ (2/6)[f(-1) + 4f(0) + f(1)] = (1/3)[A + 4*6 + A] = (1/3)[2A + 24] = 6
# 2A + 24 = 18 → 2A = -6 → A = -3

print("QUESTÃO 26")
print(f"f(-1) = f(1) = -3")
print(f"f(-0.5) = 2")
print(f"f(0) = 6")
print(f"f(0.5) = 3")
```
