**22.** Utilize a regra do ponto médio composta com \( n + 2 \) subintervalos para obter uma aproximação das integrais do Exercício 2.

**Exercício 2 (original):**
a) \(\int_{-0.25}^{0.25} (\cos x)^2 \, dx\)
b) \(\int_{-0.5}^{0} x \ln(x + 1) \, dx\)
c) \(\int_{0.75}^{1.3} [(\sin x)^2 - 2x \sin x + 1] \, dx\)
d) \(\int_{e}^{e+1} \frac{1}{x \ln x} \, dx\)

Para cada integral, use \( n = 4 \) como base (a menos que outro valor seja especificado).

```python
import numpy as np

def ponto_medio_composta(f, a, b, n):
    """Regra do ponto médio composta com n subintervalos"""
    h = (b - a) / n
    soma = 0
    for i in range(n):
        x_medio = a + (i + 0.5) * h
        soma += f(x_medio)
    return h * soma

print("=" * 70)
print("QUESTÃO 22 - Regra do ponto médio composta (n+2 subintervalos)")
print("=" * 70)

# Usando n = 4 como base
n_base = 4
n_medio = n_base + 2  # n + 2 subintervalos = 6

# 22a
f = lambda x: np.cos(x)**2
a, b = -0.25, 0.25
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"22a. ∫₋₀.₂₅⁰·²⁵ cos² x dx, n={n_medio} -> {resultado:.6f}")

# 22b
f = lambda x: x * np.log(x + 1)
a, b = -0.5, 0
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"22b. ∫₋₀.₅⁰ x ln(x+1) dx, n={n_medio} -> {resultado:.6f}")

# 22c
f = lambda x: np.sin(x)**2 - 2*x*np.sin(x) + 1
a, b = 0.75, 1.3
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"22c. ∫₀.₇₅¹·³ (sen² x - 2x sen x + 1) dx, n={n_medio} -> {resultado:.6f}")

# 22d - CORRIGIDO (usei np.e em vez de e)
f = lambda x: 1 / (x * np.log(x))
a, b = np.e, np.e + 1
resultado = ponto_medio_composta(f, a, b, n_medio)
print(f"22d. ∫_e^{np.e+1:.4f} 1/(x ln x) dx, n={n_medio} -> {resultado:.6f}")
```
