**29.** Determine os valores de \( n \) e \( h \) necessários para obter uma aproximação de  
\[
\int_{1}^{2} \frac{1}{x+4} \, dx
\]  
com precisão de \( 10^{-5} \) e calcule a aproximação. Use a:  
a. regra do trapézio composta  
b. regra de Simpson composta  
c. regra do ponto médio composta

```python
import numpy as np

def f(x):
    return 1 / (x + 4)

def f2linhas(x):
    return 2 / (x + 4)**3

def f4linhas(x):
    return 24 / (x + 4)**5

a, b = 1, 2
tolerancia = 1e-5

# Máximos das derivadas (função decrescente, máximo em x = a)
max_f2 = f2linhas(a)
max_f4 = f4linhas(a)


print("QUESTÃO 29 - ∫₁² 1/(x+4) dx, precisão = 1e-5\n")


# a) Trapézio composta
h_trap = np.sqrt((12 * tolerancia) / ((b - a) * max_f2))
n_trap = int(np.ceil((b - a) / h_trap))
h_trap = (b - a) / n_trap

x_trap = np.linspace(a, b, n_trap + 1)
integral_trap = (h_trap/2) * (f(x_trap[0]) + 2*np.sum(f(x_trap[1:n_trap])) + f(x_trap[n_trap]))

print(f"\na) Trapézio composta:")
print(f"   n = {n_trap}, h = {h_trap:.6f}")
print(f"   Aproximação = {integral_trap:.10f}")

# b) Simpson composta
h_simp = ((180 * tolerancia) / ((b - a) * max_f4)) ** 0.25
n_simp = int(np.ceil((b - a) / h_simp))
if n_simp % 2 != 0:
    n_simp += 1
h_simp = (b - a) / n_simp

x_simp = np.linspace(a, b, n_simp + 1)
integral_simp = (h_simp/3) * (f(x_simp[0]) + f(x_simp[n_simp]) + 
                               4*np.sum(f(x_simp[1:n_simp:2])) + 
                               2*np.sum(f(x_simp[2:n_simp-1:2])))

print(f"\nb) Simpson composta:")
print(f"   n = {n_simp}, h = {h_simp:.6f}")
print(f"   Aproximação = {integral_simp:.10f}")

# c) Ponto médio composta
h_pm = np.sqrt((24 * tolerancia) / ((b - a) * max_f2))
n_pm = int(np.ceil((b - a) / h_pm))
h_pm = (b - a) / n_pm

integral_pm = 0
for i in range(n_pm):
    xm = a + (i + 0.5) * h_pm
    integral_pm += f(xm)
integral_pm *= h_pm

print(f"\nc) Ponto médio composta:")
print(f"   n = {n_pm}, h = {h_pm:.6f}")
print(f"   Aproximação = {integral_pm:.10f}")

# Valor exato para referência
exato = np.log(6/5)
print(f"\nValor exato: ln(6/5) = {exato:.10f}")
```
