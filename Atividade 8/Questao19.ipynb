**19.** Use a regra de Simpson composta para aproximar as integrais do Exercício 17:

a. \(\int_{1}^{2} x \ln x \, dx, \quad n = 4\)

b. \(\int_{-2}^{2} x^3 e^x \, dx, \quad n = 4\)

c. \(\int_{0}^{\pi/2} \frac{x^2}{x^2 + 4} \, dx, \quad n = 6\)

d. \(\int_{0}^{\pi} x^2 \cos x \, dx, \quad n = 6\)

e. \(\int_{0}^{\pi} e^{2x} \sin 3x \, dx, \quad n = 8\)

f. \(\int_{1}^{3} \frac{x}{x^2 + 4} \, dx, \quad n = 8\)

g. \(\int_{3}^{5} \frac{1}{\sqrt{x^2 - 4}} \, dx, \quad n = 8\)

h. \(\int_{0}^{3\pi/8} \tan x \, dx, \quad n = 8\)

```python
import numpy as np

def simpson_composta(f, a, b, n):
    """Regra de Simpson composta (n deve ser par)"""
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    soma = f(x[0]) + f(x[n])
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(x[i])
        else:
            soma += 4 * f(x[i])
    return (h / 3) * soma


print("QUESTÃO 19 - Regra de Simpson Composta\n")


# 19a
f = lambda x: x * np.log(x)
print(f"19a. ∫₁² x ln x dx, n=4 -> {simpson_composta(f, 1, 2, 4):.6f}")

# 19b
f = lambda x: x**3 * np.exp(x)
print(f"19b. ∫₋₂² x³ eˣ dx, n=4 -> {simpson_composta(f, -2, 2, 4):.6f}")

# 19c
f = lambda x: x**2 / (x**2 + 4)
print(f"19c. ∫₀^{np.pi/2:.2f} x²/(x²+4) dx, n=6 -> {simpson_composta(f, 0, np.pi/2, 6):.6f}")

# 19d
f = lambda x: x**2 * np.cos(x)
print(f"19d. ∫₀^π x² cos x dx, n=6 -> {simpson_composta(f, 0, np.pi, 6):.6f}")

# 19e
f = lambda x: np.exp(2*x) * np.sin(3*x)
print(f"19e. ∫₀^π e²ˣ sen(3x) dx, n=8 -> {simpson_composta(f, 0, np.pi, 8):.6f}")

# 19f
f = lambda x: x / (x**2 + 4)
print(f"19f. ∫₁³ x/(x²+4) dx, n=8 -> {simpson_composta(f, 1, 3, 8):.6f}")

# 19g
f = lambda x: 1 / np.sqrt(x**2 - 4)
print(f"19g. ∫₃⁵ 1/√(x²-4) dx, n=8 -> {simpson_composta(f, 3, 5, 8):.6f}")

# 19h
f = lambda x: np.tan(x)
print(f"19h. ∫₀^{3*np.pi/8:.3f} tan x dx, n=8 -> {simpson_composta(f, 0, 3*np.pi/8, 8):.6f}")
```
