**18a.** Aproxime \(\int_{-0.5}^{0.5} \cos^2 x \, dx\) usando a regra do trapézio composta com \(n = 4\).

```python
import numpy as np

# 18a
def f18a(x):
    return np.cos(x) ** 2

a, b, n = -0.5, 0.5, 4
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_18a = (h / 2) * (f18a(x_vals[0]) + 2 * np.sum(f18a(x_vals[1:n])) + f18a(x_vals[n]))
print(f"18a. ∫₋₀.₅⁰·⁵ cos² x dx, n=4 -> {integral_18a:.6f}")

# 18b
def f18b(x):
    return x * np.log(x + 1)

a, b, n = -0.5, 0.5, 6
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_18b = (h / 2) * (f18b(x_vals[0]) + 2 * np.sum(f18b(x_vals[1:n])) + f18b(x_vals[n]))
print(f"18b. ∫₋₀.₅⁰·⁵ x ln(x+1) dx, n=6 -> {integral_18b:.6f}")

# 18c
def f18c(x):
    return np.sin(x) ** 2 - 2 * x * np.sin(x) + 1

a, b, n = 0.75, 1.75, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_18c = (h / 2) * (f18c(x_vals[0]) + 2 * np.sum(f18c(x_vals[1:n])) + f18c(x_vals[n]))
print(f"18c. ∫₀.₇₅¹·⁷⁵ (sen² x - 2x sen x + 1) dx, n=8 -> {integral_18c:.6f}")

# 18d 
def f18d(x):
    return 1 / (x * np.log(x))

a, b, n = np.e, np.e + 2, 8
h = (b - a) / n
x_vals = np.linspace(a, b, n + 1)
integral_18d = (h / 2) * (f18d(x_vals[0]) + 2 * np.sum(f18d(x_vals[1:n])) + f18d(x_vals[n]))
print(f"18d. ∫_e^{np.e+2:.4f} 1/(x ln x) dx, n=8 -> {integral_18d:.6f}")
```
