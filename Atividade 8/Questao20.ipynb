20. Use a regra de Simpson composta para aproximar as integrais do Exercício 18:

a. \(\int_{-0.5}^{0.5} \cos^2 x \, dx, \quad n = 4\)

b. \(\int_{-0.5}^{0.5} x \ln(x + 1) \, dx, \quad n = 6\)

c. \(\int_{0.75}^{1.75} (\sin^2 x - 2x \sin x + 1) \, dx, \quad n = 8\)

d. \(\int_{e}^{e+2} \frac{1}{x \ln x} \, dx, \quad n = 8\)

```python
import numpy as np

def simpson_composta(f, a, b, n):
    if n % 2 != 0:
        raise ValueError("n deve ser par para a regra de Simpson composta")
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    soma = f(x[0]) + f(x[n])
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(x[i])
        else:
            soma += 4 * f(x[i])
    return (h / 3) * soma


# 20a
f = lambda x: np.cos(x)**2
print(f"20a. ∫₋₀.₅⁰·⁵ cos² x dx, n=4 -> {simpson_composta(f, -0.5, 0.5, 4):.6f}")

# 20b
f = lambda x: x * np.log(x + 1)
print(f"20b. ∫₋₀.₅⁰·⁵ x ln(x+1) dx, n=6 -> {simpson_composta(f, -0.5, 0.5, 6):.6f}")

# 20c
f = lambda x: np.sin(x)**2 - 2*x*np.sin(x) + 1
print(f"20c. ∫₀.₇₅¹·⁷⁵ (sen² x - 2x sen x + 1) dx, n=8 -> {simpson_composta(f, 0.75, 1.75, 8):.6f}")

# 20d
f = lambda x: 1 / (x * np.log(x))
print(f"20d. ∫_e^{np.e+2:.4f} 1/(x ln x) dx, n=8 -> {simpson_composta(f, np.e, np.e + 2, 8):.6f}")
```
