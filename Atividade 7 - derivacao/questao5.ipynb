# Questão 5

Use a fórmula dos três pontos mais precisa para completar as seguintes tabelas:

## a)

| x     | f(x)       | f'(x) |
|-------|------------|-------|
| -0,3  | -0,27652   |       |
| -0,2  | -0,25074   |       |
| -0,1  | -0,16134   |       |
| 0     | 0          |       |

## b)

| x     | f(x)       | f'(x) |
|-------|------------|-------|
| 7,4   | -68,3193   |       |
| 7,6   | -71,6982   |       |
| 7,8   | -75,1576   |       |
| 8,0   | -78,6974   |       |

## c)

| x     | f(x)       | f'(x) |
|-------|------------|-------|
| 1,1   | 1,52918    |       |
| 1,2   | 1,64024    |       |
| 1,3   | 1,70470    |       |
| 1,4   | 1,71277    |       |

## d)

| x     | f(x)       | f'(x) |
|-------|------------|-------|
| -2,7  | 0,054797   |       |
| -2,5  | 0,11342    |       |
| -2,3  | 0,65536    |       |
| -2,1  | 0,98472    |       |

```python

# Método: fórmula central f'(x) = (f(x+h) - f(x-h)) / (2h)

print("RESOLUÇÃO DA QUESTÃO 5")

# 5a) Tabela: x = -0,3 / -0,2 / -0,1 / 0,0

print("\n--- 5a ---")
print("Dados: x = [-0.3, -0.2, -0.1, 0.0]")
print("       f = [-0.27652, -0.25074, -0.16134, 0.0]")
print("Passo h = 0.1\n")

# Calculando derivadas nos pontos internos
# Para x = -0,2
deriv1 = (-0.16134 - (-0.27652)) / (2 * 0.1)
print("f'(-0,2) = [f(-0,1) - f(-0,3)] / (2*0,1)")
print("         = [-0,16134 - (-0,27652)] / 0,2")
print(f"         = {deriv1:.6f}\n")

# Para x = -0,1
deriv2 = (0.0 - (-0.25074)) / (2 * 0.1)
print("f'(-0,1) = [f(0) - f(-0,2)] / (2*0,1)")
print("         = [0 - (-0,25074)] / 0,2")
print(f"         = {deriv2:.6f}")

print("\nTabela completa (5a):")
print("   x    |    f(x)     |    f'(x)")
print("-" * 40)
print(" -0,3   |  -0,27652   |    _____")
print(f" -0,2   |  -0,25074   |  {deriv1:.6f}")
print(f" -0,1   |  -0,16134   |  {deriv2:.6f}")
print("  0,0   |   0,00000   |    _____")


# 5b) Tabela: x = 7,4 / 7,6 / 7,8 / 8,0

print("\n--- 5b ---")
print("Dados: x = [7.4, 7.6, 7.8, 8.0]")
print("       f = [-68.3193, -71.6982, -75.1576, -78.6974]")
print("Passo h = 0.2\n")

# Para x = 7,6
deriv3 = (-75.1576 - (-68.3193)) / (2 * 0.2)
print("f'(7,6) = [f(7,8) - f(7,4)] / (2*0,2)")
print("        = [-75,1576 - (-68,3193)] / 0,4")
print(f"        = {deriv3:.6f}\n")

# Para x = 7,8
deriv4 = (-78.6974 - (-71.6982)) / (2 * 0.2)
print("f'(7,8) = [f(8,0) - f(7,6)] / (2*0,2)")
print("        = [-78,6974 - (-71,6982)] / 0,4")
print(f"        = {deriv4:.6f}")

print("\nTabela completa (5b):")
print("   x    |    f(x)     |    f'(x)")
print("-" * 40)
print("  7,4   |  -68,3193   |    _____")
print(f"  7,6   |  -71,6982   |  {deriv3:.6f}")
print(f"  7,8   |  -75,1576   |  {deriv4:.6f}")
print("  8,0   |  -78,6974   |    _____")


# 5c) Tabela: x = 1,1 / 1,2 / 1,3 / 1,4

print("\n--- 5c ---")
print("Dados: x = [1.1, 1.2, 1.3, 1.4]")
print("       f = [1.52918, 1.64024, 1.70470, 1.71277]")
print("Passo h = 0.1\n")

# Para x = 1,2
deriv5 = (1.70470 - 1.52918) / (2 * 0.1)
print("f'(1,2) = [f(1,3) - f(1,1)] / (2*0,1)")
print("        = [1,70470 - 1,52918] / 0,2")
print(f"        = {deriv5:.6f}\n")

# Para x = 1,3
deriv6 = (1.71277 - 1.64024) / (2 * 0.1)
print("f'(1,3) = [f(1,4) - f(1,2)] / (2*0,1)")
print("        = [1,71277 - 1,64024] / 0,2")
print(f"        = {deriv6:.6f}")

print("\nTabela completa (5c):")
print("   x    |    f(x)     |    f'(x)")
print("-" * 40)
print("  1,1   |   1,52918   |    _____")
print(f"  1,2   |   1,64024   |  {deriv5:.6f}")
print(f"  1,3   |   1,70470   |  {deriv6:.6f}")
print("  1,4   |   1,71277   |    _____")

# 5d) Tabela: x = -2,7 / -2,5 / -2,3 / -2,1

print("\n--- 5d ---")
print("Dados: x = [-2.7, -2.5, -2.3, -2.1]")
print("       f = [0.054797, 0.11342, 0.65536, 0.98472]")
print("Passo h = 0.2\n")

# Para x = -2,5
deriv7 = (0.65536 - 0.054797) / (2 * 0.2)
print("f'(-2,5) = [f(-2,3) - f(-2,7)] / (2*0,2)")
print("         = [0,65536 - 0,054797] / 0,4")
print(f"         = {deriv7:.6f}\n")

# Para x = -2,3
deriv8 = (0.98472 - 0.11342) / (2 * 0.2)
print("f'(-2,3) = [f(-2,1) - f(-2,5)] / (2*0,2)")
print("         = [0,98472 - 0,11342] / 0,4")
print(f"         = {deriv8:.6f}")

print("\nTabela completa (5d):")
print("   x    |    f(x)     |    f'(x)")
print("-" * 40)
print(" -2,7   |  0,054797   |    _____")
print(f" -2,5   |  0,113420   |  {deriv7:.6f}")
print(f" -2,3   |  0,655360   |  {deriv8:.6f}")
print(" -2,1   |  0,984720   |    _____")


```
