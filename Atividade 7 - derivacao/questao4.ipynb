# 4. Use a fórmula de três pontos mais precisa
# para completar as tabelas:

# (c)
# x     f(x)        f'(x)
# 2.9   -4.827866
# 3.0   -4.240058
# 3.1   -3.496909
# 3.2   -2.596792

# (d)
# x     f(x)        f'(x)
# 2.0   3.6887983
# 2.1   3.6905701
# 2.2   3.6688192
# 2.3   3.6245909

```python
import numpy as np

def derivada_3pontos_q4(x, f):
    n = len(x)
    df = np.zeros(n)

    i = 0
    while i < n:
        if i == 0:
            h = x[1] - x[0]
            df[i] = (-3*f[i] + 4*f[i+1] - f[i+2]) / (2*h)
        elif i == n-1:
            h = x[i] - x[i-1]
            df[i] = (3*f[i] - 4*f[i-1] + f[i-2]) / (2*h)
        else:
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i-1]) / (2*h)
        i += 1

    return df

# Dados
x_c = np.array([2.9, 3.0, 3.1, 3.2])
f_c = np.array([-4.827866, -4.240058, -3.496909, -2.596792])

x_d = np.array([2.0, 2.1, 2.2, 2.3])
f_d = np.array([3.6887983, 3.6905701, 3.6688192, 3.6245909])

print("Q4 (c):", derivada_3pontos_q4(x_c, f_c))
print("Q4 (d):", derivada_3pontos_q4(x_d, f_d))
```
