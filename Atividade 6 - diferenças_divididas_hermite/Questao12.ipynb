# Questão 12

A tabela a seguir fornece dados para a função descrita por \( f(x) = e^{0,1x^2} \). Obtenha uma aproximação de \( f(1,25) \) usando \( H_5(1,25) \) e \( H_3(1,25) \), onde \( H_5 \) usa os nós \( x_0 = 1 \), \( x_1 = 2 \) e \( x_2 = 3 \), e \( H_3 \) usa os nós \( x_0 = 1 \) e \( x_1 = 1,5 \). Encontre os limitantes de erro para essas aproximações.

| x   | f(x)       | f'(x)       |
|-----|------------|-------------|
| 1,0 | 1,105170918 | 0,2210341836 |
| 1,5 | 1,253232716 | 0,3756968148 |
| 2,0 | 1,491824698 | 0,5967298792 |
| 3,0 | 2,459603111 | 1,475761867 |

```python
import numpy as np
import math

# HERMITE

def hermite(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    for i in range(n):
        Q[2*i][0] = f[i]
        Q[2*i+1][0] = f[i]

    for i in range(n):
        Q[2*i][1] = df[i]
        Q[2*i+1][1] = df[i]

    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = (Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1])

    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q

# AVALIAR

def avaliar(z, Q, x):
    n = len(z)
    resultado = Q[0][0]
    termo = 1.0

    for i in range(1, n):
        termo *= (x - z[i-1])
        resultado += Q[i][i] * termo

    return resultado

# FUNÇÃO

f = lambda x: np.exp(0.1 * x**2)
df = lambda x: 0.2 * x * np.exp(0.1 * x**2)

# derivada de ordem 4 (aproximada)
d4f = lambda x: np.exp(0.1*x**2)*(0.12 + 0.48*x**2 + 0.16*x**4)

# DADOS

x_nodes = [1.0, 1.5]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 1.25

aprox = avaliar(z, Q, x_aprox)
real = f(x_aprox)

erro_real = abs(aprox - real)

# ERRO TEÓRICO

xs = np.linspace(1.0, 1.5, 100)
max_deriv = max(abs(d4f(x)) for x in xs)

prod = 1
for xi in x_nodes:
    prod *= (x_aprox - xi)**2

erro_teorico = (max_deriv / math.factorial(4)) * abs(prod)

# RESULTADOS

print("\nAproximação:", aprox)
print("Valor real:", real)

print("\nErro real:", erro_real)
print("Erro teórico (máximo):", erro_teorico)

if erro_real <= erro_teorico:
    print("\n✔ Erro dentro do limite teórico")
else:
    print("\n⚠ Erro maior que o esperado")
```
