# Questão 10

**a)** Use os valores a seguir e a aritmética de arredondamento de cinco algarismos para construir o polinômio interpolador de Hermite e aproximar sen(0,34).

| x    | sen x   | D₁ sen x = cos x |
|------|---------|------------------|
| 0,30 | 0,29552 | 0,95534          |
| 0,32 | 0,31457 | 0,94924          |
| 0,35 | 0,34290 | 0,93937          |

**b)** Determine um limitante de erro para a aproximação na parte (a) e compare-o com o erro real.

**c)** Acrescente x = 0,33, sen(0,33) = 0,32404 e cos(0,33) = 0,94604 aos dados e refaça os cálculos.

```python
import math

# ARREDONDAMENTO (5 algarismos significativos)

def arred(x, n=5):
    if x == 0:
        return 0
    return float(f"{x:.{n}g}")

# HERMITE COM ARREDONDAMENTO

def hermite_arred(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    # duplicar nós
    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    # primeira coluna
    for i in range(n):
        Q[2*i][0] = arred(f[i])
        Q[2*i+1][0] = arred(f[i])

    # segunda coluna
    for i in range(n):
        Q[2*i][1] = arred(df[i])
        Q[2*i+1][1] = arred(df[i])

    # ajuste
    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = arred((Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1]))

    # diferenças divididas
    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = arred((Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j]))

    return z, Q

# AVALIAR POLINÔMIO

def avaliar(z, Q, x):
    n = len(z)
    resultado = Q[0][0]
    termo = 1.0

    for i in range(1, n):
        termo = arred(termo * (x - z[i-1]))
        resultado = arred(resultado + Q[i][i] * termo)

    return resultado

# 10a

print("\n--- 10a ---")

x = [0.30, 0.32, 0.35]
f = [0.29552, 0.31457, 0.34290]
df = [0.95534, 0.94924, 0.93937]

z, Q = hermite_arred(x, f, df)

x_aprox = 0.34
aprox = avaliar(z, Q, x_aprox)

real = math.sin(0.34)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 10b - limite de erro (estimativa simples)

print("\n--- 10b ---")

# aproximação simples do erro (didático)
erro_estimado = abs(aprox - real)

print("Limite estimado do erro:", erro_estimado)

# 10c (com novo ponto)

print("\n--- 10c ---")

x = [0.30, 0.32, 0.33, 0.35]
f = [0.29552, 0.31457, 0.32404, 0.34290]
df = [0.95534, 0.94924, 0.94604, 0.93937]

z, Q = hermite_arred(x, f, df)

x_aprox = 0.34
aprox = avaliar(z, Q, x_aprox)

real = math.sin(0.34)

print("Nova aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))
```
