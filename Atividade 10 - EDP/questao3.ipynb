# Questão 3

Obtenha aproximações para as soluções das seguintes EDPs utilizando o **Algoritmo 12.1**.

## (a)

$$
\frac{\partial^2u}{\partial x^2}
+
\frac{\partial^2u}{\partial y^2}
=
0,
\qquad
0<x<1,\;
0<y<1.
$$

Condições de contorno

$$
u(x,0)=0,
$$

$$
u(x,1)=x,
$$

$$
u(0,y)=0,
$$

$$
u(1,y)=y.
$$

Utilize

$$
h=k=0.2
$$

e compare os resultados com

$$
u(x,y)=xy.
$$

---

## (b)

$$
\frac{\partial^2u}{\partial x^2}
+
\frac{\partial^2u}{\partial y^2}
=
-\left[\cos(x+y)+\cos(x-y)\right],
$$

$$
0<x<\pi,
\qquad
0<y<\frac{\pi}{2}.
$$

Condições de contorno

$$
u(0,y)=\cos y,
$$

$$
u(\pi,y)=-\cos y,
$$

$$
u(x,0)=\cos x,
$$

$$
u\left(x,\frac{\pi}{2}\right)=0.
$$

Utilize

$$
h=\frac{\pi}{5},
\qquad
k=\frac{\pi}{10}.
$$

Compare com

$$
u(x,y)=\cos x\cos y.
$$

---

## (c)

$$
\frac{\partial^2u}{\partial x^2}
+
\frac{\partial^2u}{\partial y^2}
=
(x^2+y^2)e^{xy},
$$

$$
0<x<2,
\qquad
0<y<1.
$$

Condições de contorno

$$
u(0,y)=1,
$$

$$
u(2,y)=e^{2y},
$$

$$
u(x,0)=1,
$$

$$
u(x,1)=e^x.
$$

Utilize

$$
h=0.2,
\qquad
k=0.1.
$$

Compare com

$$
u(x,y)=e^{xy}.
$$

---

## (d)

$$
\frac{\partial^2u}{\partial x^2}
+
\frac{\partial^2u}{\partial y^2}
=
\frac{x}{y}
+
\frac{y}{x},
$$

$$
1<x<2,
\qquad
1<y<2.
$$

Condições de contorno

$$
u(x,1)=x\ln x,
$$

$$
u(x,2)=x\ln(4x^2),
$$

$$
u(1,y)=y\ln y,
$$

$$
u(2,y)=2y\ln(2y).
$$

Utilize

$$
h=k=0.1.
$$

Compare com

$$
u(x,y)=xy\ln(xy).
$$

```python
import numpy as np

# ESCOLHA A ALTERNATIVA

alternativa = "a"      # Troque para "b", "c" ou "d"

# DEFINIÇÃO DO PROBLEMA

if alternativa == "a":

    a, b = 0, 1
    c, d = 0, 1

    h = 0.2
    k = 0.2

    def f(x,y):
        return 0

    def exata(x,y):
        return x*y

    def inferior(x):
        return 0

    def superior(x):
        return x

    def esquerda(y):
        return 0

    def direita(y):
        return y


elif alternativa == "b":

    a, b = 0, np.pi
    c, d = 0, np.pi/2

    h = np.pi/5
    k = np.pi/10

    def f(x,y):
        return -(np.cos(x+y)+np.cos(x-y))

    def exata(x,y):
        return np.cos(x)*np.cos(y)

    def inferior(x):
        return np.cos(x)

    def superior(x):
        return 0

    def esquerda(y):
        return np.cos(y)

    def direita(y):
        return -np.cos(y)


elif alternativa == "c":

    a, b = 0, 2
    c, d = 0, 1

    h = 0.2
    k = 0.1

    def f(x,y):
        return (x**2+y**2)*np.exp(x*y)

    def exata(x,y):
        return np.exp(x*y)

    def inferior(x):
        return 1

    def superior(x):
        return np.exp(x)

    def esquerda(y):
        return 1

    def direita(y):
        return np.exp(2*y)


elif alternativa == "d":

    a, b = 1, 2
    c, d = 1, 2

    h = 0.1
    k = 0.1

    def f(x,y):
        return x/y + y/x

    def exata(x,y):
        return x*y*np.log(x*y)

    def inferior(x):
        return x*np.log(x)

    def superior(x):
        return x*np.log(4*x*x)

    def esquerda(y):
        return y*np.log(y)

    def direita(y):
        return 2*y*np.log(2*y)

else:
    raise ValueError("Alternativa inválida.")

# MALHA

nx = int(round((b-a)/h))
ny = int(round((d-c)/k))

x = np.linspace(a,b,nx+1)
y = np.linspace(c,d,ny+1)

u = np.zeros((ny+1,nx+1))

# CONDIÇÕES DE CONTORNO

for i in range(nx+1):
    u[0,i] = inferior(x[i])
    u[ny,i] = superior(x[i])

for j in range(ny+1):
    u[j,0] = esquerda(y[j])
    u[j,nx] = direita(y[j])

# MÉTODO DE GAUSS-SEIDEL

tol = 1e-8
max_iter = 10000

for it in range(max_iter):

    erro = 0

    for j in range(1,ny):

        for i in range(1,nx):

            novo = (
                k**2*(u[j,i-1] + u[j,i+1])
                + h**2*(u[j-1,i] + u[j+1,i])
                - h**2*k**2*f(x[i],y[j])
            )/(2*(h**2+k**2))

            erro = max(erro, abs(novo-u[j,i]))

            u[j,i] = novo

    if erro < tol:
        print("Convergiu em", it+1, "iterações.")
        break

else:
    print("Número máximo de iterações atingido.")

# SOLUÇÃO EXATA

ue = np.zeros_like(u)

for j in range(ny+1):
    for i in range(nx+1):
        ue[j,i] = exata(x[i],y[j])

# RESULTADOS

print("\nSolução aproximada:\n")
print(u)

print("\nSolução exata:\n")
print(ue)

erro = np.abs(u-ue)

print("\nErro absoluto:\n")
print(erro)

print("\nErro máximo =", np.max(erro))
```
