## Questão 11

Seja a função:

\[
f(x) = 3e^x - e^{2x}
\]

**(a)** Faça a aproximação de \( f(1{,}03) \) pelo polinômio interpolador de Hermite de grau no máximo três, usando \( x_0 = 1 \) e \( x_1 = 1{,}05 \). Compare o erro real com o limitante de erro.

**(b)** Repita o item (a) com o polinômio interpolador de Hermite de grau no máximo cinco, usando \( x_0 = 1 \), \( x_1 = 1{,}05 \) e \( x_2 = 1{,}07 \).

```python
import numpy as np
import math

# HERMITE

def hermite(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    for i in range(n):
        Q[2*i][0] = f[i]
        Q[2*i+1][0] = f[i]

    for i in range(n):
        Q[2*i][1] = df[i]
        Q[2*i+1][1] = df[i]

    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = (Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1])

    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q


def avaliar(z, Q, x):
    n = len(z)
    resultado = Q[0][0]
    termo = 1.0

    for i in range(1, n):
        termo *= (x - z[i-1])
        resultado += Q[i][i] * termo

    return resultado

# FUNÇÃO DO PROBLEMA

f = lambda x: 3*np.exp(x) - np.exp(2*x)
df = lambda x: 3*np.exp(x) - 2*np.exp(2*x)

# derivada de ordem 4 (porque n=2 → 2n=4)
d4f = lambda x: 3*np.exp(x) - 16*np.exp(2*x)


# DADOS

x_nodes = [1.0, 1.05]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 1.03

aprox = avaliar(z, Q, x_aprox)
real = f(x_aprox)

erro_real = abs(aprox - real)

# ERRO TEÓRICO

# estimativa do máximo da derivada no intervalo
xs = np.linspace(1.0, 1.05, 100)
max_deriv = max(abs(d4f(x)) for x in xs)

# produto (x - xi)^2
prod = 1
for xi in x_nodes:
    prod *= (x_aprox - xi)**2

erro_teorico = (max_deriv / math.factorial(4)) * abs(prod)

# RESULTADOS

print("\nAproximação:", aprox)
print("Valor real:", real)

print("\nErro real:", erro_real)
print("Erro teórico (máximo):", erro_teorico)

print("\nComparação:")
if erro_real <= erro_teorico:
    print("✔ Erro real está dentro do limite teórico")
else:
    print("⚠ Algo inesperado aconteceu")
```
