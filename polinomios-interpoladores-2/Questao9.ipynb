```python
# Questão 9

Os dados do Exercício 7 foram gerados usando as funções a seguir. Use os polinômios construídos no Exercício 7 para o valor de x para obter uma aproximação de f(x) e calcular o erro absoluto.

**a)** f(x) = e^{2x}; aproximar f(0,43).

**b)** f(x) = x⁴ − x³ + x² − x + 1; aproximar f(0).

**c)** f(x) = x² cos x − 3x; aproximar f(0,18).

**d)** f(x) = ln(eˣ + 2); aproximar f(0,25).
```

```python
import numpy as np

def hermite(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    for i in range(n):
        Q[2*i][0] = f[i]
        Q[2*i+1][0] = f[i]

    for i in range(n):
        Q[2*i][1] = df[i]
        Q[2*i+1][1] = df[i]

    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = (Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1])

    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q

# AVALIAR POLINÔMIO

def avaliar_hermite(z, Q, x_val):
    n = len(z)
    resultado = Q[0][0]
    termo = 1.0

    for i in range(1, n):
        termo *= (x_val - z[i-1])
        resultado += Q[i][i] * termo

    return resultado


print("QUESTÃO 9 - APROXIMAÇÃO COM HERMITE")

# 9a) f(x) = e^(2x), aproximar f(0.43)
# ============================================================

print("\n--- 9a ---")

f = lambda x: np.exp(2*x)
df = lambda x: 2*np.exp(2*x)

x_nodes = [0.4, 0.5]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.43
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 9b) f(x) = x^4 - x^3 + x^2 - x + 1, aproximar f(0)

print("\n--- 9b ---")

f = lambda x: x**4 - x**3 + x**2 - x + 1
df = lambda x: 4*x**3 - 3*x**2 + 2*x - 1

x_nodes = [-0.5, 0.5]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.0
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 9c) f(x) = x^2 cos x - 3x, aproximar f(0.18)

print("\n--- 9c ---")

f = lambda x: x**2 * np.cos(x) - 3*x
df = lambda x: 2*x*np.cos(x) - x**2*np.sin(x) - 3

x_nodes = [0.1, 0.2]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.18
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 9d) f(x) = ln(e^x + 2), aproximar f(0.25)

print("\n--- 9d ---")

f = lambda x: np.log(np.exp(x) + 2)
df = lambda x: (np.exp(x)) / (np.exp(x) + 2)

x_nodes = [0.2, 0.3]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.25
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))
```
