# Questão 7

Use o Teorema 3.9 ou o Algoritmo 3.3 para construir um polinômio aproximador para os seguintes dados:

**a)**

| x   | f(x)    | f'(x)   |
|-----|---------|---------|
| 0   | 1,00000 | 2,00000 |
| 0,5 | 2,71828 | 5,43656 |

**b)**

| x     | f(x)     | f'(x)     |
|-------|----------|-----------|
| -0,25 | 1,33203  | 0,437500  |
| 0,25  | 0,800781 | -0,625000 |

**c)**

| x   | f(x)         | f'(x)      |
|-----|--------------|------------|
| 0,1 | -0,29004996  | -2,8019975 |
| 0,2 | -0,56079734  | -2,6159201 |
| 0,3 | -0,81401972  | -2,9734038 |

**d)**

| x    | f(x)       | f'(x)      |
|------|------------|------------|
| -1,0 | 0,86199480 | 0,15536240 |
| -0,5 | 0,95802009 | 0,23269654 |
| 0    | 1,0986123  | 0,33333333 |
| 0,5  | 1,2943767  | 0,45186776 |

```python


def hermite(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    # Duplica nós
    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    # Primeira coluna
    for i in range(n):
        Q[2*i][0] = f[i]
        Q[2*i+1][0] = f[i]

    # Segunda coluna
    for i in range(n):
        Q[2*i][1] = df[i]
        Q[2*i+1][1] = df[i]

    # Ajuste quando pontos são diferentes
    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = (Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1])

    # Diferenças divididas
    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q


def print_polinomio(z, Q):
    n = len(z)
    print("Coeficientes (diagonal):")
    for i in range(n):
        print(f"a{i} = {Q[i][i]}")

    print("\nForma de Newton:")
    print("P(x) =", Q[0][0], end="")

    for i in range(1, n):
        print(f" + ({Q[i][i]})", end="")
        for j in range(i):
            print(f"(x - {z[j]})", end="")
    print("\n")


print("\n--- 7a ---")

x = [0.0, 0.5]
f = [1.00000, 2.71828]
df = [2.00000, 5.43656]

z, Q = hermite(x, f, df)
print_polinomio(z, Q)

print("\n--- 7b ---")

x = [-0.25, 0.25]
f = [1.33203, 0.800781]
df = [0.437500, -0.625000]

z, Q = hermite(x, f, df)
print_polinomio(z, Q)


print("\n--- 7c ---")

x = [0.1, 0.2, 0.3]
f = [-0.29004996, -0.56079734, -0.81401972]
df = [-2.8019975, -2.6159201, -2.9734038]

z, Q = hermite(x, f, df)
print_polinomio(z, Q)


print("\n--- 7d ---")

x = [-1.0, -0.5, 0.0, 0.5]
f = [0.86199480, 0.95802009, 1.0986123, 1.2943767]
df = [0.15536240, 0.23269654, 0.33333333, 0.45186776]

z, Q = hermite(x, f, df)
print_polinomio(z, Q)
```
