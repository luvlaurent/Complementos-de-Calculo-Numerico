# Questão 8

Os dados do Exercício 6 foram gerados usando as funções a seguir. Use os polinômios construídos no Exercício 6 para o valor determinado de x para obter uma aproximação de f(x) e calcular o erro absoluto.

**a)** f(x) = x ln x; aproximar f(8,4).

**b)** f(x) = sen(eˣ − 2); aproximar f(0,9).

**c)** f(x) = x³ + 4,002x + 1,101; aproximar f(-1/3).

**d)** f(x) = x cos x − 2x² + 3x − 1; aproximar f(0,25).

```python
import numpy as np

def hermite(x, f, df):
    n = len(x)
    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    # Duplica nós
    for i in range(n):
        z.append(x[i])
        z.append(x[i])

    # Primeira coluna
    for i in range(n):
        Q[2*i][0] = f[i]
        Q[2*i+1][0] = f[i]

    # Segunda coluna
    for i in range(n):
        Q[2*i][1] = df[i]
        Q[2*i+1][1] = df[i]

    # Ajuste quando pontos são diferentes
    for i in range(1, 2*n):
        if z[i] != z[i-1]:
            Q[i][1] = (Q[i][0] - Q[i-1][0]) / (z[i] - z[i-1])

    # Diferenças divididas
    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q

# AVALIAÇÃO DO POLINÔMIO

def avaliar_hermite(z, Q, x_val):
    n = len(z)
    resultado = Q[0][0]
    termo = 1.0

    for i in range(1, n):
        termo *= (x_val - z[i-1])
        resultado += Q[i][i] * termo

    return resultado

print("QUESTÃO 8 - APROXIMAÇÃO COM HERMITE")

# 8a) f(x) = x ln x, aproximar f(8.4)

print("\n--- 8a ---")

f = lambda x: x*np.log(x)
df = lambda x: np.log(x) + 1

x_nodes = [7, 8, 9]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 8.4
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 8b) f(x) = sen(e^x - 2), aproximar f(0.9)

print("\n--- 8b ---")

f = lambda x: np.sin(np.exp(x) - 2)
df = lambda x: np.cos(np.exp(x) - 2) * np.exp(x)

x_nodes = [0.8, 1.0]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.9
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 8c) f(x) = x³ + 4.002x + 1.101, aproximar f(-1/3)

print("\n--- 8c ---")

f = lambda x: x**3 + 4.002*x + 1.101
df = lambda x: 3*x**2 + 4.002

x_nodes = [-1, 0]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = -1/3
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))

# 8d) f(x) = x cos x - 2x² + 3x - 1, aproximar f(0.25)

print("\n--- 8d ---")

f = lambda x: x*np.cos(x) - 2*x**2 + 3*x - 1
df = lambda x: np.cos(x) - x*np.sin(x) - 4*x + 3

x_nodes = [0.2, 0.3]
f_vals = [f(x) for x in x_nodes]
df_vals = [df(x) for x in x_nodes]

z, Q = hermite(x_nodes, f_vals, df_vals)

x_aprox = 0.25
aprox = avaliar_hermite(z, Q, x_aprox)
real = f(x_aprox)

print("Aproximação:", aprox)
print("Valor real:", real)
print("Erro absoluto:", abs(aprox - real))
```
