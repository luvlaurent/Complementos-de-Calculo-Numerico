Use o Teorema 3.9 ou o Algoritmo 3.3 para construir um polinômio aproximador para os seguintes dados:

**a)**

| x   | f(x)     | f'(x)   |
|-----|----------|---------|
| 8,3 | 17,56492 | 3,116256 |
| 8,6 | 18,50515 | 3,151762 |

**b)**

| x   | f(x)      | f'(x)    |
|-----|-----------|----------|
| 0,8 | 0,22363362 | 2,1691753 |
| 1,0 | 0,65809197 | 2,0466965 |

**c)**

| x    | f(x)        | f'(x)    |
|------|-------------|----------|
| -0,5 | -0,02475000 | 0,7510000 |
| -0,25| 0,33493750  | 2,1890000 |
| 0    | 1,10100000  | 4,0020000 |

**d)**

| x   | f(x)         | f'(x)     |
|-----|--------------|-----------|
| 0,1 | -0,62049958  | 3,58502082 |
| 0,2 | -0,28398668  | 3,14033271 |
| 0,3 | 0,00660095   | 2,66668043 |
| 0,4 | 0,24842440   | 2,16529366 |

```python
import numpy as np

def hermite(x, y, dy):
    n = len(x)
    z = np.zeros(2*n)
    Q = np.zeros((2*n, 2*n))

    for i in range(n):
        z[2*i] = z[2*i+1] = x[i]
        Q[2*i][0] = Q[2*i+1][0] = y[i]
        Q[2*i+1][1] = dy[i]

        if i != 0:
            Q[2*i][1] = (Q[2*i][0] - Q[2*i-1][0]) / (z[2*i] - z[2*i-1])

    for i in range(2, 2*n):
        for j in range(2, i+1):
            Q[i][j] = (Q[i][j-1] - Q[i-1][j-1]) / (z[i] - z[i-j])

    return z, Q


def avalia_hermite(z, Q, x):
    n = len(z)
    resultado = Q[0][0]
    produto = 1

    for i in range(1, n):
        produto *= (x - z[i-1])
        resultado += Q[0][i] * produto

    return resultado


# Exemplo (6a)
x = np.array([8.3, 8.6])
y = np.array([17.56492, 18.50515])
dy = np.array([3.116256, 3.151762])

z, Q = hermite(x, y, dy)

valor = 8.4
p = avalia_hermite(z, Q, valor)

print("Hermite:", p)
```
