# 2. Use as fórmulas de diferenças progressivas e regressivas
# para completar as tabelas:

# (a)
# x      f(x)     f'(x)
# -0.3   1.9507
# -0.2   2.0421
# -0.1   2.0601

# (b)
# x     f(x)     f'(x)
# 1.0   1.0000
# 1.2   1.2625
# 1.4   1.6595

```python
import numpy as np

def derivadas_q2(x, f):
    n = len(x)
    df = np.zeros(n)

    i = 0
    while i < n:
        if i == 0:
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i]) / h
        elif i == n-1:
            h = x[i] - x[i-1]
            df[i] = (f[i] - f[i-1]) / h
        else:
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i-1]) / (2*h)
        i += 1

    return df

# Dados
x_a = np.array([-0.3, -0.2, -0.1])
f_a = np.array([1.9507, 2.0421, 2.0601])

x_b = np.array([1.0, 1.2, 1.4])
f_b = np.array([1.0000, 1.2625, 1.6595])

print("Q2 (a):", derivadas_q2(x_a, f_a))
print("Q2 (b):", derivadas_q2(x_b, f_b))
```
