# 3. Use a fórmula de três pontos mais precisa
# para completar as tabelas:

# (a)
# x     f(x)      f'(x)
# 1.1   9.025013
# 1.2   11.02318
# 1.3   13.46374
# 1.4   16.44465

# (b)
# x     f(x)      f'(x)
# 8.1   16.94410
# 8.3   17.56492
# 8.5   18.19056
# 8.7   18.82091

```python
import numpy as np

def derivada_3pontos(x, f):
    n = len(x)
    df = np.zeros(n)

    i = 0
    while i < n:
        if i == 0:
            # progressiva 3 pontos
            h = x[1] - x[0]
            df[i] = (-3*f[i] + 4*f[i+1] - f[i+2]) / (2*h)
        elif i == n-1:
            # regressiva 3 pontos
            h = x[i] - x[i-1]
            df[i] = (3*f[i] - 4*f[i-1] + f[i-2]) / (2*h)
        else:
            # centrada
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i-1]) / (2*h)
        i += 1

    return df

# Dados
x_a = np.array([1.1, 1.2, 1.3, 1.4])
f_a = np.array([9.025013, 11.02318, 13.46374, 16.44465])

x_b = np.array([8.1, 8.3, 8.5, 8.7])
f_b = np.array([16.94410, 17.56492, 18.19056, 18.82091])

print("Q3 (a):", derivada_3pontos(x_a, f_a))
print("Q3 (b):", derivada_3pontos(x_b, f_b))
```
