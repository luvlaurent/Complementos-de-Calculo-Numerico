Utilize as fórmulas fornecidas para determinar, com maior precisão possível, aproximações para completar as tabelas.

| 2,1  | -1,709847   |       |
| 2,2  | -1,373823   |       |
| 2,3  | -1,119214   |       |

## b)

| x    | f(x)        | f'(x) |
|------|-------------|-------|
| -3,0 | 9,367879    |       |

| -2,6 | 7,180350    |       |
| -2,4 | 6,209329    |       |
| -2,2 | 5,320305    |       |
| -2,0 | 4,513417    |       |

```python
# Questão 6 - Derivadas numéricas com maior precisão possível
# Fórmulas usadas:
# - Fórmula de 5 pontos (mais precisa): f'(x) = [f(x-2h) - 8f(x-h) + 8f(x+h) - f(x+2h)] / (12h)
# - Fórmula de 3 pontos: f'(x) = [f(x+h) - f(x-h)] / (2h)
# - Fórmula de 2 pontos (extremos): f'(x) = [f(x+h) - f(x)] / h  ou  [f(x) - f(x-h)] / h

#Pontos internos (com 2 vizinhos de cada lado): fórmula de 5 pontos (4ª ordem)")
#Pontos internos (com apenas 1 vizinho de cada lado): fórmula de 3 pontos (2ª ordem)")
#Extremos: fórmula de 2 pontos (progressiva ou regressiva)\n")

print("RESOLUÇÃO DA QUESTÃO 6")


print("Questão 6a")

# Dados da tabela a)
x_a = [2.1, 2.2, 2.3, 2.4, 2.5, 2.6]
f_a = [-1.709847, -1.373823, -1.119214, -0.9160143, -0.7470223, -0.6015966]
h_a = x_a[1] - x_a[0]  # h = 0.1

print("\nDados:")
print("x    =", x_a)
print("f(x) =", f_a)
print(f"Passo h = {h_a}\n")


# Lista para armazenar as derivadas
derivadas_a = []

for i in range(len(x_a)):
    if i >= 2 and i <= len(x_a) - 3:
        # Fórmula de 5 pontos (mais precisa)
        deriv = (f_a[i-2] - 8*f_a[i-1] + 8*f_a[i+1] - f_a[i+2]) / (12 * h_a)
        print(f"\nPara x = {x_a[i]}:")
        print(f"  Usando fórmula de 5 pontos:")
        print(f"  f'(x) = [f(x-2h) - 8f(x-h) + 8f(x+h) - f(x+2h)] / (12h)")
        print(f"  f'({x_a[i]}) = [{f_a[i-2]:.6f} - 8*{f_a[i-1]:.6f} + 8*{f_a[i+1]:.6f} - {f_a[i+2]:.6f}] / (12*{h_a})")
        print(f"  f'({x_a[i]}) = {deriv:.8f}")
        derivadas_a.append(deriv)
    
    elif i >= 1 and i <= len(x_a) - 2:
        # Fórmula de 3 pontos
        deriv = (f_a[i+1] - f_a[i-1]) / (2 * h_a)
        print(f"\nPara x = {x_a[i]}:")
        print(f"  Usando fórmula de 3 pontos:")
        print(f"  f'(x) = [f(x+h) - f(x-h)] / (2h)")
        print(f"  f'({x_a[i]}) = [{f_a[i+1]:.6f} - {f_a[i-1]:.6f}] / (2*{h_a})")
        print(f"  f'({x_a[i]}) = {deriv:.8f}")
        derivadas_a.append(deriv)
    
    elif i == 0:
        # Primeiro ponto - fórmula progressiva de 2 pontos
        deriv = (f_a[i+1] - f_a[i]) / h_a
        print(f"\nPara x = {x_a[i]} (extremo esquerdo):")
        print(f"  Usando fórmula progressiva de 2 pontos:")
        print(f"  f'(x) = [f(x+h) - f(x)] / h")
        print(f"  f'({x_a[i]}) = [{f_a[i+1]:.6f} - {f_a[i]:.6f}] / {h_a}")
        print(f"  f'({x_a[i]}) = {deriv:.8f}")
        derivadas_a.append(deriv)
    
    elif i == len(x_a) - 1:
        # Último ponto - fórmula regressiva de 2 pontos
        deriv = (f_a[i] - f_a[i-1]) / h_a
        print(f"\nPara x = {x_a[i]} (extremo direito):")
        print(f"  Usando fórmula regressiva de 2 pontos:")
        print(f"  f'(x) = [f(x) - f(x-h)] / h")
        print(f"  f'({x_a[i]}) = [{f_a[i]:.6f} - {f_a[i-1]:.6f}] / {h_a}")
        print(f"  f'({x_a[i]}) = {deriv:.8f}")
        derivadas_a.append(deriv)

# Exibir tabela completa da 6a
print("\n" + "-" * 60)
print("Tabela completa (6a):")
print("-" * 60)
print("   x    |     f(x)      |      f'(x)")
print("-" * 60)
for i in range(len(x_a)):
    print(f"  {x_a[i]:.1f}   |  {f_a[i]:12.7f}  |  {derivadas_a[i]:12.8f}")


# QUESTÃO 6b

print("Questão 6b")


# Dados da tabela b)
x_b = [-3.0, -2.8, -2.6, -2.4, -2.2, -2.0]
f_b = [9.367879, 8.233241, 7.180350, 6.209329, 5.320305, 4.513417]
h_b = x_b[1] - x_b[0]  # h = 0.2

print("\nDados:")
print("x    =", x_b)
print("f(x) =", f_b)
print(f"Passo h = {h_b}\n")

print("-" * 60)
print("Cálculos passo a passo:")
print("-" * 60)

# Lista para armazenar as derivadas
derivadas_b = []

for i in range(len(x_b)):
    if i >= 2 and i <= len(x_b) - 3:
        # Fórmula de 5 pontos (mais precisa)
        deriv = (f_b[i-2] - 8*f_b[i-1] + 8*f_b[i+1] - f_b[i+2]) / (12 * h_b)
        print(f"\nPara x = {x_b[i]}:")
        print(f"  Usando fórmula de 5 pontos:")
        print(f"  f'({x_b[i]}) = [{f_b[i-2]:.6f} - 8*{f_b[i-1]:.6f} + 8*{f_b[i+1]:.6f} - {f_b[i+2]:.6f}] / (12*{h_b})")
        print(f"  f'({x_b[i]}) = {deriv:.8f}")
        derivadas_b.append(deriv)
    
    elif i >= 1 and i <= len(x_b) - 2:
        # Fórmula de 3 pontos
        deriv = (f_b[i+1] - f_b[i-1]) / (2 * h_b)
        print(f"\nPara x = {x_b[i]}:")
        print(f"  Usando fórmula de 3 pontos:")
        print(f"  f'({x_b[i]}) = [{f_b[i+1]:.6f} - {f_b[i-1]:.6f}] / (2*{h_b})")
        print(f"  f'({x_b[i]}) = {deriv:.8f}")
        derivadas_b.append(deriv)
    
    elif i == 0:
        # Primeiro ponto - fórmula progressiva de 2 pontos
        deriv = (f_b[i+1] - f_b[i]) / h_b
        print(f"\nPara x = {x_b[i]} (extremo esquerdo):")
        print(f"  Usando fórmula progressiva de 2 pontos:")
        print(f"  f'({x_b[i]}) = [{f_b[i+1]:.6f} - {f_b[i]:.6f}] / {h_b}")
        print(f"  f'({x_b[i]}) = {deriv:.8f}")
        derivadas_b.append(deriv)
    
    elif i == len(x_b) - 1:
        # Último ponto - fórmula regressiva de 2 pontos
        deriv = (f_b[i] - f_b[i-1]) / h_b
        print(f"\nPara x = {x_b[i]} (extremo direito):")
        print(f"  Usando fórmula regressiva de 2 pontos:")
        print(f"  f'({x_b[i]}) = [{f_b[i]:.6f} - {f_b[i-1]:.6f}] / {h_b}")
        print(f"  f'({x_b[i]}) = {deriv:.8f}")
        derivadas_b.append(deriv)

# Exibir tabela completa da 6b
print("\n" + "-" * 60)
print("Tabela completa (6b):")
print("-" * 60)
print("   x    |     f(x)      |      f'(x)")
print("-" * 60)
for i in range(len(x_b)):
    print(f"  {x_b[i]:.1f}   |  {f_b[i]:12.6f}  |  {derivadas_b[i]:12.8f}")


```
