# Questão 7

Seja \( f(x) = 3xe^x - \cos x \). Use os dados a seguir e a Equação (4.9) para obter uma aproximação de \( f''(1,3) \) com \( h = 0,1 \) e com \( h = 0,01 \).

| x     | f(x)     |
|-------|----------|
| 1,20  | 11,59006 |
| 1,29  | 13,78176 |
| 1,30  | 14,04276 |
| 1,31  | 14,30741 |
| 1,40  | 16,86187 |

Compare seus resultados com \( f''(1,3) \).

```python
# Questão 7 - Aproximação da segunda derivada f''(1,3)
# Equação (4.9): f''(x) ≈ [f(x+h) - 2f(x) + f(x-h)] / h²
# Dados fornecidos na tabela
# Usando os valores mais próximos de x = 1,30

print("RESOLUÇÃO DA QUESTÃO 7")

x_valores = {
    1.20: 11.59006,
    1.29: 13.78176,
    1.30: 14.04276,
    1.31: 14.30741,
    1.40: 16.86187
}

print("\nDados fornecidos:")
print("-" * 40)
print("   x    |     f(x)")
print("-" * 40)
for x in sorted(x_valores.keys()):
    print(f"  {x:.2f}  |  {x_valores[x]:.8f}")
print("-" * 40)

# Cálculo da derivada segunda exata (para comparação)
# f(x) = 3x·e^x - cos(x)
# f''(x) = 3·e^x·(x + 2) + cos(x)

print("CÁLCULO DO VALOR EXATO DE f''(1,3)")

# Função para calcular e^x manualmente (sem importar bibliotecas)
def exp_manual(x):
    # Usando a série de Taylor para e^x com precisão suficiente
    resultado = 1.0
    termo = 1.0
    n = 1
    while abs(termo) > 1e-12:
        termo = termo * x / n
        resultado += termo
        n += 1
    return resultado

x0 = 1.3
e_130 = exp_manual(x0)

# f''(x) = 3·e^x·(x + 2) + cos(x)
# cos(1.3) usando série de Taylor
def cos_manual(x):
    resultado = 1.0
    termo = 1.0
    n = 1
    sinal = -1
    while abs(termo) > 1e-12:
        termo = termo * x * x / ((2*n - 1) * (2*n))
        resultado += sinal * termo
        sinal = -sinal
        n += 1
    return resultado

cos_130 = cos_manual(x0)

f2_exato = 3 * e_130 * (x0 + 2) + cos_130

print(f"\nPara x = {x0}:")
print(f"  e^1.3 ≈ {e_130:.10f}")
print(f"  cos(1.3) ≈ {cos_130:.10f}")
print(f"  f''(1.3) exato = 3·e^1.3·(1.3 + 2) + cos(1.3)")
print(f"  f''(1.3) exato = {f2_exato:.10f}")

# Aproximação com h = 0,1
# Usando x = 1,30 como ponto central e h = 0,1
# f(x+h) = f(1,40) = 16,86187
# f(x-h) = f(1,20) = 11,59006

print("APROXIMAÇÃO COM h = 0,1")

x_central = 1.30
h1 = 0.1

# Valores da tabela
f_central = x_valores[1.30]
f_mais_h = x_valores[1.40]      # 1,30 + 0,1 = 1,40
f_menos_h = x_valores[1.20]     # 1,30 - 0,1 = 1,20

print(f"\nPonto central: x = {x_central}")
print(f"h = {h1}")
print(f"f(x) = f({x_central}) = {f_central:.8f}")
print(f"f(x+h) = f({x_central + h1}) = f(1,40) = {f_mais_h:.8f}")
print(f"f(x-h) = f({x_central - h1}) = f(1,20) = {f_menos_h:.8f}")

# Fórmula (4.9): f''(x) ≈ [f(x+h) - 2f(x) + f(x-h)] / h²
aprox_h1 = (f_mais_h - 2*f_central + f_menos_h) / (h1 ** 2)

print(f"\nAplicando a fórmula:")
print(f"f''(1,3) ≈ [f(1,40) - 2·f(1,30) + f(1,20)] / ({h1})²")
print(f"f''(1,3) ≈ [{f_mais_h:.8f} - 2·{f_central:.8f} + {f_menos_h:.8f}] / {h1*h1}")
print(f"f''(1,3) ≈ [{(f_mais_h - 2*f_central + f_menos_h):.8f}] / {h1*h1:.2f}")
print(f"f''(1,3) ≈ {aprox_h1:.10f}")

erro1 = abs(aprox_h1 - f2_exato)
erro_relativo1 = (erro1 / abs(f2_exato)) * 100

print(f"\nComparação com o valor exato:")
print(f"  Valor exato de f''(1,3) = {f2_exato:.10f}")
print(f"  Aproximação com h = 0,1 = {aprox_h1:.10f}")
print(f"  Erro absoluto = {erro1:.10f}")
print(f"  Erro relativo = {erro_relativo1:.6f}%")

# Aproximação com h = 0,01
# Usando os valores da tabela: f(1,31) e f(1,29)
# h = 0,01

print("APROXIMAÇÃO COM h = 0,01")

h2 = 0.01

# Valores da tabela para h = 0,01
f_mais_h2 = x_valores[1.31]     # 1,30 + 0,01 = 1,31
f_menos_h2 = x_valores[1.29]    # 1,30 - 0,01 = 1,29

print(f"\nPonto central: x = {x_central}")
print(f"h = {h2}")
print(f"f(x) = f({x_central}) = {f_central:.8f}")
print(f"f(x+h) = f({x_central + h2}) = f(1,31) = {f_mais_h2:.8f}")
print(f"f(x-h) = f({x_central - h2}) = f(1,29) = {f_menos_h2:.8f}")

# Fórmula (4.9): f''(x) ≈ [f(x+h) - 2f(x) + f(x-h)] / h²
aprox_h2 = (f_mais_h2 - 2*f_central + f_menos_h2) / (h2 ** 2)

print(f"\nAplicando a fórmula:")
print(f"f''(1,3) ≈ [f(1,31) - 2·f(1,30) + f(1,29)] / ({h2})²")
print(f"f''(1,3) ≈ [{f_mais_h2:.8f} - 2·{f_central:.8f} + {f_menos_h2:.8f}] / {h2*h2:.4f}")
print(f"f''(1,3) ≈ [{(f_mais_h2 - 2*f_central + f_menos_h2):.10f}] / {h2*h2:.4f}")
print(f"f''(1,3) ≈ {aprox_h2:.10f}")

erro2 = abs(aprox_h2 - f2_exato)
erro_relativo2 = (erro2 / abs(f2_exato)) * 100

print(f"\nComparação com o valor exato:")
print(f"  Valor exato de f''(1,3) = {f2_exato:.10f}")
print(f"  Aproximação com h = 0,01 = {aprox_h2:.10f}")
print(f"  Erro absoluto = {erro2:.10f}")
print(f"  Erro relativo = {erro_relativo2:.6f}%")

print("RESUMO FINAL - COMPARAÇÃO DOS RESULTADOS")


print("\n" + "-" * 60)
print("h       |  f''(1,3) aproximado  |  Erro absoluto  |  Erro relativo")
print("-" * 60)
print(f"0,1     |  {aprox_h1:.10f}     |  {erro1:.2e}      |  {erro_relativo1:.6f}%")
print(f"0,01    |  {aprox_h2:.10f}     |  {erro2:.2e}      |  {erro_relativo2:.6f}%")
print(f"EXATO   |  {f2_exato:.10f}     |     ---         |     ---")
print("-" * 60)


print("""
Observações importantes:

1. Com h = 0,1, o erro é maior porque a aproximação é menos precisa.
   A fórmula de diferença central tem erro da ordem O(h²).

2. Com h = 0,01, o erro é aproximadamente 100 vezes menor,
   conforme esperado para uma fórmula de segunda ordem.

3. A redução de h por fator 10 reduziu o erro por fator ~100,
   confirmando a ordem O(h²) da Equação (4.9).

4. Quanto menor o h, mais precisa é a aproximação, desde que
   os valores de f(x) não tenham erros de arredondamento.
""")
```
