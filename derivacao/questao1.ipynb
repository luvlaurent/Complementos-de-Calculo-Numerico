# 1. Use as fórmulas de diferenças progressivas e regressivas
# para completar as tabelas:

# (a)
# x     f(x)     f'(x)
# 0.5   0.4794
# 0.6   0.5646
# 0.7   0.6442

# (b)
# x     f(x)     f'(x)
# 0.0   0.00000
# 0.2   0.74140
# 0.4   1.3718

```python
import numpy as np

def derivadas_q1(x, f):
    n = len(x)
    df = np.zeros(n)

    i = 0
    while i < n:
        if i == 0:
            # progressiva
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i]) / h
        elif i == n-1:
            # regressiva
            h = x[i] - x[i-1]
            df[i] = (f[i] - f[i-1]) / h
        else:
            # centrada
            h = x[i+1] - x[i]
            df[i] = (f[i+1] - f[i-1]) / (2*h)
        i += 1

    return df

# Dados
x_a = np.array([0.5, 0.6, 0.7])
f_a = np.array([0.4794, 0.5646, 0.6442])

x_b = np.array([0.0, 0.2, 0.4])
f_b = np.array([0.00000, 0.74140, 1.3718])

print("Q1 (a):", derivadas_q1(x_a, f_a))
print("Q1 (b):", derivadas_q1(x_b, f_b))
```
