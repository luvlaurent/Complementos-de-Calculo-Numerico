## Questão 2

Utilize um método de iteração de ponto fixo para determinar uma solução com precisão de \(10^{-2}\) para

\[x^3 - x - 1 = 0\]

no intervalo \([1, 2]\). Utilize \(p_0 = 1\).

```python
# Questão 2 - Ponto Fixo

def ponto_fixo(p0, tol=1e-2):
    def g(x):
        return (x + 1)**(1/3)

    p = p0

    while True:
        p_novo = g(p)

        if abs(p_novo - p) < tol:
            return p_novo

        p = p_novo

raiz = ponto_fixo(1)
print("Aproximação =", raiz)
```
