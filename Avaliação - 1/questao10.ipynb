## b)

∫₀³ x³ sin(2x) dx

Trapézio = -11,316328

Simpson = -2,819549

Ponto médio = 1,428840

Valor exato ≈ -12,583190

```python
# QUESTAO 10

import math

#a

def f_a(t):
    return t**4

a = 0.5
b = 1.0

trap = (b - a) * (f_a(a) + f_a(b)) / 2

simp = (b - a) * (
    f_a(a)
    + 4 * f_a((a + b) / 2)
    + f_a(b)
) / 6

mid = (b - a) * f_a((a + b) / 2)

exato = (b**5 - a**5) / 5

print("QUESTÃO 10A")
print("Trapézio =", trap)
print("Simpson =", simp)
print("Ponto Médio =", mid)
print("Valor Exato =", exato)

#b

def f_b(x):
    return math.exp(3*x) * math.sin(2*x)

a = 0
b = 3

trap = (b - a) * (f_b(a) + f_b(b)) / 2

simp = (b - a) * (
    f_b(a)
    + 4 * f_b((a + b) / 2)
    + f_b(b)
) / 6

mid = (b - a) * f_b((a + b) / 2)

print("\nQUESTÃO 10B")
print("Trapézio =", trap)
print("Simpson =", simp)
print("Ponto Médio =", mid)
```
