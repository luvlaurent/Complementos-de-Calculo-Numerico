# Questão 7

Utilizando diferenças progressivas e regressivas.

## Letra A

Dados:

| x | f(x) |
|---|---|
|0,5|0,4794|
|0,6|0,5646|
|0,7|0,6442|

Passo:

h = 0,1

Diferença progressiva:

f'(0,5) ≈ (f(0,6)-f(0,5))/h

Diferença central:

f'(0,6) ≈ (f(0,7)-f(0,5))/(2h)

Diferença regressiva:

f'(0,7) ≈ (f(0,7)-f(0,6))/h

Resultados:

| x | f'(x) |
|---|---|
|0,5|0,852000|
|0,6|0,824000|
|0,7|0,796000|


## Letra B

Dados:

| x | f(x) |
|---|---|
|0,0|0,00000|
|0,2|0,74110|
|0,4|1,37180|

Passo:

h = 0,2

Resultados:

| x | f'(x) |
|---|---|
|0,0|3,705500|
|0,2|3,429500|
|0,4|3,153500|

```python

x = [0.5, 0.6, 0.7]
f = [0.4794, 0.5646, 0.6442]

h = 0.1

# progressiva
f1_05 = (f[1] - f[0]) / h

# central (usando os vizinhos)
f1_06 = (f[2] - f[0]) / (2*h)

# regressiva
f1_07 = (f[2] - f[1]) / h

print(f"x = {x[0]} -> f'(x) = {f1_05:.6f}")
print(f"x = {x[1]} -> f'(x) = {f1_06:.6f}")
print(f"x = {x[2]} -> f'(x) = {f1_07:.6f}")

#b
x = [0.0, 0.2, 0.4]
f = [0.00000, 0.74140, 1.37180]

h = 0.2

# progressiva
f1_00 = (f[1] - f[0]) / h

# central
f1_02 = (f[2] - f[0]) / (2*h)

# regressiva
f1_04 = (f[2] - f[1]) / h

print(f"x = {x[0]} -> f'(x) = {f1_00:.6f}")
print(f"x = {x[1]} -> f'(x) = {f1_02:.6f}")
print(f"x = {x[2]} -> f'(x) = {f1_04:.6f}")
```
