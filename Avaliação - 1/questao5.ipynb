## Questão 5

Use o método de Neville para obter aproximações para os polinômios interpoladores de Lagrange de graus um, dois e três e aproximar o seguinte:

- **a)** \(f(0,46)\) se  
  \(f(0) = 1\),  
  \(f(0,25) = 1,64872\),  
  \(f(0,5) = 2,71828\),  
  \(f(0,75) = 4,48169\)

- **d)** \(f(0,3)\) se  
  \(f(-1) = 0,86199480\),  
  \(f(-0,5) = 0,95802009\),  
  \(f(0) = 1,0986123\),  
  \(f(0,5) = 1,2943767\)

```python
# Questão 5 - Método de Neville

def neville(x, xs, ys):
    n = len(xs)

    Q = [[0]*n for _ in range(n)]

    for i in range(n):
        Q[i][0] = ys[i]

    for j in range(1, n):
        for i in range(n-j):
            Q[i][j] = (
                (x - xs[i+j])*Q[i][j-1]
                - (x - xs[i])*Q[i+1][j-1]
            )/(xs[i] - xs[i+j])

    return Q[0][n-1]


# Item a
xs = [0, 0.25, 0.5, 0.75]
ys = [1, 1.64872, 2.71828, 4.48169]

for grau in [2, 3, 4]:
    print(
        f"Grau {grau-1}:",
        neville(0.46, xs[:grau], ys[:grau])
    )


# Item b
xs = [-1, -0.5, 0, 0.5]
ys = [0.86199480, 0.95802009, 1.0986123, 1.2943767]

for grau in [2, 3, 4]:
    print(
        f"Grau {grau-1}:",
        neville(0.3, xs[:grau], ys[:grau])
    )
```
