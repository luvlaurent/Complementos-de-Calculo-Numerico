## Questão 3

A equação \[x^2 - 10 \cos x = 0\]

tem duas soluções, \(\pm 1.3793646\). Utilize o método de Newton para encontrar uma aproximação das soluções com precisão de \(10^{-5}\), com os valores de \(p_0\) a seguir:

- a) \(p_0 = -60\)
- b) \(p_0 = -15\)
- c) \(p_0 = 15\)
- d) \(p_0 = 60\)

```python
# Questão 3 - Método de Newton

import math

def newton(p0, tol=1e-5):
    def f(x):
        return x**2 - 10*math.cos(x)

    def df(x):
        return 2*x + 10*math.sin(x)

    p = p0

    while True:
        p_novo = p - f(p)/df(p)

        if abs(p_novo - p) < tol:
            return p_novo

        p = p_novo


iniciais = [-60, -15, 15, 60]

for p0 in iniciais:
    raiz = newton(p0)
    print(f"p0={p0} -> {raiz}")
```
