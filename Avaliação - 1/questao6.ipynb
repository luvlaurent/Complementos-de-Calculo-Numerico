## Questão 6

Seja \(f(x) = 3xe^x - e^{2x}\).

- **a)** Faça a aproximação de \(f(1,02)\) pelo polinômio interpolador de Hermite de grau no máximo três usando \(x_0 = 1\) e \(x_1 = 1,05\). Compare o erro real com o limitante de erro.

- **b)** Repita (a) com o polinômio interpolador de Hermite de grau no máximo cinco, usando \(x_0 = 1\), \(x_1 = 1,05\) e \(x_2 = 1,07\).

```python
# Questão 6 - Hermite

import math

def f(x):
    return 3*x*math.exp(x) - math.exp(2*x)

def df(x):
    return 3*math.exp(x) + 3*x*math.exp(x) - 2*math.exp(2*x)


def hermite(x_nodes, y_nodes, dy_nodes, x):

    n = len(x_nodes)

    z = []
    Q = [[0]*(2*n) for _ in range(2*n)]

    for i in range(n):
        z.append(x_nodes[i])
        z.append(x_nodes[i])

    for i in range(n):
        Q[2*i][0] = y_nodes[i]
        Q[2*i+1][0] = y_nodes[i]

        Q[2*i+1][1] = dy_nodes[i]

        if i != 0:
            Q[2*i][1] = (
                Q[2*i][0] - Q[2*i-1][0]
            )/(z[2*i] - z[2*i-1])

    m = 2*n

    for j in range(2, m):
        for i in range(m-j):
            Q[i][j] = (
                Q[i+1][j-1] - Q[i][j-1]
            )/(z[i+j] - z[i])

    p = Q[0][0]
    prod = 1

    for i in range(1, m):
        prod *= (x - z[i-1])
        p += Q[0][i]*prod

    return p


# Item a
xs = [1.0, 1.05]
ys = [f(x) for x in xs]
dys = [df(x) for x in xs]

aprox = hermite(xs, ys, dys, 1.02)

print("Hermite grau 3 =", aprox)


# Item b
xs = [1.0, 1.05, 1.07]
ys = [f(x) for x in xs]
dys = [df(x) for x in xs]

aprox = hermite(xs, ys, dys, 1.02)

print("Hermite grau 5 =", aprox)
```
