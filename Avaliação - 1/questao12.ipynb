```python
import math

def gauss1(f, a, b):

    x1 = 0
    w1 = 2

    t1 = ((b-a)/2)*x1 + (a+b)/2

    return ((b-a)/2) * w1 * f(t1)


def gauss2(f, a, b):

    pontos = [
        -1/math.sqrt(3),
         1/math.sqrt(3)
    ]

    pesos = [1, 1]

    soma = 0

    for p, w in zip(pontos, pesos):

        t = ((b-a)/2)*p + (a+b)/2

        soma += w*f(t)

    return ((b-a)/2) * soma


def gauss3(f, a, b):

    pontos = [
        -math.sqrt(3/5),
         0,
         math.sqrt(3/5)
    ]

    pesos = [
        5/9,
        8/9,
        5/9
    ]

    soma = 0

    for p, w in zip(pontos, pesos):

        t = ((b-a)/2)*p + (a+b)/2

        soma += w*f(t)

    return ((b-a)/2) * soma


#a
# Integral de 1 até 15
# x² ln(x)

print("QUESTÃO 12A")

def f(x):
    return x**2 * math.log(x)

a = 1
b = 1.5

print("n = 1 :", gauss1(f, a, b))
print("n = 2 :", gauss2(f, a, b))
print("n = 3 :", gauss3(f, a, b))

#b
# Integral de 0 até pi/4
# x² sen(x)

print("\nQUESTÃO 12B")

def g(x):
    return x**2 * math.sin(x)

a = 0
b = math.pi/4

print("n = 1 :", gauss1(g, a, b))
print("n = 2 :", gauss2(g, a, b))
print("n = 3 :", gauss3(g, a, b))
```
