Use o método de Neville para obter aproximações para os polinômios interpoladores de Lagrange de graus um, dois e três e aproximar o seguinte:

**a)** f(0,43) se f(0) = 1, f(0,25) = 1,64872, f(0,5) = 2,71828, f(0,75) = 4,48169.

**b)** f(0) se f(-0,5) = 1,93750, f(-0,25) = 1,33203, f(0,25) = 0,800781, f(0,5) = 0,687500.

**c)** f(0,18) se f(0) = -0,29004986, f(0,2) = -0,56079734, f(0,3) = -0,81401972, f(0,4) = -1,0526302.

**d)** f(0,25) se f(-1) = 0,86199480, f(-0,5) = 0,95802009, f(0) = 1,0986123, f(0,5) = 1,2943767.

```python
# Método de Neville
def neville(xp, yp, x):
    n = len(xp)
    Q = [[0]*n for _ in range(n)]

    for i in range(n):
        Q[i][0] = yp[i]

    for j in range(1, n):
        for i in range(n - j):
            Q[i][j] = ((x - xp[i+j])*Q[i][j-1] - (x - xp[i])*Q[i+1][j-1])/(xp[i] - xp[i+j])

    return Q[0][n-1]


# a
xp = [8.1, 8.3, 8.6, 8.7]
yp = [16.94410, 17.56492, 18.50515, 18.82091]
print("1a:", neville(xp, yp, 8.4))

# b
xp = [-0.75, -0.5, -0.25, 0]
yp = [-0.07181250, -0.02475000, 0.33493750, 1.10100000]
print("1b:", neville(xp, yp, -1/3))

# c
xp = [0.1, 0.2, 0.3, 0.4]
yp = [0.62049958, -0.28398668, 0.00660095, 0.24842440]
print("1c:", neville(xp, yp, 0.25))

# d
xp = [0.6, 0.7, 0.8, 1.0]
yp = [-0.17694460, 0.01375227, 0.22363362, 0.65809197]
print("1d:", neville(xp, yp, 0.9))

```
