# Questão 7 - Método de Runge-Kutta de Quarta Ordem

Repita o Exercício 1 usando o método de Runge-Kutta de quarta ordem.

## Letra a)
$$y' = te^{3t} - 2y, \quad 0 \leq t \leq 1, \quad y(0) = 0, \quad \text{com } h = 0,5$$

## Letra b)
$$y' = 1 + (t - y)^2, \quad 2 \leq t \leq 3, \quad y(2) = 1, \quad \text{com } h = 0,5$$

## Letra c)
$$y' = 1 + \frac{y}{t}, \quad 1 \leq t \leq 2, \quad y(1) = 2, \quad \text{com } h = 0,25$$

## Letra d)
$$y' = \cos(2t) + \sin(3t), \quad 0 \leq t \leq 1, \quad y(0) = 1, \quad \text{com } h = 0,25$$

```python
import numpy as np
import matplotlib.pyplot as plt

def rk4_method(f, t0, y0, h, t_end):
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    y = np.zeros(n+1)
    t[0] = t0
    y[0] = y0
    
    for i in range(n):
        k1 = f(t[i], y[i])
        k2 = f(t[i] + h/2, y[i] + (h/2)*k1)
        k3 = f(t[i] + h/2, y[i] + (h/2)*k2)
        k4 = f(t[i] + h, y[i] + h*k3)
        
        y[i+1] = y[i] + (h/6)*(k1 + 2*k2 + 2*k3 + k4)
        t[i+1] = t[i] + h
    
    return t, y

# Letra A
def f7a(t, y):
    return t * np.exp(3*t) - 2*y

t0, y0, h, t_end = 0, 0, 0.5, 1
t_a, y_a = rk4_method(f7a, t0, y0, h, t_end)

print("QUESTÃO 7 - RUNGE-KUTTA 4ª ORDEM (Questão 1)")

print("\n--- LETRA A ---")
print("t\t\ty (aproximado)")

for i in range(len(t_a)):
    print(f"{t_a[i]:.2f}\t\t{y_a[i]:.6f}")

#Letra B
def f7b(t, y):
    return 1 + (t - y)**2

t0, y0, h, t_end = 2, 1, 0.5, 3
t_b, y_b = rk4_method(f7b, t0, y0, h, t_end)

print("\n--- LETRA B ---")
print("t\t\ty (aproximado)")
print("-"*30)
for i in range(len(t_b)):
    print(f"{t_b[i]:.2f}\t\t{y_b[i]:.6f}")

#Letra C
def f7c(t, y):
    return 1 + y/t

t0, y0, h, t_end = 1, 2, 0.25, 2
t_c, y_c = rk4_method(f7c, t0, y0, h, t_end)

print("\n--- LETRA C ---")
print("t\t\ty (aproximado)")
print("-"*30)
for i in range(len(t_c)):
    print(f"{t_c[i]:.2f}\t\t{y_c[i]:.6f}")

#Letra D
def f7d(t, y):
    return np.cos(2*t) + np.sin(3*t)

t0, y0, h, t_end = 0, 1, 0.25, 1
t_d, y_d = rk4_method(f7d, t0, y0, h, t_end)

print("\n--- LETRA D ---")
print("t\t\ty (aproximado)")
print("-"*30)
for i in range(len(t_d)):
    print(f"{t_d[i]:.2f}\t\t{y_d[i]:.6f}")

#Gráficos
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(t_a, y_a, 'bo-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('7a - y\' = te^(3t) - 2y')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(t_b, y_b, 'ro-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('7b - y\' = 1 + (t-y)^2')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(t_c, y_c, 'go-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('7c - y\' = 1 + y/t')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(t_d, y_d, 'mo-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('7d - y\' = cos(2t) + sin(3t)')
plt.grid(True)
plt.legend()

plt.suptitle('Questão 7 - Runge-Kutta 4ª Ordem', fontsize=16)
plt.tight_layout()
plt.show()
```
