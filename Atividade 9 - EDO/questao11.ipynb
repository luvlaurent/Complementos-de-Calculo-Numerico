# Questão 11 - Runge-Kutta para Sistemas de EDOs de Ordem Superior

Use o Algoritmo de Runge-Kutta para sistemas para encontrar aproximações das soluções dos seguintes sistemas de equações diferenciais de ordem superior e compare os resultados às soluções reais.

---

## Letra a)

$$y'' - 2y' + y = te^t - t, \quad 0 \leq t \leq 1, \quad y(0) = y'(0) = 0; \quad \text{com } h = 0,1$$

**Solução real:**

$$y(t) = \frac{1}{6}t^3e^t - te^t + 2e^t - t - 2$$

---

## Letra b)

$$t^2y'' - 2ty' + 2y = t^3 \ln t, \quad 1 \leq t \leq 2, \quad y(1) = 1, \quad y'(1) = 0; \quad \text{com } h = 0,1$$

**Solução real:**

$$y(t) = \frac{7}{4}t + \frac{1}{2}t^3 \ln t - \frac{3}{4}t^3$$

---

## Letra c)

$$y''' + 2y'' - y' - 2y = e^t, \quad 0 \leq t \leq 3, \quad y(0) = 1, \quad y'(0) = 2, \quad y''(0) = 0; \quad \text{com } h = 0,2$$

**Solução real:**

$$y(t) = \frac{43}{36}e^t + \frac{1}{4}e^{-t} - \frac{4}{9}e^{-2t} + \frac{1}{6}te^t$$

---

## Letra d)

$$t^3y''' - t^2y'' + 3ty' - 4y = 5t^3 \ln t + 9t^3, \quad 1 \leq t \leq 2, \quad y(1) = 0, \quad y'(1) = 1, \quad y''(1) = 3; \quad \text{com } h = 0,1$$

**Solução real:**

$$y(t) = -t^2 + t \cos(\ln t) + t \sin(\ln t) + t^3 \ln t$$

---

## Transformação para Sistemas de Primeira Ordem

### Letra a)
Seja $u_1 = y$ e $u_2 = y'$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = 2u_2 - u_1 + te^t - t
\end{cases}$$

com $u_1(0) = 0$ e $u_2(0) = 0$.

---

### Letra b)
Seja $u_1 = y$ e $u_2 = y'$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = \frac{2}{t}u_2 - \frac{2}{t^2}u_1 + t\ln t
\end{cases}$$

com $u_1(1) = 1$ e $u_2(1) = 0$.

---

### Letra c)
Seja $u_1 = y$, $u_2 = y'$ e $u_3 = y''$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = u_3 \\
u_3' = -2u_3 + u_2 + 2u_1 + e^t
\end{cases}$$

com $u_1(0) = 1$, $u_2(0) = 2$ e $u_3(0) = 0$.

---

### Letra d)
Seja $u_1 = y$, $u_2 = y'$ e $u_3 = y''$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = u_3 \\
u_3' = \frac{1}{t}u_3 - \frac{3}{t^2}u_2 + \frac{4}{t^3}u_1 + 5\ln t + 9
\end{cases}$$

com $u_1(1) = 0$, $u_2(1) = 1$ e $u_3(1) = 3$.

---

## Observações

- Utilize o método de Runge-Kutta de 4ª ordem para sistemas
- Compare os resultados obtidos com as soluções reais fornecidas
- Analise os erros das aproximações

```python
import numpy as np
import matplotlib.pyplot as plt

def rk4_system_2d(f1, f2, t0, u1_0, u2_0, h, t_end):
    """
    Método de Runge-Kutta de 4ª ordem para sistemas de 2 EDOs de primeira ordem
    """
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    u1 = np.zeros(n+1)
    u2 = np.zeros(n+1)
    t[0] = t0
    u1[0] = u1_0
    u2[0] = u2_0
    
    for i in range(n):
        t_curr = t[i]
        u1_curr = u1[i]
        u2_curr = u2[i]
        
        k1_1 = f1(t_curr, u1_curr, u2_curr)
        k1_2 = f2(t_curr, u1_curr, u2_curr)
        
        k2_1 = f1(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2)
        k2_2 = f2(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2)
        
        k3_1 = f1(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2)
        k3_2 = f2(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2)
        
        k4_1 = f1(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2)
        k4_2 = f2(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2)
        
        u1[i+1] = u1_curr + (h/6)*(k1_1 + 2*k2_1 + 2*k3_1 + k4_1)
        u2[i+1] = u2_curr + (h/6)*(k1_2 + 2*k2_2 + 2*k3_2 + k4_2)
        t[i+1] = t_curr + h
    
    return t, u1, u2


def rk4_system_3d(f1, f2, f3, t0, u1_0, u2_0, u3_0, h, t_end):
    """
    Método de Runge-Kutta de 4ª ordem para sistemas de 3 EDOs de primeira ordem
    """
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    u1 = np.zeros(n+1)
    u2 = np.zeros(n+1)
    u3 = np.zeros(n+1)
    t[0] = t0
    u1[0] = u1_0
    u2[0] = u2_0
    u3[0] = u3_0
    
    for i in range(n):
        t_curr = t[i]
        u1_curr = u1[i]
        u2_curr = u2[i]
        u3_curr = u3[i]
        
        k1_1 = f1(t_curr, u1_curr, u2_curr, u3_curr)
        k1_2 = f2(t_curr, u1_curr, u2_curr, u3_curr)
        k1_3 = f3(t_curr, u1_curr, u2_curr, u3_curr)
        
        k2_1 = f1(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        k2_2 = f2(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        k2_3 = f3(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        
        k3_1 = f1(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        k3_2 = f2(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        k3_3 = f3(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        
        k4_1 = f1(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        k4_2 = f2(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        k4_3 = f3(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        
        u1[i+1] = u1_curr + (h/6)*(k1_1 + 2*k2_1 + 2*k3_1 + k4_1)
        u2[i+1] = u2_curr + (h/6)*(k1_2 + 2*k2_2 + 2*k3_2 + k4_2)
        u3[i+1] = u3_curr + (h/6)*(k1_3 + 2*k2_3 + 2*k3_3 + k4_3)
        t[i+1] = t_curr + h
    
    return t, u1, u2, u3

# LETRA A
# y'' - 2y' + y = t*e^t - t
# y(0) = 0, y'(0) = 0
# 
# Transformação: u1 = y, u2 = y'
# u1' = u2
# u2' = 2u2 - u1 + t*e^t - t

def f11a1(t, u1, u2):
    """u1' = u2"""
    return u2

def f11a2(t, u1, u2):
    """u2' = 2u2 - u1 + t*e^t - t"""
    return 2*u2 - u1 + t*np.exp(t) - t

def sol11a(t):
    """Solução real: y(t) = (1/6)t^3*e^t - t*e^t + 2*e^t - t - 2"""
    return (1/6)*t**3*np.exp(t) - t*np.exp(t) + 2*np.exp(t) - t - 2

# Parâmetros da Letra A
t0_a, u1_0_a, u2_0_a, h_a, t_end_a = 0, 0, 0, 0.1, 1

# Executar RK4
t_a, u1_a, u2_a = rk4_system_2d(f11a1, f11a2, t0_a, u1_0_a, u2_0_a, h_a, t_end_a)

# Solução real
y_real_a = sol11a(t_a)

# LETRA B
# t^2*y'' - 2t*y' + 2y = t^3*ln(t)
# y(1) = 1, y'(1) = 0
# 
# Transformação: u1 = y, u2 = y'
# u1' = u2
# u2' = (2/t)*u2 - (2/t^2)*u1 + t*ln(t)

def f11b1(t, u1, u2):
    """u1' = u2"""
    return u2

def f11b2(t, u1, u2):
    """u2' = (2/t)*u2 - (2/t^2)*u1 + t*ln(t)"""
    return (2/t)*u2 - (2/t**2)*u1 + t*np.log(t)

def sol11b(t):
    """Solução real: y(t) = (7/4)t + (1/2)t^3*ln(t) - (3/4)t^3"""
    return (7/4)*t + (1/2)*t**3*np.log(t) - (3/4)*t**3

# Parâmetros da Letra B
t0_b, u1_0_b, u2_0_b, h_b, t_end_b = 1, 1, 0, 0.1, 2

# Executar RK4
t_b, u1_b, u2_b = rk4_system_2d(f11b1, f11b2, t0_b, u1_0_b, u2_0_b, h_b, t_end_b)

# Solução real
y_real_b = sol11b(t_b)

# LETRA C
# y''' + 2y'' - y' - 2y = e^t
# y(0) = 1, y'(0) = 2, y''(0) = 0
# 
# Transformação: u1 = y, u2 = y', u3 = y''
# u1' = u2
# u2' = u3
# u3' = -2u3 + u2 + 2u1 + e^t

def f11c1(t, u1, u2, u3):
    """u1' = u2"""
    return u2

def f11c2(t, u1, u2, u3):
    """u2' = u3"""
    return u3

def f11c3(t, u1, u2, u3):
    """u3' = -2u3 + u2 + 2u1 + e^t"""
    return -2*u3 + u2 + 2*u1 + np.exp(t)

def sol11c(t):
    """Solução real: y(t) = (43/36)e^t + (1/4)e^(-t) - (4/9)e^(-2t) + (1/6)t*e^t"""
    return (43/36)*np.exp(t) + (1/4)*np.exp(-t) - (4/9)*np.exp(-2*t) + (1/6)*t*np.exp(t)

# Parâmetros da Letra C
t0_c, u1_0_c, u2_0_c, u3_0_c, h_c, t_end_c = 0, 1, 2, 0, 0.2, 3

# Executar RK4
t_c, u1_c, u2_c, u3_c = rk4_system_3d(
    f11c1, f11c2, f11c3,
    t0_c, u1_0_c, u2_0_c, u3_0_c,
    h_c, t_end_c
)

# Solução real
y_real_c = sol11c(t_c)

# LETRA D
# t^3*y''' - t^2*y'' + 3t*y' - 4y = 5t^3*ln(t) + 9t^3
# y(1) = 0, y'(1) = 1, y''(1) = 3
# 
# Transformação: u1 = y, u2 = y', u3 = y''
# u1' = u2
# u2' = u3
# u3' = (1/t)*u3 - (3/t^2)*u2 + (4/t^3)*u1 + 5*ln(t) + 9


def f11d1(t, u1, u2, u3):
    """u1' = u2"""
    return u2

def f11d2(t, u1, u2, u3):
    """u2' = u3"""
    return u3

def f11d3(t, u1, u2, u3):
    """u3' = (1/t)*u3 - (3/t^2)*u2 + (4/t^3)*u1 + 5*ln(t) + 9"""
    return (1/t)*u3 - (3/t**2)*u2 + (4/t**3)*u1 + 5*np.log(t) + 9

def sol11d(t):
    """Solução real: y(t) = -t^2 + t*cos(ln(t)) + t*sin(ln(t)) + t^3*ln(t)"""
    return -t**2 + t*np.cos(np.log(t)) + t*np.sin(np.log(t)) + t**3*np.log(t)

# Parâmetros da Letra D
t0_d, u1_0_d, u2_0_d, u3_0_d, h_d, t_end_d = 1, 0, 1, 3, 0.1, 2

# Executar RK4
t_d, u1_d, u2_d, u3_d = rk4_system_3d(
    f11d1, f11d2, f11d3,
    t0_d, u1_0_d, u2_0_d, u3_0_d,
    h_d, t_end_d
)

# Solução real
y_real_d = sol11d(t_d)

# IMPRESSÃO DOS RESULTADOS

print("QUESTÃO 11 - RUNGE-KUTTA PARA SISTEMAS DE ORDEM SUPERIOR")

# Letra A
print("LETRA A")

print("\nEDO: y'' - 2y' + y = t*e^t - t")
print("y(0) = 0, y'(0) = 0")
print(f"Intervalo: 0 <= t <= 1,  h = {h_a}")
print("\n" + "-"*100)
print("t\t\ty (aproximado)\t\ty (real)\t\tErro")
print("-"*100)
for i in range(len(t_a)):
    erro = abs(u1_a[i] - y_real_a[i])
    print(f"{t_a[i]:.2f}\t\t{u1_a[i]:.10f}\t\t{y_real_a[i]:.10f}\t\t{erro:.10f}")

erro_max_a = np.max(np.abs(u1_a - y_real_a))
print(f"\nErro máximo: {erro_max_a:.10f}")

# Letra B

print("LETRA B")

print("\nEDO: t^2*y'' - 2t*y' + 2y = t^3*ln(t)")
print("y(1) = 1, y'(1) = 0")
print(f"Intervalo: 1 <= t <= 2,  h = {h_b}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_b)):
    erro = abs(u1_b[i] - y_real_b[i])
    print(f"{t_b[i]:.2f}\t\t{u1_b[i]:.10f}\t\t{y_real_b[i]:.10f}\t\t{erro:.10f}")

erro_max_b = np.max(np.abs(u1_b - y_real_b))
print(f"\nErro máximo: {erro_max_b:.10f}")

# Letraa C

print("LETRA C")

print("\nEDO: y''' + 2y'' - y' - 2y = e^t")
print("y(0) = 1, y'(0) = 2, y''(0) = 0")
print(f"Intervalo: 0 <= t <= 3,  h = {h_c}")
print("\n" + "-"*100)
print("t\t\ty (aproximado)\t\ty (real)\t\tErro")
print("-"*100)
for i in range(len(t_c)):
    erro = abs(u1_c[i] - y_real_c[i])
    print(f"{t_c[i]:.2f}\t\t{u1_c[i]:.10f}\t\t{y_real_c[i]:.10f}\t\t{erro:.10f}")

erro_max_c = np.max(np.abs(u1_c - y_real_c))
print(f"\nErro máximo: {erro_max_c:.10f}")

# Letra D

print("LETRA D")

print("\nEDO: t^3*y''' - t^2*y'' + 3t*y' - 4y = 5t^3*ln(t) + 9t^3")
print("y(1) = 0, y'(1) = 1, y''(1) = 3")
print(f"Intervalo: 1 <= t <= 2,  h = {h_d}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_d)):
    erro = abs(u1_d[i] - y_real_d[i])
    print(f"{t_d[i]:.2f}\t\t{u1_d[i]:.10f}\t\t{y_real_d[i]:.10f}\t\t{erro:.10f}")

erro_max_d = np.max(np.abs(u1_d - y_real_d))
print(f"\nErro máximo: {erro_max_d:.10f}")

# GRÁFICOS

# GRÁFICO LETRA A 
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
plt.plot(t_a, u1_a, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_a, y_real_a, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra A - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

# GRÁFICO LETRA B
plt.subplot(2, 2, 2)
plt.plot(t_b, u1_b, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_b, y_real_b, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra B - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

# GRÁFICO LETRA C
plt.subplot(2, 2, 3)
plt.plot(t_c, u1_c, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_c, y_real_c, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra C - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

#GRÁFICO LETRA D 
plt.subplot(2, 2, 4)
plt.plot(t_d, u1_d, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_d, y_real_d, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra D - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

plt.suptitle('Questão 11 - Runge-Kutta para Sistemas de Ordem Superior', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# GRÁFICOS DOS ERROS

plt.figure(figsize=(14, 10))

# Erro Letra A
plt.subplot(2, 2, 1)
plt.plot(t_a, np.abs(u1_a - y_real_a), 'b-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra A - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra B
plt.subplot(2, 2, 2)
plt.plot(t_b, np.abs(u1_b - y_real_b), 'r-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra B - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra C
plt.subplot(2, 2, 3)
plt.plot(t_c, np.abs(u1_c - y_real_c), 'g-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra C - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra D
plt.subplot(2, 2, 4)
plt.plot(t_d, np.abs(u1_d - y_real_d), 'm-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra D - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

plt.suptitle('Questão 11 - Erros das Aproximações', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# RESUMO FINAL

print("\n" + "="*100)
print("RESUMO FINAL - ERROS MÁXIMOS")
print("="*100)

print(f"\nLetra A (y'' - 2y' + y = t*e^t - t):")
print(f"  Erro máximo: {erro_max_a:.10f}")

print(f"\nLetra B (t^2*y'' - 2t*y' + 2y = t^3*ln(t)):")
print(f"  Erro máximo: {erro_max_b:.10f}")

print(f"\nLetra C (y''' + 2y'' - y' - 2y = e^t):")
print(f"  Erro máximo: {erro_max_c:.10f}")

print(f"\nLetra D (t^3*y''' - t^2*y'' + 3t*y' - 4y = 5t^3*ln(t) + 9t^3):")
print(f"  Erro máximo: {erro_max_d:.10f}")

print("\n" + "="*100)
```
