Use o método de Euler para obter uma aproximação para as soluções de cada um dos seguintes problemas de valor inicial.

**a)** y′ = e^(t − y), 0 ≤ t ≤ 1, y(0) = 1, com h = 0,5

**b)** y′ = (1 + t)/(1 + y), 1 ≤ t ≤ 2, y(1) = 2, com h = 0,5

**c)** y′ = −y + t·y^(1/2), 2 ≤ t ≤ 3, y(2) = 2, com h = 0,25

**d)** y′ = t⁻² (sen(2t) − 2t·y), 1 ≤ t ≤ 2, y(1) = 2, com h = 0,25

```python
import numpy as np

def euler(f, t0, y0, tf, h, label):
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        y = y + h * f(t, y)
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

def f2a(t, y): return np.exp(t - y)
def f2b(t, y): return (1 + t) / (1 + y)
def f2c(t, y): return -y + t * np.sqrt(y)
def f2d(t, y): return (t**(-2)) * (np.sin(2*t) - 2*t*y)

dados = [
    (f2a, 0.0, 1.0, 1.0, 0.5, "2a) y' = e^(t-y), y(0)=1, h=0.5"),
    (f2b, 1.0, 2.0, 2.0, 0.5, "2b) y' = (1+t)/(1+y), y(1)=2, h=0.5"),
    (f2c, 2.0, 2.0, 3.0, 0.25, "2c) y' = -y + t√y, y(2)=2, h=0.25"),
    (f2d, 1.0, 2.0, 2.0, 0.25, "2d) y' = t^-2(sin2t - 2ty), y(1)=2, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    euler(f, t0, y0, tf, h, label)
```
