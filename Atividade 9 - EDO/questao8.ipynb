# Questão 8 - Método de Runge-Kutta de Quarta Ordem

Repita o Exercício 2 usando o método de Runge-Kutta de quarta ordem.

## Letra a)
$$y' = e^{t-y}, \quad 0 \leq t \leq 1, \quad y(0) = 1, \quad \text{com } h = 0,5$$

## Letra b)
$$y' = \frac{1+t}{1+y}, \quad 1 \leq t \leq 2, \quad y(1) = 2, \quad \text{com } h = 0,5$$

## Letra c)
$$y' = -y + ty^{1/2}, \quad 2 \leq t \leq 3, \quad y(2) = 2, \quad \text{com } h = 0,25$$

## Letra d)
$$y' = t^{-2}(\sin(2t) - 2ty), \quad 1 \leq t \leq 2, \quad y(1) = 2, \quad \text{com } h = 0,25$$

```python
import numpy as np
import matplotlib.pyplot as plt

def rk4_method(f, t0, y0, h, t_end):
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    y = np.zeros(n+1)
    t[0] = t0
    y[0] = y0
    
    for i in range(n):
        k1 = f(t[i], y[i])
        k2 = f(t[i] + h/2, y[i] + (h/2)*k1)
        k3 = f(t[i] + h/2, y[i] + (h/2)*k2)
        k4 = f(t[i] + h, y[i] + h*k3)
        
        y[i+1] = y[i] + (h/6)*(k1 + 2*k2 + 2*k3 + k4)
        t[i+1] = t[i] + h
    
    return t, y

#Letra A
def f8a(t, y):
    return np.exp(t - y)

t0, y0, h, t_end = 0, 1, 0.5, 1
t_a, y_a = rk4_method(f8a, t0, y0, h, t_end)

print("QUESTÃO 8 - RUNGE-KUTTA 4ª ORDEM (Questão 2)\n")

print("\nLETRA A")
print("t\t\ty (aproximado)")

for i in range(len(t_a)):
    print(f"{t_a[i]:.2f}\t\t{y_a[i]:.6f}")

#Letra B
def f8b(t, y):
    return (1 + t) / (1 + y)

t0, y0, h, t_end = 1, 2, 0.5, 2
t_b, y_b = rk4_method(f8b, t0, y0, h, t_end)

print("\n LETRA B ")

for i in range(len(t_b)):
    print(f"{t_b[i]:.2f}\t\t{y_b[i]:.6f}")

#Letra C
def f8c(t, y):
    return -y + t * np.sqrt(y)

t0, y0, h, t_end = 2, 2, 0.25, 3
t_c, y_c = rk4_method(f8c, t0, y0, h, t_end)

print("\n LETRA C ")
print("t\t\ty (aproximado)")

for i in range(len(t_c)):
    print(f"{t_c[i]:.2f}\t\t{y_c[i]:.6f}")

#Letra D
def f8d(t, y):
    return t**(-2) * (np.sin(2*t) - 2*t*y)

t0, y0, h, t_end = 1, 2, 0.25, 2
t_d, y_d = rk4_method(f8d, t0, y0, h, t_end)

print("\n LETRA D ")
print("t\t\ty (aproximado)")

for i in range(len(t_d)):
    print(f"{t_d[i]:.2f}\t\t{y_d[i]:.6f}")

# Gráficos
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(t_a, y_a, 'bo-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('8a - y\' = e^(t-y)')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(t_b, y_b, 'ro-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('8b - y\' = (1+t)/(1+y)')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(t_c, y_c, 'go-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('8c - y\' = -y + t*sqrt(y)')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(t_d, y_d, 'mo-', label='RK4')
plt.xlabel('t')
plt.ylabel('y')
plt.title('8d - y\' = t^(-2)(sin(2t) - 2ty)')
plt.grid(True)
plt.legend()

plt.suptitle('Questão 8 - Runge-Kutta 4ª Ordem', fontsize=16)
plt.tight_layout()
plt.show()
```
