Repita o Exercício 4 usando o método de Taylor de quarta ordem.

```python
import numpy as np
from sympy import symbols, exp, sqrt, sin, diff

t_sym, y_sym = symbols('t y', real=True)

f2a_sym = exp(t_sym - y_sym)
f2b_sym = (1 + t_sym) / (1 + y_sym)
f2c_sym = -y_sym + t_sym * sqrt(y_sym)
f2d_sym = (t_sym**(-2)) * (sin(2*t_sym) - 2*t_sym*y_sym)

def taylor4(f_expr, t0, y0, tf, h, label):
    f_derivs = [f_expr]
    for k in range(3):
        f_derivs.append(diff(f_derivs[-1], t_sym) + diff(f_derivs[-1], y_sym) * f_expr)
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        vals = [float(der.subs({t_sym: t, y_sym: y})) for der in f_derivs]
        y = y + h*vals[0] + (h**2/2)*vals[1] + (h**3/6)*vals[2] + (h**4/24)*vals[3]
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

dados = [
    (f2a_sym, 0.0, 1.0, 1.0, 0.5, "6a) y'=e^(t-y), y(0)=1, h=0.5"),
    (f2b_sym, 1.0, 2.0, 2.0, 0.5, "6b) y'=(1+t)/(1+y), y(1)=2, h=0.5"),
    (f2c_sym, 2.0, 2.0, 3.0, 0.25, "6c) y'=-y+t√y, y(2)=2, h=0.25"),
    (f2d_sym, 1.0, 2.0, 2.0, 0.25, "6d) y'=t^-2(sin2t-2ty), y(1)=2, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    taylor4(f, t0, y0, tf, h, label)
```
