# Questão 1

Use o método de Euler para obter uma aproximação para as soluções de cada um dos seguintes problemas de valor inicial.

**a)** y′ = t·e³ᵗ − 2y, 0 ≤ t ≤ 1, y(0) = 0, com h = 0,5

**b)** y′ = 1 + (t − y)², 2 ≤ t ≤ 3, y(2) = 1, com h = 0,5

**c)** y′ = 1 + y/t, 1 ≤ t ≤ 2, y(1) = 2, com h = 0,25

**d)** y′ = cos(2t) + sen(3t), 0 ≤ t ≤ 1, y(0) = 1, com h = 0,25

```python
import numpy as np

def euler(f, t0, y0, tf, h, label):
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        y = y + h * f(t, y)
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

# Definição das funções f(t,y)
def f1a(t, y): return t * np.exp(3*t) - 2*y
def f1b(t, y): return 1 + (t - y)**2
def f1c(t, y): return 1 + y/t
def f1d(t, y): return np.cos(2*t) + np.sin(3*t)

# Execução
dados = [
    (f1a, 0.0, 0.0, 1.0, 0.5, "1a) y' = t e^(3t) - 2y, y(0)=0, h=0.5"),
    (f1b, 2.0, 1.0, 3.0, 0.5, "1b) y' = 1+(t-y)^2, y(2)=1, h=0.5"),
    (f1c, 1.0, 2.0, 2.0, 0.25, "1c) y' = 1+y/t, y(1)=2, h=0.25"),
    (f1d, 0.0, 1.0, 1.0, 0.25, "1d) y' = cos2t+sin3t, y(0)=1, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    euler(f, t0, y0, tf, h, label)
```
