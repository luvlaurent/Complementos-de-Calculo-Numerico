Repita o Exercício 3 usando o método de Taylor de quarta ordem.

```python
import numpy as np
from sympy import symbols, exp, cos, sin, diff

t_sym, y_sym = symbols('t y', real=True)

f1a_sym = t_sym * exp(3*t_sym) - 2*y_sym
f1b_sym = 1 + (t_sym - y_sym)**2
f1c_sym = 1 + y_sym/t_sym
f1d_sym = cos(2*t_sym) + sin(3*t_sym)

def taylor4(f_expr, t0, y0, tf, h, label):
    # Calcula f, f', f'', f''' (derivada total)
    f_derivs = [f_expr]
    for k in range(3):  # precisamos até ordem 3 (f''')
        f_derivs.append(diff(f_derivs[-1], t_sym) + diff(f_derivs[-1], y_sym) * f_expr)
    # f_derivs[0]=f, [1]=f', [2]=f'', [3]=f'''
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        vals = [float(der.subs({t_sym: t, y_sym: y})) for der in f_derivs]
        y = y + h*vals[0] + (h**2/2)*vals[1] + (h**3/6)*vals[2] + (h**4/24)*vals[3]
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

dados = [
    (f1a_sym, 0.0, 0.0, 1.0, 0.5, "5a) y'=t e^(3t)-2y, y(0)=0, h=0.5"),
    (f1b_sym, 2.0, 1.0, 3.0, 0.5, "5b) y'=1+(t-y)^2, y(2)=1, h=0.5"),
    (f1c_sym, 1.0, 2.0, 2.0, 0.25, "5c) y'=1+y/t, y(1)=2, h=0.25"),
    (f1d_sym, 0.0, 1.0, 1.0, 0.25, "5d) y'=cos2t+sin3t, y(0)=1, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    taylor4(f, t0, y0, tf, h, label)
```
