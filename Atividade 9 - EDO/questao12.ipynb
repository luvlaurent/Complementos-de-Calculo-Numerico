# Questão 12 - Runge-Kutta para Sistemas de EDOs de Ordem Superior

Use o algoritmo de Runge-Kutta para sistemas para encontrar aproximações das soluções dos seguintes sistemas de equações diferenciais de ordem superior e compare os resultados às soluções reais.

---

## Letra a)

$$y'' - 3y' + 2y = 6e^{-t}, \quad 0 \leq t \leq 1, \quad y(0) = y'(0) = 2; \quad \text{com } h = 0,1$$

**Solução real:**

$$y(t) = 2e^{2t} - e^t + e^{-t}$$

---

## Letra b)

$$t^2y'' + ty' - 4y = -3t, \quad 1 \leq t \leq 3, \quad y(1) = 4, \quad y'(1) = 3; \quad \text{com } h = 0,2$$

**Solução real:**

$$y(t) = 2t^2 + t + t^{-2}$$

---

## Letra c)

$$y'''' + y'' - 4y' - 4y = 0, \quad 0 \leq t \leq 2, \quad y(0) = 3, \quad y'(0) = -1, \quad y''(0) = 9; \quad \text{com } h = 0,2$$

**Solução real:**

$$y(t) = e^{-t} + e^{2t} + e^{-2t}$$

---

## Letra d)

$$t^3y''' + t^2y'' - 2ty' + 2y = 8t^3 - 2, \quad 1 \leq t \leq 2, \quad y(1) = 2, \quad y'(1) = 8, \quad y''(1) = 6; \quad \text{com } h = 0,1$$

**Solução real:**

$$y(t) = 2t - t^{-1} + t^2 + t^3 - 1$$

## Transformação para Sistemas de Primeira Ordem

### Letra a)
Seja $u_1 = y$ e $u_2 = y'$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = 3u_2 - 2u_1 + 6e^{-t}
\end{cases}$$

com $u_1(0) = 2$ e $u_2(0) = 2$.

---

### Letra b)
Seja $u_1 = y$ e $u_2 = y'$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = -\frac{1}{t}u_2 + \frac{4}{t^2}u_1 - \frac{3}{t}
\end{cases}$$

com $u_1(1) = 4$ e $u_2(1) = 3$.

---

### Letra c)
Seja $u_1 = y$, $u_2 = y'$, $u_3 = y''$ e $u_4 = y'''$. Então:

$$\begin{cases}
u_1' = u_2 \\
u_2' = u_3 \\
u_3' = u_4 \\
u_4' = -u_3 + 4u_2 + 4u_1
\end{cases}$$

com $u_1(0) = 3$, $u_2(0) = -1$, $u_3(0) = 9$.

Portanto, $u_4(0) = -1$.

```python
import numpy as np
import matplotlib.pyplot as plt

def rk4_system_2d(f1, f2, t0, u1_0, u2_0, h, t_end):
    """
    Método de Runge-Kutta de 4ª ordem para sistemas de 2 EDOs de primeira ordem
    """
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    u1 = np.zeros(n+1)
    u2 = np.zeros(n+1)
    t[0] = t0
    u1[0] = u1_0
    u2[0] = u2_0
    
    for i in range(n):
        t_curr = t[i]
        u1_curr = u1[i]
        u2_curr = u2[i]
        
        k1_1 = f1(t_curr, u1_curr, u2_curr)
        k1_2 = f2(t_curr, u1_curr, u2_curr)
        
        k2_1 = f1(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2)
        k2_2 = f2(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2)
        
        k3_1 = f1(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2)
        k3_2 = f2(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2)
        
        k4_1 = f1(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2)
        k4_2 = f2(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2)
        
        u1[i+1] = u1_curr + (h/6)*(k1_1 + 2*k2_1 + 2*k3_1 + k4_1)
        u2[i+1] = u2_curr + (h/6)*(k1_2 + 2*k2_2 + 2*k3_2 + k4_2)
        t[i+1] = t_curr + h
    
    return t, u1, u2


def rk4_system_3d(f1, f2, f3, t0, u1_0, u2_0, u3_0, h, t_end):
    """
    Método de Runge-Kutta de 4ª ordem para sistemas de 3 EDOs de primeira ordem
    """
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    u1 = np.zeros(n+1)
    u2 = np.zeros(n+1)
    u3 = np.zeros(n+1)
    t[0] = t0
    u1[0] = u1_0
    u2[0] = u2_0
    u3[0] = u3_0
    
    for i in range(n):
        t_curr = t[i]
        u1_curr = u1[i]
        u2_curr = u2[i]
        u3_curr = u3[i]
        
        k1_1 = f1(t_curr, u1_curr, u2_curr, u3_curr)
        k1_2 = f2(t_curr, u1_curr, u2_curr, u3_curr)
        k1_3 = f3(t_curr, u1_curr, u2_curr, u3_curr)
        
        k2_1 = f1(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        k2_2 = f2(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        k2_3 = f3(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, u3_curr + (h/2)*k1_3)
        
        k3_1 = f1(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        k3_2 = f2(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        k3_3 = f3(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, u3_curr + (h/2)*k2_3)
        
        k4_1 = f1(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        k4_2 = f2(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        k4_3 = f3(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, u3_curr + h*k3_3)
        
        u1[i+1] = u1_curr + (h/6)*(k1_1 + 2*k2_1 + 2*k3_1 + k4_1)
        u2[i+1] = u2_curr + (h/6)*(k1_2 + 2*k2_2 + 2*k3_2 + k4_2)
        u3[i+1] = u3_curr + (h/6)*(k1_3 + 2*k2_3 + 2*k3_3 + k4_3)
        t[i+1] = t_curr + h
    
    return t, u1, u2, u3


def rk4_system_4d(f1, f2, f3, f4, t0, u1_0, u2_0, u3_0, u4_0, h, t_end):
    """
    Método de Runge-Kutta de 4ª ordem para sistemas de 4 EDOs de primeira ordem
    """
    n = int((t_end - t0) / h)
    t = np.zeros(n+1)
    u1 = np.zeros(n+1)
    u2 = np.zeros(n+1)
    u3 = np.zeros(n+1)
    u4 = np.zeros(n+1)
    t[0] = t0
    u1[0] = u1_0
    u2[0] = u2_0
    u3[0] = u3_0
    u4[0] = u4_0
    
    for i in range(n):
        t_curr = t[i]
        u1_curr = u1[i]
        u2_curr = u2[i]
        u3_curr = u3[i]
        u4_curr = u4[i]
        
        k1_1 = f1(t_curr, u1_curr, u2_curr, u3_curr, u4_curr)
        k1_2 = f2(t_curr, u1_curr, u2_curr, u3_curr, u4_curr)
        k1_3 = f3(t_curr, u1_curr, u2_curr, u3_curr, u4_curr)
        k1_4 = f4(t_curr, u1_curr, u2_curr, u3_curr, u4_curr)
        
        k2_1 = f1(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, 
                  u3_curr + (h/2)*k1_3, u4_curr + (h/2)*k1_4)
        k2_2 = f2(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, 
                  u3_curr + (h/2)*k1_3, u4_curr + (h/2)*k1_4)
        k2_3 = f3(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, 
                  u3_curr + (h/2)*k1_3, u4_curr + (h/2)*k1_4)
        k2_4 = f4(t_curr + h/2, u1_curr + (h/2)*k1_1, u2_curr + (h/2)*k1_2, 
                  u3_curr + (h/2)*k1_3, u4_curr + (h/2)*k1_4)
        
        k3_1 = f1(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, 
                  u3_curr + (h/2)*k2_3, u4_curr + (h/2)*k2_4)
        k3_2 = f2(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, 
                  u3_curr + (h/2)*k2_3, u4_curr + (h/2)*k2_4)
        k3_3 = f3(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, 
                  u3_curr + (h/2)*k2_3, u4_curr + (h/2)*k2_4)
        k3_4 = f4(t_curr + h/2, u1_curr + (h/2)*k2_1, u2_curr + (h/2)*k2_2, 
                  u3_curr + (h/2)*k2_3, u4_curr + (h/2)*k2_4)
        
        k4_1 = f1(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, 
                  u3_curr + h*k3_3, u4_curr + h*k3_4)
        k4_2 = f2(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, 
                  u3_curr + h*k3_3, u4_curr + h*k3_4)
        k4_3 = f3(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, 
                  u3_curr + h*k3_3, u4_curr + h*k3_4)
        k4_4 = f4(t_curr + h, u1_curr + h*k3_1, u2_curr + h*k3_2, 
                  u3_curr + h*k3_3, u4_curr + h*k3_4)
        
        u1[i+1] = u1_curr + (h/6)*(k1_1 + 2*k2_1 + 2*k3_1 + k4_1)
        u2[i+1] = u2_curr + (h/6)*(k1_2 + 2*k2_2 + 2*k3_2 + k4_2)
        u3[i+1] = u3_curr + (h/6)*(k1_3 + 2*k2_3 + 2*k3_3 + k4_3)
        u4[i+1] = u4_curr + (h/6)*(k1_4 + 2*k2_4 + 2*k3_4 + k4_4)
        t[i+1] = t_curr + h
    
    return t, u1, u2, u3, u4

# LETRA A
# y'' - 3y' + 2y = 6e^(-t)
# y(0) = 2, y'(0) = 2
# 
# Transformação: u1 = y, u2 = y'
# u1' = u2
# u2' = 3u2 - 2u1 + 6e^(-t)

def f12a1(t, u1, u2):
    """u1' = u2"""
    return u2

def f12a2(t, u1, u2):
    """u2' = 3u2 - 2u1 + 6e^(-t)"""
    return 3*u2 - 2*u1 + 6*np.exp(-t)

def sol12a(t):
    """Solução real: y(t) = 2e^(2t) - e^t + e^(-t)"""
    return 2*np.exp(2*t) - np.exp(t) + np.exp(-t)

# Parâmetros da Letra A
t0_a, u1_0_a, u2_0_a, h_a, t_end_a = 0, 2, 2, 0.1, 1

# Executar RK4
t_a, u1_a, u2_a = rk4_system_2d(f12a1, f12a2, t0_a, u1_0_a, u2_0_a, h_a, t_end_a)

# Solução real
y_real_a = sol12a(t_a)

# LETRA B
# t^2*y'' + t*y' - 4y = -3t
# y(1) = 4, y'(1) = 3
# 
# Transformação: u1 = y, u2 = y'
# u1' = u2
# u2' = -(1/t)*u2 + (4/t^2)*u1 - 3/t


def f12b1(t, u1, u2):
    """u1' = u2"""
    return u2

def f12b2(t, u1, u2):
    """u2' = -(1/t)*u2 + (4/t^2)*u1 - 3/t"""
    return -(1/t)*u2 + (4/t**2)*u1 - 3/t

def sol12b(t):
    """Solução real: y(t) = 2t^2 + t + t^(-2)"""
    return 2*t**2 + t + t**(-2)

# Parâmetros da Letra B
t0_b, u1_0_b, u2_0_b, h_b, t_end_b = 1, 4, 3, 0.2, 3

# Executar RK4
t_b, u1_b, u2_b = rk4_system_2d(f12b1, f12b2, t0_b, u1_0_b, u2_0_b, h_b, t_end_b)

# Solução real
y_real_b = sol12b(t_b)


# LETRA C
# y'''' + y'' - 4y' - 4y = 0
# y(0) = 3, y'(0) = -1, y''(0) = 9, y'''(0) = ?
# 
# Transformação: u1 = y, u2 = y', u3 = y'', u4 = y'''
# u1' = u2
# u2' = u3
# u3' = u4
# u4' = -u3 + 4u2 + 4u1
# 
# Observação: Precisamos de y'''(0). Da EDO: y'''' = -y'' + 4y' + 4y
# Em t=0: y''''(0) = -9 + 4*(-1) + 4*3 = -9 - 4 + 12 = -1


def f12c1(t, u1, u2, u3, u4):
    """u1' = u2"""
    return u2

def f12c2(t, u1, u2, u3, u4):
    """u2' = u3"""
    return u3

def f12c3(t, u1, u2, u3, u4):
    """u3' = u4"""
    return u4

def f12c4(t, u1, u2, u3, u4):
    """u4' = -u3 + 4u2 + 4u1"""
    return -u3 + 4*u2 + 4*u1

def sol12c(t):
    """Solução real: y(t) = e^(-t) + e^(2t) + e^(-2t)"""
    return np.exp(-t) + np.exp(2*t) + np.exp(-2*t)

# Parâmetros da Letra C
# y'''(0) = -1 (calculado acima)
t0_c, u1_0_c, u2_0_c, u3_0_c, u4_0_c, h_c, t_end_c = 0, 3, -1, 9, -1, 0.2, 2

# Executar RK4
t_c, u1_c, u2_c, u3_c, u4_c = rk4_system_4d(
    f12c1, f12c2, f12c3, f12c4,
    t0_c, u1_0_c, u2_0_c, u3_0_c, u4_0_c,
    h_c, t_end_c
)

# Solução real
y_real_c = sol12c(t_c)

# LETRA D
# t^3*y''' + t^2*y'' - 2t*y' + 2y = 8t^3 - 2
# y(1) = 2, y'(1) = 8, y''(1) = 6
# 
# Transformação: u1 = y, u2 = y', u3 = y''
# u1' = u2
# u2' = u3
# u3' = -(1/t)*u3 + (2/t^2)*u2 - (2/t^3)*u1 + 8 - 2/t^3


def f12d1(t, u1, u2, u3):
    """u1' = u2"""
    return u2

def f12d2(t, u1, u2, u3):
    """u2' = u3"""
    return u3

def f12d3(t, u1, u2, u3):
    """u3' = -(1/t)*u3 + (2/t^2)*u2 - (2/t^3)*u1 + 8 - 2/t^3"""
    return -(1/t)*u3 + (2/t**2)*u2 - (2/t**3)*u1 + 8 - 2/t**3

def sol12d(t):
    """Solução real: y(t) = 2t - t^(-1) + t^2 + t^3 - 1"""
    return 2*t - t**(-1) + t**2 + t**3 - 1

# Parâmetros da Letra D
t0_d, u1_0_d, u2_0_d, u3_0_d, h_d, t_end_d = 1, 2, 8, 6, 0.1, 2

# Executar RK4
t_d, u1_d, u2_d, u3_d = rk4_system_3d(
    f12d1, f12d2, f12d3,
    t0_d, u1_0_d, u2_0_d, u3_0_d,
    h_d, t_end_d
)

# Solução real
y_real_d = sol12d(t_d)

# IMPRESSÃO DOS RESULTADOS

print("QUESTÃO 12 - RUNGE-KUTTA PARA SISTEMAS DE ORDEM SUPERIOR")

#  Letra A

print("LETRA A")

print("\nEDO: y'' - 3y' + 2y = 6e^(-t)")
print("y(0) = 2, y'(0) = 2")
print(f"Intervalo: 0 <= t <= 1,  h = {h_a}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_a)):
    erro = abs(u1_a[i] - y_real_a[i])
    print(f"{t_a[i]:.2f}\t\t{u1_a[i]:.10f}\t\t{y_real_a[i]:.10f}\t\t{erro:.10f}")

erro_max_a = np.max(np.abs(u1_a - y_real_a))
print(f"\nErro máximo: {erro_max_a:.10f}")

# Letra B 

print("LETRA B")

print("\nEDO: t^2*y'' + t*y' - 4y = -3t")
print("y(1) = 4, y'(1) = 3")
print(f"Intervalo: 1 <= t <= 3,  h = {h_b}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_b)):
    erro = abs(u1_b[i] - y_real_b[i])
    print(f"{t_b[i]:.2f}\t\t{u1_b[i]:.10f}\t\t{y_real_b[i]:.10f}\t\t{erro:.10f}")

erro_max_b = np.max(np.abs(u1_b - y_real_b))
print(f"\nErro máximo: {erro_max_b:.10f}")

# Letra C 

print("LETRA C")

print("\nEDO: y'''' + y'' - 4y' - 4y = 0")
print("y(0) = 3, y'(0) = -1, y''(0) = 9")
print(f"y'''(0) = -1 (calculado a partir da EDO)")
print(f"Intervalo: 0 <= t <= 2,  h = {h_c}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_c)):
    erro = abs(u1_c[i] - y_real_c[i])
    print(f"{t_c[i]:.2f}\t\t{u1_c[i]:.10f}\t\t{y_real_c[i]:.10f}\t\t{erro:.10f}")

erro_max_c = np.max(np.abs(u1_c - y_real_c))
print(f"\nErro máximo: {erro_max_c:.10f}")

# Letra D

print("LETRA D")

print("\nEDO: t^3*y''' + t^2*y'' - 2t*y' + 2y = 8t^3 - 2")
print("y(1) = 2, y'(1) = 8, y''(1) = 6")
print(f"Intervalo: 1 <= t <= 2,  h = {h_d}")

print("t\t\ty (aproximado)\t\ty (real)\t\tErro")

for i in range(len(t_d)):
    erro = abs(u1_d[i] - y_real_d[i])
    print(f"{t_d[i]:.2f}\t\t{u1_d[i]:.10f}\t\t{y_real_d[i]:.10f}\t\t{erro:.10f}")

erro_max_d = np.max(np.abs(u1_d - y_real_d))
print(f"\nErro máximo: {erro_max_d:.10f}")

# GRÁFICOS

#GRÁFICO LETRA A 
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
plt.plot(t_a, u1_a, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_a, y_real_a, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra A - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

# GRÁFICO LETRA B 
plt.subplot(2, 2, 2)
plt.plot(t_b, u1_b, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_b, y_real_b, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra B - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

# GRÁFICO LETRA C
plt.subplot(2, 2, 3)
plt.plot(t_c, u1_c, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_c, y_real_c, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra C - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

#GRÁFICO LETRA D 
plt.subplot(2, 2, 4)
plt.plot(t_d, u1_d, 'bo-', linewidth=2, markersize=6, label='RK4 - y aprox')
plt.plot(t_d, y_real_d, 'r--', linewidth=2, label='y real')
plt.xlabel('t', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Letra D - y(t)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)

plt.suptitle('Questão 12 - Runge-Kutta para Sistemas de Ordem Superior', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# GRÁFICOS DOS ERROS

plt.figure(figsize=(14, 10))

# Erro Letra A
plt.subplot(2, 2, 1)
plt.plot(t_a, np.abs(u1_a - y_real_a), 'b-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra A - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra B
plt.subplot(2, 2, 2)
plt.plot(t_b, np.abs(u1_b - y_real_b), 'r-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra B - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra C
plt.subplot(2, 2, 3)
plt.plot(t_c, np.abs(u1_c - y_real_c), 'g-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra C - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

# Erro Letra D
plt.subplot(2, 2, 4)
plt.plot(t_d, np.abs(u1_d - y_real_d), 'm-', linewidth=2)
plt.xlabel('t', fontsize=12)
plt.ylabel('Erro Absoluto', fontsize=12)
plt.title('Letra D - Erro', fontsize=14)
plt.grid(True, alpha=0.3)

plt.suptitle('Questão 12 - Erros das Aproximações', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# RESUMO FINAL

print("RESUMO FINAL - ERROS MÁXIMOS")

print(f"\nLetra A (y'' - 3y' + 2y = 6e^(-t)):")
print(f"  Erro máximo: {erro_max_a:.10f}")

print(f"\nLetra B (t^2*y'' + t*y' - 4y = -3t):")
print(f"  Erro máximo: {erro_max_b:.10f}")

print(f"\nLetra C (y'''' + y'' - 4y' - 4y = 0):")
print(f"  Erro máximo: {erro_max_c:.10f}")

print(f"\nLetra D (t^3*y''' + t^2*y'' - 2t*y' + 2y = 8t^3 - 2):")
print(f"  Erro máximo: {erro_max_d:.10f}")
```
