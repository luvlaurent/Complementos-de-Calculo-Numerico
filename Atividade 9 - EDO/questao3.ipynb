# Questão 3

Use o método de Taylor de segunda ordem para encontrar uma aproximação das soluções de cada um dos seguintes problemas de valor inicial.

**a)** y′ = t·e³ᵗ − 2y, 0 ≤ t ≤ 1, y(0) = 0, com h = 0,5

**b)** y′ = 1 + (t − y)², 2 ≤ t ≤ 3, y(2) = 1, com h = 0,5

**c)** y′ = 1 + y/t, 1 ≤ t ≤ 2, y(1) = 2, com h = 0,25

**d)** y′ = cos(2t) + sen(3t), 0 ≤ t ≤ 1, y(0) = 1, com h = 0,25

```python
import numpy as np
from sympy import symbols, exp, cos, sin, diff

t_sym, y_sym = symbols('t y', real=True)

# Definição simbólica das f's
f1a_sym = t_sym * exp(3*t_sym) - 2*y_sym
f1b_sym = 1 + (t_sym - y_sym)**2
f1c_sym = 1 + y_sym/t_sym
f1d_sym = cos(2*t_sym) + sin(3*t_sym)

def taylor2(f_expr, t0, y0, tf, h, label):
    # Calcula derivada total: f' = df/dt + df/dy * f
    f_prime = diff(f_expr, t_sym) + diff(f_expr, y_sym) * f_expr
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        f_val = float(f_expr.subs({t_sym: t, y_sym: y}))
        fp_val = float(f_prime.subs({t_sym: t, y_sym: y}))
        y = y + h * f_val + (h**2 / 2) * fp_val
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

dados = [
    (f1a_sym, 0.0, 0.0, 1.0, 0.5, "3a) y'=t e^(3t)-2y, y(0)=0, h=0.5"),
    (f1b_sym, 2.0, 1.0, 3.0, 0.5, "3b) y'=1+(t-y)^2, y(2)=1, h=0.5"),
    (f1c_sym, 1.0, 2.0, 2.0, 0.25, "3c) y'=1+y/t, y(1)=2, h=0.25"),
    (f1d_sym, 0.0, 1.0, 1.0, 0.25, "3d) y'=cos2t+sin3t, y(0)=1, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    taylor2(f, t0, y0, tf, h, label)
```
