Use o método de Taylor de segunda ordem para encontrar uma aproximação das soluções de cada um dos seguintes problemas de valor inicial.

**a)** y′ = e^(t − y), 0 ≤ t ≤ 1, y(0) = 1, com h = 0,5

**b)** y′ = (1 + t)/(1 + y), 1 ≤ t ≤ 2, y(1) = 2, com h = 0,5

**c)** y′ = −y + t·y^(1/2), 2 ≤ t ≤ 3, y(2) = 2, com h = 0,25

**d)** y′ = t⁻² (sen(2t) − 2t·y), 1 ≤ t ≤ 2, y(1) = 2, com h = 0,25

```python
import numpy as np
from sympy import symbols, exp, sqrt, sin, diff

t_sym, y_sym = symbols('t y', real=True)

f2a_sym = exp(t_sym - y_sym)
f2b_sym = (1 + t_sym) / (1 + y_sym)
f2c_sym = -y_sym + t_sym * sqrt(y_sym)
f2d_sym = (t_sym**(-2)) * (sin(2*t_sym) - 2*t_sym*y_sym)

def taylor2(f_expr, t0, y0, tf, h, label):
    f_prime = diff(f_expr, t_sym) + diff(f_expr, y_sym) * f_expr
    n = int(round((tf - t0) / h))
    t = t0
    y = y0
    print(f"\n--- {label} ---")
    print(f"t_0 = {t:.4f}, y_0 = {y:.10f}")
    for i in range(1, n+1):
        f_val = float(f_expr.subs({t_sym: t, y_sym: y}))
        fp_val = float(f_prime.subs({t_sym: t, y_sym: y}))
        y = y + h * f_val + (h**2 / 2) * fp_val
        t = t + h
        print(f"t_{i} = {t:.4f}, y_{i} = {y:.10f}")

dados = [
    (f2a_sym, 0.0, 1.0, 1.0, 0.5, "4a) y'=e^(t-y), y(0)=1, h=0.5"),
    (f2b_sym, 1.0, 2.0, 2.0, 0.5, "4b) y'=(1+t)/(1+y), y(1)=2, h=0.5"),
    (f2c_sym, 2.0, 2.0, 3.0, 0.25, "4c) y'=-y+t√y, y(2)=2, h=0.25"),
    (f2d_sym, 1.0, 2.0, 2.0, 0.25, "4d) y'=t^-2(sin2t-2ty), y(1)=2, h=0.25")
]

for f, t0, y0, tf, h, label in dados:
    taylor2(f, t0, y0, tf, h, label)
```
