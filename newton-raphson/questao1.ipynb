# 1. Seja f(x) = x^2 - 6 e p0 = 1.
# Utilize o método de Newton para determinar p2.

```python
import math

# função f(x)
def f(x):
    return x**2 - 6

# derivada f'(x)
def df(x):
    return 2*x

# valor inicial
p = 1

print("Iteração | Valor de p")

# vamos calcular até p2
for i in range(2):
    p_new = p - f(p)/df(p)  # calcula próximo valor
    print(i+1, "      |", p_new)
    p = p_new               # atualiza p
```
