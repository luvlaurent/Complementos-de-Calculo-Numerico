# Questão 23

Obtenha uma aproximação de \(\int_1^2 x^2 \ln(x^2 + 1) \, dx\) usando \(h = 0,25\). Use a:

**a)** regra do trapézio composta

**b)** regra de Simpson composta

**c)** regra do ponto médio composta

```python

import math

def f23(x):
    return x**2 * math.log(x**2 + 1)

a, b = 1, 2
h = 0.25
n = int((b - a) / h)  # n = 4

# Valores de x e f(x)
x_vals = [a + i*h for i in range(n+1)]
f_vals = [f23(x) for x in x_vals]

print(f"Intervalo: [{a}, {b}]")
print(f"h = {h}")
print(f"n = {n}")
print("\nTabela de pontos:")
print("   x    |    f(x)")
print("-" * 25)
for i in range(len(x_vals)):
    print(f"  {x_vals[i]:.2f}  |  {f_vals[i]:.8f}")

# a) Regra do trapézio composta

print("\n--- a) Trapézio Composta ---")
trap23 = (h/2) * (f_vals[0] + f_vals[-1] + 2*sum(f_vals[1:-1]))
print(f"Aproximação: {trap23:.8f}")

# Valor exato (aproximado com alta precisão)
def integracao_alta(f, a, b, n=10000):
    hh = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * hh)
    return (hh / 2) * soma
exato23 = integracao_alta(f23, 1, 2, 100000)
print(f"Valor exato (referência): {exato23:.8f}")
print(f"Erro: {abs(exato23 - trap23):.2e}")

# b) Regra de Simpson composta

print("\n--- b) Simpson Composta ---")
if n % 2 == 0:
    simp23 = (h/3) * (f_vals[0] + f_vals[-1] + 4*sum(f_vals[1:-1:2]) + 2*sum(f_vals[2:-1:2]))
    print(f"Aproximação: {simp23:.8f}")
    print(f"Erro: {abs(exato23 - simp23):.2e}")
else:
    print("n não é par para Simpson!")

# c) Regra do ponto médio composta

print("\n--- c) Ponto Médio Composta ---")
pm23 = 0
for i in range(n):
    x_mid = a + (i + 0.5) * h
    pm23 += f23(x_mid)
pm23 = pm23 * h
print(f"Aproximação: {pm23:.8f}")
print(f"Erro: {abs(exato23 - pm23):.2e}")
```
