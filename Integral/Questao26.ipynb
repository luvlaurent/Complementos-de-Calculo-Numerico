# Questão 26

A regra do ponto médio para obter uma aproximação de \(\int_{-1}^1 f(x) \, dx\) resulta em \(12\), a regra do ponto médio composta com \(n = 2\) resulta em \(5\) e a regra de Simpson composta resulta em \(6\). Utilize o fato de que \(f(-1) = f(1)\) e \(f(-0,5) = f(0,5) - 1\) para determinar \(f(-1)\), \(f(-0,5)\), \(f(0)\), \(f(0,5)\) e \(f(1)\).
