# Questão 26

A regra do ponto médio para obter uma aproximação de \(\int_{-1}^1 f(x) \, dx\) resulta em \(12\), a regra do ponto médio composta com \(n = 2\) resulta em \(5\) e a regra de Simpson composta resulta em \(6\). Utilize o fato de que \(f(-1) = f(1)\) e \(f(-0,5) = f(0,5) - 1\) para determinar \(f(-1)\), \(f(-0,5)\), \(f(0)\), \(f(0,5)\) e \(f(1)\).

```python

import math

print("""
Dados:
- Regra do ponto médio simples (n=1) para ∫₋₁¹ f(x) dx resulta em 12
- Regra do ponto médio composta com n=2 resulta em 5
- Regra de Simpson composta resulta em 6
- f(-1) = f(1)
- f(-0.5) = f(0.5) - 1

Precisamos encontrar: f(-1), f(-0.5), f(0), f(0.5), f(1)
""")

# Seja: a = f(-1) = f(1), b = f(-0.5), c = f(0), d = f(0.5)
# Dado: b = d - 1  →  d = b + 1

print("\nDefinindo variáveis:")
print("a = f(-1) = f(1)")
print("b = f(-0.5)")
print("c = f(0)")
print("d = f(0.5)")
print("Relação: d = b + 1")

# Regra do ponto médio simples (n=1): h = 2, ponto médio = 0
# I ≈ 2 * f(0) = 2c = 12 → c = 6
print("\n1) Ponto médio simples (n=1):")
print("   h = 2, ponto médio = 0")
print("   I ≈ 2 * f(0) = 2c = 12")
print("   → c = 6")

# Regra do ponto médio composta (n=2): h = 1, pontos médios: -0.5 e 0.5
# I ≈ 1 * [f(-0.5) + f(0.5)] = b + d = b + (b+1) = 2b + 1 = 5
print("\n2) Ponto médio composta (n=2):")
print("   h = 1")
print("   I ≈ 1 * [f(-0.5) + f(0.5)] = b + d = b + (b+1) = 2b + 1")
print("   2b + 1 = 5")
print("   → 2b = 4")
print("   → b = 2")
print("   → d = b + 1 = 3")

# Regra de Simpson composta (n=2): h = 1, pontos: x0=-1, x1=0, x2=1
# I ≈ (h/3)[f(-1) + 4f(0) + f(1)] = (1/3)[a + 4c + a] = (1/3)[2a + 4c] = (2a/3) + (4c/3)
# 2a/3 + 4c/3 = 6
print("\n3) Simpson composta (n=2):")
print("   h = 1")
print("   I ≈ (1/3)[f(-1) + 4f(0) + f(1)] = (1/3)[a + 4*6 + a]")
print("   = (1/3)[2a + 24] = (2a/3) + 8")
print("   (2a/3) + 8 = 6")
print("   (2a/3) = -2")
print("   → 2a = -6")
print("   → a = -3")

print("\n" + "=" * 50)
print("RESULTADOS:")
print(f"f(-1) = {(-3)}")
print(f"f(-0.5) = {2}")
print(f"f(0) = {6}")
print(f"f(0.5) = {3}")
print(f"f(1) = {(-3)}")
```
