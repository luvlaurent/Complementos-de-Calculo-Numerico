Suponha que \(f(0) = 1\), \(f(0,5) = 2,5\), \(f(1) = 2\) e \(f(0,25) = f(0,75) = \alpha\). Encontre \(\alpha\) se a regra do trapézio composta com \(n = 4\) resultar o valor \(1,75\) para \(\int_0^1 f(x) \, dx\).

```python

import math


print("""
Dados: f(0) = 1, f(0.5) = 2.5, f(1) = 2
       f(0.25) = f(0.75) = α
       Regra do trapézio composta com n = 4 resulta 1.75 para ∫₀¹ f(x) dx

Precisamos encontrar α.
""")

# Com n = 4, h = 0.25
# Pontos: x0=0, x1=0.25, x2=0.5, x3=0.75, x4=1
# f0=1, f1=α, f2=2.5, f3=α, f4=2

h = 0.25
f0 = 1
f1_alpha = "α"
f2 = 2.5
f3_alpha = "α"
f4 = 2

print("\nFórmula do trapézio composta com n=4:")
print("I = (h/2)[f0 + 2f1 + 2f2 + 2f3 + f4]")
print(f"I = ({h}/2)[{f0} + 2α + 2*{f2} + 2α + {f4}]")
print(f"I = {h/2}[{f0} + {f4} + 2*{f2} + 4α]")
print(f"I = {h/2}[1 + 2 + 5 + 4α]")
print(f"I = {h/2}[8 + 4α]")
print(f"I = {h/2} * 4 * [2 + α]")
print(f"I = {h} * 2 * [2 + α]")
print(f"I = {0.25*2} * [2 + α]")
print(f"I = 0.5 * [2 + α]")

print("\nSabemos que I = 1.75")
print("0.5 * [2 + α] = 1.75")
print("2 + α = 3.5")
print(f"α = 1.5")

print("\n" + "=" * 50)
print(f"RESPOSTA: α = 1.5")
```
