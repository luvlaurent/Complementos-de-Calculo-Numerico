# Questão 21

Utilize a regra do ponto médio composta com \(n + 2\) subintervalos para obter uma aproximação das integrais no Exercício 1.

```python

import math

def ponto_medio_composta(f, a, b, n):
    """Regra do ponto médio composta"""
    h = (b - a) / n
    soma = 0
    for i in range(n):
        xi = a + (i + 0.5) * h
        soma += f(xi)
    return h * soma

# A questão pede "n + 2 subintervalos"
# Vamos usar n = 2 como base (já que as integrais da questão 1 são simples)

print("\n--- Questão 1a) ∫₀.₅¹ x⁴ dx ---")
def f21a(x):
    return x**4
exato21a = 1/5 - (0.5)**5/5
n_base = 2
n_uso = n_base + 2
aprox21a = ponto_medio_composta(f21a, 0.5, 1, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21a:.8f}")
print(f"Exato: {exato21a:.8f}")
print(f"Erro: {abs(exato21a - aprox21a):.2e}")

print("\n--- Questão 1b) ∫₀⁰·⁵ 2/(x-4) dx ---")
def f21b(x):
    return 2/(x-4)
exato21b = 2 * math.log(3.5/4)
aprox21b = ponto_medio_composta(f21b, 0, 0.5, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21b:.8f}")
print(f"Exato: {exato21b:.8f}")
print(f"Erro: {abs(exato21b - aprox21b):.2e}")

print("\n--- Questão 1c) ∫₁¹·⁵ x² ln x dx ---")
def f21c(x):
    return x**2 * math.log(x)
def antiderivada21c(x):
    return (x**3/3) * math.log(x) - (x**3/9)
exato21c = antiderivada21c(1.5) - antiderivada21c(1)
aprox21c = ponto_medio_composta(f21c, 1, 1.5, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21c:.8f}")
print(f"Exato: {exato21c:.8f}")
print(f"Erro: {abs(exato21c - aprox21c):.2e}")

print("\n--- Questão 1d) ∫₀¹ x² e^{-x} dx ---")
def f21d(x):
    return x**2 * math.exp(-x)
def antiderivada21d(x):
    return -math.exp(-x) * (x**2 + 2*x + 2)
exato21d = antiderivada21d(1) - antiderivada21d(0)
aprox21d = ponto_medio_composta(f21d, 0, 1, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21d:.8f}")
print(f"Exato: {exato21d:.8f}")
print(f"Erro: {abs(exato21d - aprox21d):.2e}")

print("\n--- Questão 1e) ∫₁¹·⁶ 2x/(x²-4) dx ---")
def f21e(x):
    return 2*x/(x**2 - 4)
exato21e = math.log((1.6**2 - 4)/(1**2 - 4))
aprox21e = ponto_medio_composta(f21e, 1, 1.6, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21e:.8f}")
print(f"Exato: {exato21e:.8f}")
print(f"Erro: {abs(exato21e - aprox21e):.2e}")

print("\n--- Questão 1f) ∫₀⁰·³⁵ 2/(x²-4) dx ---")
def f21f(x):
    return 2/(x**2 - 4)
def antiderivada21f(x):
    return (1/2) * math.log(abs((x-2)/(x+2)))
exato21f = antiderivada21f(0.35) - antiderivada21f(0)
aprox21f = ponto_medio_composta(f21f, 0, 0.35, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21f:.8f}")
print(f"Exato: {exato21f:.8f}")
print(f"Erro: {abs(exato21f - aprox21f):.2e}")

print("\n--- Questão 1g) ∫₀^{π/4} x sen x dx ---")
def f21g(x):
    return x * math.sin(x)
def antiderivada21g(x):
    return math.sin(x) - x*math.cos(x)
exato21g = antiderivada21g(math.pi/4) - antiderivada21g(0)
aprox21g = ponto_medio_composta(f21g, 0, math.pi/4, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21g:.8f}")
print(f"Exato: {exato21g:.8f}")
print(f"Erro: {abs(exato21g - aprox21g):.2e}")

print("\n--- Questão 1h) ∫₀³ e^{3x} sen(2x) dx ---")
def f21h(x):
    return math.exp(3*x) * math.sin(2*x)
def antiderivada21h(x):
    return math.exp(3*x) * (3*math.sin(2*x) - 2*math.cos(2*x)) / 13
exato21h = antiderivada21h(3) - antiderivada21h(0)
aprox21h = ponto_medio_composta(f21h, 0, 3, n_uso)
print(f"Aproximação (n={n_uso}): {aprox21h:.8f}")
print(f"Exato: {exato21h:.8f}")
print(f"Erro: {abs(exato21h - aprox21h):.2e}")
```
