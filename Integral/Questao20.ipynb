Utilize a regra de Simpson composta para obter uma aproximação das integrais no Exercício 18.

```python
# Questão 20 - Regra de Simpson composta (usando as integrais da 18)
import math

def simpson_composta(f, a, b, n):
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(a + i * h)
        else:
            soma += 4 * f(a + i * h)
    return (h / 3) * soma

# 18a) ∫₋₀.₅⁰.⁵ cos² x dx
print("\n--- 18a (Simpson) ---")
def f20a(x):
    return math.cos(x)**2
def antiderivada20a(x):
    return (x/2) + math.sin(2*x)/4
exato20a = antiderivada20a(0.5) - antiderivada20a(-0.5)
aprox20a = simpson_composta(f20a, -0.5, 0.5, 4)
print(f"Aproximação (n=4): {aprox20a:.8f}")
print(f"Erro: {abs(exato20a - aprox20a):.2e}")

# 18b) ∫₋₀.₅⁰.⁵ x ln(x+1) dx
print("\n--- 18b (Simpson) ---")
def f20b(x):
    return x * math.log(x + 1)
def antiderivada20b(x):
    return (x**2/2) * math.log(x+1) - (x**2/4) + (x/2) - (1/2)*math.log(x+1)
exato20b = antiderivada20b(0.5) - antiderivada20b(-0.5)
aprox20b = simpson_composta(f20b, -0.5, 0.5, 6)
print(f"Aproximação (n=6): {aprox20b:.8f}")
print(f"Erro: {abs(exato20b - aprox20b):.2e}")

# 18c) ∫₀.₇₅¹.₇₅ (sen²x - 2x sen x + 1) dx
print("\n--- 18c (Simpson) ---")
def f20c(x):
    return math.sin(x)**2 - 2*x*math.sin(x) + 1
def integracao_alta_precisao(f, a, b, n=10000):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma
exato20c = integracao_alta_precisao(f20c, 0.75, 1.75, 10000)
aprox20c = simpson_composta(f20c, 0.75, 1.75, 8)
print(f"Aproximação (n=8): {aprox20c:.8f}")
print(f"Erro: {abs(exato20c - aprox20c):.2e}")
```
