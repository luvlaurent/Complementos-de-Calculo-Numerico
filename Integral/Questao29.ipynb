# Questão 29

Determine os valores de \(n\) e \(h\) necessários para obter uma aproximação de

\[
\int_{1}^{2} \frac{1}{x+4} \, dx
\]

com precisão de \(10^{-5}\) e calcule a aproximação. Use a:

**a)** regra do trapézio composta

**b)** regra de Simpson composta

**c)** regra do ponto médio composta

```python

import math


def f29(x):
    return 1 / (x + 4)

# Derivadas
def f_linha29(x):
    return -1 / (x + 4)**2

def f_2linha29(x):
    return 2 / (x + 4)**3

def f_4linha29(x):
    return 24 / (x + 4)**5

a, b = 1, 2

print(f"Intervalo: [{a}, {b}]")

# Máximos no intervalo [1,2] ocorrem no menor x (denominador menor)
M2 = abs(f_2linha29(1))  # 2/(5³) = 2/125 = 0.016
M4 = abs(f_4linha29(1))  # 24/(5⁵) = 24/3125 = 0.00768

print(f"Máximo |f''(x)| = {M2:.6f}")
print(f"Máximo |f⁽⁴⁾(x)| = {M4:.6f}")

# a) Trapézio

print("\n--- a) Trapézio Composta ---")
precisao = 1e-5
n_trap = math.sqrt((b-a)**3 * M2 / (12 * precisao))
n_trap = int(n_trap) + 1
h_trap = (b-a) / n_trap

print(f"n ≥ {n_trap}")
print(f"h = {h_trap:.6f}")

# Cálculo da aproximação
def trapezio_composta(f, a, b, n):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma

aprox_trap = trapezio_composta(f29, a, b, n_trap)
exato = math.log(6/5)  # ∫₁² 1/(x+4) dx = ln(6/5)
print(f"\nAproximação com n={n_trap}: {aprox_trap:.10f}")
print(f"Valor exato: {exato:.10f}")
print(f"Erro: {abs(exato - aprox_trap):.2e}")

# b) Simpson

print("\n--- b) Simpson Composta ---")
n_simp = ((b-a)**5 * M4 / (180 * precisao)) ** (1/4)
n_simp = int(n_simp) + 1
if n_simp % 2 != 0:
    n_simp += 1
h_simp = (b-a) / n_simp

print(f"n ≥ {n_simp}")
print(f"h = {h_simp:.6f}")

def simpson_composta(f, a, b, n):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(a + i * h)
        else:
            soma += 4 * f(a + i * h)
    return (h / 3) * soma

aprox_simp = simpson_composta(f29, a, b, n_simp)
print(f"\nAproximação com n={n_simp}: {aprox_simp:.10f}")
print(f"Valor exato: {exato:.10f}")
print(f"Erro: {abs(exato - aprox_simp):.2e}")


# c) Ponto médio

print("\n--- c) Ponto Médio Composta ---")
n_pm = math.sqrt((b-a)**3 * M2 / (24 * precisao))
n_pm = int(n_pm) + 1
h_pm = (b-a) / n_pm

print(f"n ≥ {n_pm}")
print(f"h = {h_pm:.6f}")

def ponto_medio_composta(f, a, b, n):
    h = (b - a) / n
    soma = 0
    for i in range(n):
        xi = a + (i + 0.5) * h
        soma += f(xi)
    return h * soma

aprox_pm = ponto_medio_composta(f29, a, b, n_pm)
print(f"\nAproximação com n={n_pm}: {aprox_pm:.10f}")
print(f"Valor exato: {exato:.10f}")
print(f"Erro: {abs(exato - aprox_pm):.2e}")
```
