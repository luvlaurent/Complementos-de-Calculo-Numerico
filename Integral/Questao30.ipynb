Repita o exercício 29 para a integral \(\int_1^2 x \ln x \, dx\).

```python

import math

def f30(x):
    return x * math.log(x)

# Derivadas
def f_linha30(x):
    return math.log(x) + 1

def f_2linha30(x):
    return 1/x

def f_4linha30(x):
    return -6 / x**4

a, b = 1, 2
exato = 2*math.log(2) - 3/4  # ≈ 0.636294361

print(f"Intervalo: [{a}, {b}]")
print(f"Valor exato: {exato:.10f}")

# Máximos no intervalo [1,2]
M2 = max(abs(f_2linha30(x)) for x in [1, 2])  # max = 1 em x=1
M4 = max(abs(f_4linha30(x)) for x in [1, 2])  # max = 6 em x=1

print(f"Máximo |f''(x)| = {M2:.6f}")
print(f"Máximo |f⁽⁴⁾(x)| = {M4:.6f}")

# a) Trapézio

print("\n--- a) Trapézio Composta ---")
precisao = 1e-5
n_trap = math.sqrt((b-a)**3 * M2 / (12 * precisao))
n_trap = int(n_trap) + 1
h_trap = (b-a) / n_trap

print(f"n ≥ {n_trap}")
print(f"h = {h_trap:.6f}")

def trapezio_composta(f, a, b, n):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma

aprox_trap = trapezio_composta(f30, a, b, n_trap)
print(f"\nAproximação com n={n_trap}: {aprox_trap:.10f}")
print(f"Erro: {abs(exato - aprox_trap):.2e}")

# b) Simpson

print("\n--- b) Simpson Composta ---")
n_simp = ((b-a)**5 * M4 / (180 * precisao)) ** (1/4)
n_simp = int(n_simp) + 1
if n_simp % 2 != 0:
    n_simp += 1
h_simp = (b-a) / n_simp

print(f"n ≥ {n_simp}")
print(f"h = {h_simp:.6f}")

def simpson_composta(f, a, b, n):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(a + i * h)
        else:
            soma += 4 * f(a + i * h)
    return (h / 3) * soma

aprox_simp = simpson_composta(f30, a, b, n_simp)
print(f"\nAproximação com n={n_simp}: {aprox_simp:.10f}")
print(f"Erro: {abs(exato - aprox_simp):.2e}")

# c) Ponto médio

print("\n--- c) Ponto Médio Composta ---")
n_pm = math.sqrt((b-a)**3 * M2 / (24 * precisao))
n_pm = int(n_pm) + 1
h_pm = (b-a) / n_pm

print(f"n ≥ {n_pm}")
print(f"h = {h_pm:.6f}")

def ponto_medio_composta(f, a, b, n):
    h = (b - a) / n
    soma = 0
    for i in range(n):
        xi = a + (i + 0.5) * h
        soma += f(xi)
    return h * soma

aprox_pm = ponto_medio_composta(f30, a, b, n_pm)
print(f"\nAproximação com n={n_pm}: {aprox_pm:.10f}")
print(f"Erro: {abs(exato - aprox_pm):.2e}")
```
