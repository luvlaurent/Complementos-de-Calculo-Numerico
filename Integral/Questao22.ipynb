# Questão 22

Utilize a regra do ponto médio composta com \(n + 2\) subintervalos para obter uma aproximação das integrais no Exercício 2.

```python

import math

def ponto_medio_composta(f, a, b, n):
    h = (b - a) / n
    soma = 0
    for i in range(n):
        xi = a + (i + 0.5) * h
        soma += f(xi)
    return h * soma

# 2a) ∫₋₀·₂₅⁰·²⁵ cos² x dx
print("\n--- Questão 2a ---")
def f22a(x):
    return math.cos(x)**2
def antiderivada22a(x):
    return (x/2) + math.sin(2*x)/4
exato22a = antiderivada22a(0.25) - antiderivada22a(-0.25)
aprox22a = ponto_medio_composta(f22a, -0.25, 0.25, 2 + 2)
print(f"Aproximação: {aprox22a:.8f}")
print(f"Exato: {exato22a:.8f}")

# 2b) ∫₋₀·₅⁰ x ln(x+1) dx
print("\n--- Questão 2b ---")
def f22b(x):
    return x * math.log(x + 1)
def antiderivada22b(x):
    return (x**2/2) * math.log(x+1) - (x**2/4) + (x/2) - (1/2)*math.log(x+1)
exato22b = antiderivada22b(0) - antiderivada22b(-0.5)
aprox22b = ponto_medio_composta(f22b, -0.5, 0, 2 + 2)
print(f"Aproximação: {aprox22b:.8f}")
print(f"Exato: {exato22b:.8f}")

# 2c) ∫₀·₇₅¹·³ (sen²x - 2x sen x + 1) dx
print("\n--- Questão 2c ---")
def f22c(x):
    return math.sin(x)**2 - 2*x*math.sin(x) + 1
def integracao_alta(f, a, b, n=10000):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma
exato22c = integracao_alta(f22c, 0.75, 1.3, 10000)
aprox22c = ponto_medio_composta(f22c, 0.75, 1.3, 2 + 2)
print(f"Aproximação: {aprox22c:.8f}")
print(f"Exato: {exato22c:.8f}")

# 2d) ∫_e^{e+1} 1/(x ln x) dx
print("\n--- Questão 2d ---")
def f22d(x):
    return 1 / (x * math.log(x))
def antiderivada22d(x):
    return math.log(math.log(x))
exato22d = antiderivada22d(math.e + 1) - antiderivada22d(math.e)
aprox22d = ponto_medio_composta(f22d, math.e, math.e + 1, 2 + 2)
print(f"Aproximação: {aprox22d:.8f}")
print(f"Exato: {exato22d:.8f}")
```
