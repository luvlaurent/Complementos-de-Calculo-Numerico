# Questão 27

Determine os valores de \(n\) e \(h\) necessários para obter uma aproximação de

\[
\int_{0}^{\pi} e^{2x} \sin 3x \, dx
\]

com precisão de \(10^{-4}\). Use a:

**a)** regra do trapézio composta

**b)** regra de Simpson composta

**c)** regra do ponto médio composta

```python

import math

print("=" * 70)
print("QUESTÃO 27 - DETERMINAR n E h PARA PRECISÃO 10⁻⁴")
print("=" * 70)

def f27(x):
    return math.exp(2*x) * math.sin(3*x)

# Derivadas para o limitante de erro
def f_linha(x):
    # f'(x) = e^{2x}(2 sen(3x) + 3 cos(3x))
    return math.exp(2*x) * (2*math.sin(3*x) + 3*math.cos(3*x))

def f_2linha(x):
    # f''(x) = e^{2x}(-5 sen(3x) + 12 cos(3x))
    return math.exp(2*x) * (-5*math.sin(3*x) + 12*math.cos(3*x))

def f_4linha(x):
    # Aproximação: derivada quarta - para Simpson
    return math.exp(2*x) * (119*math.sin(3*x) - 120*math.cos(3*x))

a, b = 0, math.pi

# Encontrando máximos das derivadas no intervalo
print(f"Intervalo: [{a}, {b}]")

# Amostragem para encontrar máximos aproximados
x_amostras = [a + i*(b-a)/1000 for i in range(1001)]
max_f2 = max(abs(f_2linha(x)) for x in x_amostras)
max_f4 = max(abs(f_4linha(x)) for x in x_amostras)

print(f"Máximo |f''(x)| ≈ {max_f2:.2e}")
print(f"Máximo |f⁽⁴⁾(x)| ≈ {max_f4:.2e}")

# ------------------------------------------------------------
# a) Regra do trapézio composta
# Erro: E = -(b-a)³/(12n²) * f''(ξ)
# |E| ≤ (b-a)³/(12n²) * M2
# n ≥ √[(b-a)³ * M2 / (12 * precisão)]
# ------------------------------------------------------------
print("\n--- a) Trapézio Composta ---")
precisao = 1e-4
M2 = max_f2
n_trap = math.sqrt((b-a)**3 * M2 / (12 * precisao))
n_trap = int(n_trap) + 1
h_trap = (b-a) / n_trap
print(f"n ≥ {n_trap}")
print(f"h = {h_trap:.6f}")

# ------------------------------------------------------------
# b) Regra de Simpson composta
# Erro: E = -(b-a)⁵/(180n⁴) * f⁽⁴⁾(ξ)
# |E| ≤ (b-a)⁵/(180n⁴) * M4
# n ≥ [(b-a)⁵ * M4 / (180 * precisão)]^(1/4)
# ------------------------------------------------------------
print("\n--- b) Simpson Composta ---")
M4 = max_f4
n_simp = ((b-a)**5 * M4 / (180 * precisao)) ** (1/4)
n_simp = int(n_simp) + 1
if n_simp % 2 != 0:
    n_simp += 1
h_simp = (b-a) / n_simp
print(f"n ≥ {n_simp}")
print(f"h = {h_simp:.6f}")

# ------------------------------------------------------------
# c) Regra do ponto médio composta
# Erro: E = (b-a)³/(24n²) * f''(ξ)
# |E| ≤ (b-a)³/(24n²) * M2
# n ≥ √[(b-a)³ * M2 / (24 * precisão)]
# ------------------------------------------------------------
print("\n--- c) Ponto Médio Composta ---")
n_pm = math.sqrt((b-a)**3 * M2 / (24 * precisao))
n_pm = int(n_pm) + 1
h_pm = (b-a) / n_pm
print(f"n ≥ {n_pm}")
print(f"h = {h_pm:.6f}")
```
