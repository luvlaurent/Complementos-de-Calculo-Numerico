```python

import math

def simpson_composta(f, a, b, n):
    """Regra de Simpson composta (n deve ser par)"""
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(a + i * h)
        else:
            soma += 4 * f(a + i * h)
    return (h / 3) * soma

# 17a) ∫₁² x ln x dx
print("\n--- 17a (Simpson) ---")
def f19a(x):
    return x * math.log(x)
exato19a = 2*math.log(2) - 3/4
aprox19a = simpson_composta(f19a, 1, 2, 4)
print(f"Aproximação (n=4): {aprox19a:.8f}")
print(f"Erro: {abs(exato19a - aprox19a):.2e}")

# 17b) ∫₋₂² x³ e^x dx
print("\n--- 17b (Simpson) ---")
def f19b(x):
    return x**3 * math.exp(x)
def antiderivada19b(x):
    return math.exp(x) * (x**3 - 3*x**2 + 6*x - 6)
exato19b = antiderivada19b(2) - antiderivada19b(-2)
aprox19b = simpson_composta(f19b, -2, 2, 4)
print(f"Aproximação (n=4): {aprox19b:.8f}")
print(f"Erro: {abs(exato19b - aprox19b):.2e}")

# 17c) ∫₀² 2/(x²+4) dx
print("\n--- 17c (Simpson) ---")
def f19c(x):
    return 2 / (x**2 + 4)
exato19c = math.atan(1)
aprox19c = simpson_composta(f19c, 0, 2, 6)
print(f"Aproximação (n=6): {aprox19c:.8f}")
print(f"Erro: {abs(exato19c - aprox19c):.2e}")

# 17d) ∫₀^π x² cos x dx
print("\n--- 17d (Simpson) ---")
def f19d(x):
    return x**2 * math.cos(x)
def antiderivada19d(x):
    return x**2 * math.sin(x) + 2*x*math.cos(x) - 2*math.sin(x)
exato19d = antiderivada19d(math.pi) - antiderivada19d(0)
aprox19d = simpson_composta(f19d, 0, math.pi, 6)
print(f"Aproximação (n=6): {aprox19d:.8f}")
print(f"Erro: {abs(exato19d - aprox19d):.2e}")

# 17e) ∫₀^π e^{2x} sen(3x) dx
print("\n--- 17e (Simpson) ---")
def f19e(x):
    return math.exp(2*x) * math.sin(3*x)
def antiderivada19e(x):
    return math.exp(2*x) * (2*math.sin(3*x) - 3*math.cos(3*x)) / 13
exato19e = antiderivada19e(math.pi) - antiderivada19e(0)
aprox19e = simpson_composta(f19e, 0, math.pi, 8)
print(f"Aproximação (n=8): {aprox19e:.8f}")
print(f"Erro: {abs(exato19e - aprox19e):.2e}")

# 17f) ∫₁³ x/(x²+4) dx
print("\n--- 17f (Simpson) ---")
def f19f(x):
    return x / (x**2 + 4)
exato19f = 0.5 * math.log(13/5)
aprox19f = simpson_composta(f19f, 1, 3, 8)
print(f"Aproximação (n=8): {aprox19f:.8f}")
print(f"Erro: {abs(exato19f - aprox19f):.2e}")

# 17g) ∫₃⁵ 1/√(x²-4) dx
print("\n--- 17g (Simpson) ---")
def f19g(x):
    return 1 / math.sqrt(x**2 - 4)
exato19g = math.log((5 + math.sqrt(21)) / (3 + math.sqrt(5)))
aprox19g = simpson_composta(f19g, 3, 5, 8)
print(f"Aproximação (n=8): {aprox19g:.8f}")
print(f"Erro: {abs(exato19g - aprox19g):.2e}")

# 17h) ∫₀^{3π/8} tan x dx
print("\n--- 17h (Simpson) ---")
def f19h(x):
    return math.tan(x)
exato19h = -math.log(math.cos(3*math.pi/8))
aprox19h = simpson_composta(f19h, 0, 3*math.pi/8, 8)
print(f"Aproximação (n=8): {aprox19h:.8f}")
print(f"Erro: {abs(exato19h - aprox19h):.2e}")
```
