# Questão 24

Obtenha uma aproximação de \(\int_0^2 x^2 e^{-x^2} \, dx\) usando \(h = 0,25\). Use a:

**a)** regra do trapézio composta

**b)** regra de Simpson composta

**c)** regra do ponto médio composta

```python
# Questão 24 - ∫₀² x² e^{-x²} dx com h = 0.25
import math

def f24(x):
    return x**2 * math.exp(-x**2)

a, b = 0, 2
h = 0.25
n = int((b - a) / h)  # n = 8

x_vals = [a + i*h for i in range(n+1)]
f_vals = [f24(x) for x in x_vals]

print(f"Intervalo: [{a}, {b}]")
print(f"h = {h}")
print(f"n = {n}")
print("\nTabela de pontos:")
print("   x    |    f(x)")
print("-" * 25)
for i in range(0, len(x_vals), 2):
    print(f"  {x_vals[i]:.2f}  |  {f_vals[i]:.8f}")

# Valor exato aproximado (integral gaussiana relacionada)
# ∫₀^∞ x² e^{-x²} dx = √π/4 ≈ 0.443113462726
# Como vamos até 2, a contribuição além é pequena
def integracao_alta(f, a, b, n=10000):
    hh = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * hh)
    return (hh / 2) * soma
exato24 = integracao_alta(f24, 0, 2, 100000)
print(f"\nValor exato (referência): {exato24:.8f}")

# a) Trapézio composta

print("\n--- a) Trapézio Composta ---")
trap24 = (h/2) * (f_vals[0] + f_vals[-1] + 2*sum(f_vals[1:-1]))
print(f"Aproximação: {trap24:.8f}")
print(f"Erro: {abs(exato24 - trap24):.2e}")

# b) Simpson composta

print("\n--- b) Simpson Composta ---")
if n % 2 == 0:
    simp24 = (h/3) * (f_vals[0] + f_vals[-1] + 4*sum(f_vals[1:-1:2]) + 2*sum(f_vals[2:-1:2]))
    print(f"Aproximação: {simp24:.8f}")
    print(f"Erro: {abs(exato24 - simp24):.2e}")
else:
    print("n não é par para Simpson!")

# ------------------------------------------------------------
# c) Ponto médio composta
# ------------------------------------------------------------
print("\n--- c) Ponto Médio Composta ---")
pm24 = 0
for i in range(n):
    x_mid = a + (i + 0.5) * h
    pm24 += f24(x_mid)
pm24 = pm24 * h
print(f"Aproximação: {pm24:.8f}")
print(f"Erro: {abs(exato24 - pm24):.2e}")
```
