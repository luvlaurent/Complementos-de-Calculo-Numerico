```python

import math

def trapezio_composta(f, a, b, n):
    """Regra do trapézio composta"""
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma

# 17a) ∫₁² x ln x dx, n = 4

print("\n--- 17a ---")
print("∫₁² x ln x dx, n = 4")

def f17a(x):
    return x * math.log(x)

exato17a = 2*math.log(2) - 3/4  # antiderivada: (x²/2)ln x - x²/4
print(f"Valor exato: {exato17a:.8f}")

aprox17a = trapezio_composta(f17a, 1, 2, 4)
print(f"Aproximação (n=4): {aprox17a:.8f}")
print(f"Erro: {abs(exato17a - aprox17a):.2e}")

# 17b) ∫₋₂² x³ e^x dx, n = 4

print("\n--- 17b ---")
print("∫₋₂² x³ e^x dx, n = 4")

def f17b(x):
    return x**3 * math.exp(x)

def antiderivada17b(x):
    return math.exp(x) * (x**3 - 3*x**2 + 6*x - 6)

exato17b = antiderivada17b(2) - antiderivada17b(-2)
print(f"Valor exato: {exato17b:.8f}")

aprox17b = trapezio_composta(f17b, -2, 2, 4)
print(f"Aproximação (n=4): {aprox17b:.8f}")
print(f"Erro: {abs(exato17b - aprox17b):.2e}")

# 17c) ∫₀² 2/(x²+4) dx, n = 6

print("\n--- 17c ---")
print("∫₀² 2/(x²+4) dx, n = 6")

def f17c(x):
    return 2 / (x**2 + 4)

exato17c = math.atan(1)  # resulta em π/4
print(f"Valor exato: {exato17c:.8f}")

aprox17c = trapezio_composta(f17c, 0, 2, 6)
print(f"Aproximação (n=6): {aprox17c:.8f}")
print(f"Erro: {abs(exato17c - aprox17c):.2e}")

# 17d) ∫₀^π x² cos x dx, n = 6

print("\n--- 17d ---")
print("∫₀^π x² cos x dx, n = 6")

def f17d(x):
    return x**2 * math.cos(x)

def antiderivada17d(x):
    return x**2 * math.sin(x) + 2*x*math.cos(x) - 2*math.sin(x)

exato17d = antiderivada17d(math.pi) - antiderivada17d(0)
print(f"Valor exato: {exato17d:.8f}")

aprox17d = trapezio_composta(f17d, 0, math.pi, 6)
print(f"Aproximação (n=6): {aprox17d:.8f}")
print(f"Erro: {abs(exato17d - aprox17d):.2e}")

# 17e) ∫₀^π e^{2x} sen(3x) dx, n = 8

print("\n--- 17e ---")
print("∫₀^π e^{2x} sen(3x) dx, n = 8")

def f17e(x):
    return math.exp(2*x) * math.sin(3*x)

# antiderivada: e^{2x}(2 sen(3x) - 3 cos(3x))/13
def antiderivada17e(x):
    return math.exp(2*x) * (2*math.sin(3*x) - 3*math.cos(3*x)) / 13

exato17e = antiderivada17e(math.pi) - antiderivada17e(0)
print(f"Valor exato: {exato17e:.8f}")

aprox17e = trapezio_composta(f17e, 0, math.pi, 8)
print(f"Aproximação (n=8): {aprox17e:.8f}")
print(f"Erro: {abs(exato17e - aprox17e):.2e}")

# 17f) ∫₁³ x/(x²+4) dx, n = 8

print("\n--- 17f ---")
print("∫₁³ x/(x²+4) dx, n = 8")

def f17f(x):
    return x / (x**2 + 4)

exato17f = 0.5 * math.log((9+4)/(1+4))
print(f"Valor exato: {exato17f:.8f}")

aprox17f = trapezio_composta(f17f, 1, 3, 8)
print(f"Aproximação (n=8): {aprox17f:.8f}")
print(f"Erro: {abs(exato17f - aprox17f):.2e}")

# 17g) ∫₃⁵ 1/√(x²-4) dx, n = 8

print("\n--- 17g ---")
print("∫₃⁵ 1/√(x²-4) dx, n = 8")

def f17g(x):
    return 1 / math.sqrt(x**2 - 4)

exato17g = math.log((5 + math.sqrt(21)) / (3 + math.sqrt(5)))
print(f"Valor exato: {exato17g:.8f}")

aprox17g = trapezio_composta(f17g, 3, 5, 8)
print(f"Aproximação (n=8): {aprox17g:.8f}")
print(f"Erro: {abs(exato17g - aprox17g):.2e}")

# 17h) ∫₀^{3π/8} tan x dx, n = 8

print("\n--- 17h ---")
print("∫₀^{3π/8} tan x dx, n = 8")

def f17h(x):
    return math.tan(x)

exato17h = -math.log(math.cos(3*math.pi/8))
print(f"Valor exato: {exato17h:.8f}")

aprox17h = trapezio_composta(f17h, 0, 3*math.pi/8, 8)
print(f"Aproximação (n=8): {aprox17h:.8f}")
print(f"Erro: {abs(exato17h - aprox17h):.2e}")
```
