Repita o Exercício 27 para a integral \(\int_0^\pi x^2 \cos x \, dx\).

```python

import math


def f28(x):
    return x**2 * math.cos(x)

# Derivadas
def f_linha28(x):
    return 2*x*math.cos(x) - x**2*math.sin(x)

def f_2linha28(x):
    return 2*math.cos(x) - 4*x*math.sin(x) - x**2*math.cos(x)

def f_4linha28(x):
    # Derivada quarta aproximada
    return (12 - x**2)*math.cos(x) + 8*x*math.sin(x)

a, b = 0, math.pi

print(f"Intervalo: [{a:.2f}, {b:.2f}]")

# Encontrando máximos
x_amostras = [a + i*(b-a)/1000 for i in range(1001)]
max_f2 = max(abs(f_2linha28(x)) for x in x_amostras)
max_f4 = max(abs(f_4linha28(x)) for x in x_amostras)

print(f"Máximo |f''(x)| ≈ {max_f2:.2e}")
print(f"Máximo |f⁽⁴⁾(x)| ≈ {max_f4:.2e}")

# a) Trapézio

print("\n--- a) Trapézio Composta ---")
precisao = 1e-4
n_trap = math.sqrt((b-a)**3 * max_f2 / (12 * precisao))
n_trap = int(n_trap) + 1
h_trap = (b-a) / n_trap
print(f"n ≥ {n_trap}")
print(f"h = {h_trap:.6f}")

# b) Simpson

print("\n--- b) Simpson Composta ---")
n_simp = ((b-a)**5 * max_f4 / (180 * precisao)) ** (1/4)
n_simp = int(n_simp) + 1
if n_simp % 2 != 0:
    n_simp += 1
h_simp = (b-a) / n_simp
print(f"n ≥ {n_simp}")
print(f"h = {h_simp:.6f}")

# c) Ponto médio

print("\n--- c) Ponto Médio Composta ---")
n_pm = math.sqrt((b-a)**3 * max_f2 / (24 * precisao))
n_pm = int(n_pm) + 1
h_pm = (b-a) / n_pm
print(f"n ≥ {n_pm}")
print(f"h = {h_pm:.6f}")
```
