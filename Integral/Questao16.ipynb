# Questão 16

Aproxime as seguintes integrais usando as fórmulas (4.25) a (4.32). Qual das partes (c) e (d) fornece a melhor aproximação?

**a)** \(\int_2^{2.5} \frac{(\ln x)^3}{3x} \, dx\)

**b)** \(\int_{0.5}^{1} 5xe^{x^2} \, dx\)

**c)** \(\int_1^{10} \frac{1}{x} \, dx\)

**d)** \(\int_1^{5.5} \frac{1}{x} \, dx + \int_{5.5}^{10} \frac{1}{x} \, dx\)

```python

import math

# 16a) ∫₂^{2.5} (ln x)³/(3x) dx

print("\n--- 16a ---")
print("∫₂^{2.5} (ln x)³/(3x) dx")

def f16a(x):
    return (math.log(x)**3) / (3*x)

# Integral exata: 1/12 * (ln(2.5)⁴ - ln(2)⁴)
exato16a = (math.log(2.5)**4 - math.log(2)**4) / 12
print(f"Valor exato: {exato16a:.8f}")

a, b = 2, 2.5
trap16a = (b - a) * (f16a(a) + f16a(b)) / 2
simp16a = (b - a) * (f16a(a) + 4*f16a((a+b)/2) + f16a(b)) / 6
pm16a = (b - a) * f16a((a+b)/2)

print(f"Trapézio simples: {trap16a:.8f}")
print(f"Simpson simples: {simp16a:.8f}")
print(f"Ponto médio simples: {pm16a:.8f}")
print(f"Erro trapézio: {abs(exato16a - trap16a):.2e}")

# 16b) ∫_{0.5}^{1} 5x·e^{x²} dx

print("\n--- 16b ---")
print("∫_{0.5}^{1} 5x·e^{x²} dx")

def f16b(x):
    return 5 * x * math.exp(x**2)

# Integral exata: (5/2)(e¹ - e^{0.25})
exato16b = (5/2) * (math.exp(1) - math.exp(0.25))
print(f"Valor exato: {exato16b:.8f}")

a, b = 0.5, 1
trap16b = (b - a) * (f16b(a) + f16b(b)) / 2
simp16b = (b - a) * (f16b(a) + 4*f16b((a+b)/2) + f16b(b)) / 6
pm16b = (b - a) * f16b((a+b)/2)

print(f"Trapézio simples: {trap16b:.8f}")
print(f"Simpson simples: {simp16b:.8f}")
print(f"Ponto médio simples: {pm16b:.8f}")
print(f"Erro trapézio: {abs(exato16b - trap16b):.2e}")

# 16c) ∫₁¹⁰ 1/x dx

print("\n--- 16c ---")
print("∫₁¹⁰ 1/x dx")

def f16c(x):
    return 1/x

exato16c = math.log(10)
print(f"Valor exato: {exato16c:.8f}")

a, b = 1, 10
trap16c = (b - a) * (f16c(a) + f16c(b)) / 2
simp16c = (b - a) * (f16c(a) + 4*f16c((a+b)/2) + f16c(b)) / 6
pm16c = (b - a) * f16c((a+b)/2)

print(f"Trapézio simples: {trap16c:.8f}")
print(f"Simpson simples: {simp16c:.8f}")
print(f"Ponto médio simples: {pm16c:.8f}")
print(f"Erro trapézio: {abs(exato16c - trap16c):.2e}")

# 16d) ∫₁^{5.5} 1/x dx + ∫_{5.5}¹⁰ 1/x dx

print("\n--- 16d ---")
print("∫₁^{5.5} 1/x dx + ∫_{5.5}¹⁰ 1/x dx")

exato16d = math.log(10)  # A soma é a mesma que ln(10)
print(f"Valor exato: {exato16d:.8f}")

# Primeira integral
a1, b1 = 1, 5.5
trap1 = (b1 - a1) * (f16c(a1) + f16c(b1)) / 2
simp1 = (b1 - a1) * (f16c(a1) + 4*f16c((a1+b1)/2) + f16c(b1)) / 6
pm1 = (b1 - a1) * f16c((a1+b1)/2)

# Segunda integral
a2, b2 = 5.5, 10
trap2 = (b2 - a2) * (f16c(a2) + f16c(b2)) / 2
simp2 = (b2 - a2) * (f16c(a2) + 4*f16c((a2+b2)/2) + f16c(b2)) / 6
pm2 = (b2 - a2) * f16c((a2+b2)/2)

trap16d = trap1 + trap2
simp16d = simp1 + simp2
pm16d = pm1 + pm2

print(f"Trapézio composto (2 subintervalos): {trap16d:.8f}")
print(f"Simpson composto (2 subintervalos): {simp16d:.8f}")
print(f"Ponto médio composto (2 subintervalos): {pm16d:.8f}")

print(f"\nComparação entre (c) e (d):")
print(f"Parte (c) - trapézio: {trap16c:.8f}, erro: {abs(exato16c - trap16c):.2e}")
print(f"Parte (d) - trapézio: {trap16d:.8f}, erro: {abs(exato16d - trap16d):.2e}")
print(f"A parte (d) fornece melhor aproximação!")
```
