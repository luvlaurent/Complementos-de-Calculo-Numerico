No cálculo de muitas variáveis e em cursos de estatística, é mostrado que

\[
\int_{-\infty}^{\infty} \frac{1}{\sigma \sqrt{2\pi}} e^{-(1/2)(x/\sigma)^2} \, dx = 1,
\]

para todo \(\sigma\) positivo. A função

\[
f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-(1/2)(x/\sigma)^2}
\]

é a função de densidade normal com média \(\mu = 0\) e desvio padrão \(\sigma\). A probabilidade de um valor descrito por essa distribuição, escolhido aleatoriamente, estar em \([a, b]\) é dada por \(\int_a^b f(x) \, dx\). Obtenha uma aproximação, com a precisão de \(10^{-5}\), da probabilidade de um valor escolhido aleatoriamente, descrito por essa distribuição, estar em

**a)** \([-\sigma, \sigma]\)

**b)** \([-2\sigma, 2\sigma]\)

**c)** \([-3\sigma, 3\sigma]\)

```python

import math

print("""
Distribuição normal com média µ = 0 e desvio padrão σ
Função densidade: f(x) = 1/(σ√(2π)) * e^(-x²/(2σ²))

Vamos calcular P(-kσ ≤ X ≤ kσ) para k = 1, 2, 3
A integral de -kσ a kσ de f(x) dx

Por simetria, = 2 * ∫₀^{kσ} f(x) dx
""")

def normal_pdf(x, sigma):
    """Função densidade da normal com média 0"""
    return (1/(sigma * math.sqrt(2*math.pi))) * math.exp(-x**2/(2*sigma**2))

def integracao_simpson_alta(f, a, b, sigma, n=1000):
    """Integração de Simpson com alta precisão"""
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    soma = f(a, sigma) + f(b, sigma)
    for i in range(1, n):
        if i % 2 == 0:
            soma += 2 * f(a + i*h, sigma)
        else:
            soma += 4 * f(a + i*h, sigma)
    return (h/3) * soma

def probabilidade(k, sigma=1):
    """
    Calcula P(-kσ ≤ X ≤ kσ)
    Integra de -kσ a kσ
    """
    limite = k * sigma
    # Por simetria, integramos de 0 a limite e multiplicamos por 2
    prob = 2 * integracao_simpson_alta(normal_pdf, 0, limite, sigma, n=10000)
    return prob

print("\n" + "-" * 50)
print("RESULTADOS (com σ = 1):")
print("-" * 50)

# a) [-σ, σ] = [-1, 1]
k1 = 1
prob1 = probabilidade(k1)
print(f"\na) P(-σ ≤ X ≤ σ) = P(-{k1} ≤ X ≤ {k1})")
print(f"   Aproximação: {prob1:.10f}")
print(f"   Valor esperado (teórico): 0.6826894921")
print(f"   Erro: {abs(prob1 - 0.6826894921):.2e}")

# b) [-2σ, 2σ] = [-2, 2]
k2 = 2
prob2 = probabilidade(k2)
print(f"\nb) P(-2σ ≤ X ≤ 2σ)")
print(f"   Aproximação: {prob2:.10f}")
print(f"   Valor esperado (teórico): 0.9544997361")
print(f"   Erro: {abs(prob2 - 0.9544997361):.2e}")

# c) [-3σ, 3σ] = [-3, 3]
k3 = 3
prob3 = probabilidade(k3)
print(f"\nc) P(-3σ ≤ X ≤ 3σ)")
print(f"   Aproximação: {prob3:.10f}")
print(f"   Valor esperado (teórico): 0.9973002039")
print(f"   Erro: {abs(prob3 - 0.9973002039):.2e}")

print("\n" + "-" * 50)
print("REGRAS EMPÍRICAS (68-95-99.7):")
print("-" * 50)
print(f"±1σ: 68.27% → {prob1*100:.4f}%")
print(f"±2σ: 95.45% → {prob2*100:.4f}%")
print(f"±3σ: 99.73% → {prob3*100:.4f}%")
```
