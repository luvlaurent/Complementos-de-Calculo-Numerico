# Questão 18

Utilize a regra do trapézio composta com os valores indicados de \(n\) para obter uma aproximação das seguintes integrais.

**a)** \(\int_{-0.5}^{0.5} \cos^2 x \, dx, \quad n = 4\)

**b)** \(\int_{-0.5}^{0.5} x \ln(x+1) \, dx, \quad n = 6\)

**c)** \(\int_{0.75}^{1.75} (\sin^2 x - 2x \sin x + 1) \, dx, \quad n = 8\)

```python

import math


def trapezio_composta(f, a, b, n):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma

# 18a) ∫₋₀.₅⁰.⁵ cos² x dx, n = 4

print("\n--- 18a ---")
print("∫₋₀.₅⁰.⁵ cos² x dx, n = 4")

def f18a(x):
    return math.cos(x)**2

def antiderivada18a(x):
    return (x/2) + math.sin(2*x)/4

exato18a = antiderivada18a(0.5) - antiderivada18a(-0.5)
print(f"Valor exato: {exato18a:.8f}")

aprox18a = trapezio_composta(f18a, -0.5, 0.5, 4)
print(f"Aproximação (n=4): {aprox18a:.8f}")
print(f"Erro: {abs(exato18a - aprox18a):.2e}")

# 18b) ∫₋₀.₅⁰.⁵ x ln(x+1) dx, n = 6

print("\n--- 18b ---")
print("∫₋₀.₅⁰.⁵ x ln(x+1) dx, n = 6")

def f18b(x):
    return x * math.log(x + 1)

# Antiderivada: (x²/2)ln(x+1) - (x²/4) + (x/2) - (1/2)ln(x+1)
def antiderivada18b(x):
    return (x**2/2) * math.log(x+1) - (x**2/4) + (x/2) - (1/2)*math.log(x+1)

exato18b = antiderivada18b(0.5) - antiderivada18b(-0.5)
print(f"Valor exato: {exato18b:.8f}")

aprox18b = trapezio_composta(f18b, -0.5, 0.5, 6)
print(f"Aproximação (n=6): {aprox18b:.8f}")
print(f"Erro: {abs(exato18b - aprox18b):.2e}")

# 18c) ∫₀.₇₅¹.₇₅ (sen²x - 2x sen x + 1) dx, n = 8

print("\n--- 18c ---")
print("∫₀.₇₅¹.₇₅ (sen²x - 2x sen x + 1) dx, n = 8")

def f18c(x):
    return math.sin(x)**2 - 2*x*math.sin(x) + 1

# Integral pode ser calculada numericamente com alta precisão
def integracao_romberg(f, a, b, n=100):
    h = (b - a) / n
    soma = f(a) + f(b)
    for i in range(1, n):
        soma += 2 * f(a + i * h)
    return (h / 2) * soma

exato18c = integracao_romberg(f18c, 0.75, 1.75, 10000)
print(f"Valor exato (aproximado): {exato18c:.8f}")

aprox18c = trapezio_composta(f18c, 0.75, 1.75, 8)
print(f"Aproximação (n=8): {aprox18c:.8f}")
print(f"Erro: {abs(exato18c - aprox18c):.2e}")
```
