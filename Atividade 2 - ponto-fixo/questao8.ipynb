# 8. Utilize um método de iteração de ponto fixo para determinar
# uma solução com precisão de 1e-2 para:
# x^3 - x - 1 = 0, no intervalo [1, 2].
# Use p0 = 1.

```python
import math  # Biblioteca matemática (não é necessária aqui, mas mantida por padronização)

def g(x):
    return (x + 1)**(1/3)

p = 1
tol = 1e-5
erro = 1
iteracao = 0
max_iter = 20

while erro > tol and iteracao < max_iter:

    # Calcula a nova aproximação usando a função g(x)
    p_new = g(p)

    # Calcula o erro absoluto entre as iterações consecutivas
    erro = abs(p_new - p)

    # Atualiza o valor de p para a próxima iteração
    p = p_new

    # Incrementa o contador de iterações
    iteracao += 1

if iteracao == max_iter:
    print("Atenção: número máximo de iterações atingido!")
else:
    print("Convergência atingida!")


# Exibe os resultados finais
print(f"Solução aproximada: {p:.4f}")
print(f"Iterações: {iteracao}")
print(f"Erro final: {erro:.6f}")
```
