# 7. Utilize um método de iteração de ponto fixo para determinar
# uma solução com precisão de 1e-2 para:
# x^4 - 3x^2 - 3 = 0, no intervalo [1, 2].
# Use p0 = 1.

```python
import math

def g(x):
    return math.sqrt((x**4 - 3)/3)

p = 1
tol = 1e-2
erro = 1
iteracao = 0

while erro > tol:
    p_new = g(p)
    erro = abs(p_new - p)

    p = p_new
    iteracao += 1

print(f"Solução aproximada: {p:.4f}")
print(f"Iterações: {iteracao}")
```
