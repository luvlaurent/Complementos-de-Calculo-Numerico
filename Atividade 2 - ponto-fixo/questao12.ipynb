# 12. Utilize um método de ponto fixo para encontrar uma aproximação
# de raiz cúbica de 25 com precisão de 1e-4.

# Compare o resultado e o número de iterações com o Exercício 13 da Seção 2.1.

```python
import math

def g(x):
    return (2*x + 25/(x**2))/3

p = 1
tol = 1e-4
erro = 1
iteracao = 0

while erro > tol:
    p_new = g(p)
    erro = abs(p_new - p)

    p = p_new
    iteracao += 1

print(f"Raiz cúbica de 25 ≈ {p:.6f}")
print(f"Iterações: {iteracao}")
```
