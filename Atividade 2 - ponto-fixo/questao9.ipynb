# Utilize o Teorema 2.3 para mostrar que g(x) = x + 0.5*sin(x/2)
# tem um único ponto fixo em [0, 2*pi].

# Aplique a iteração de ponto fixo para encontrar uma aproximação
# com precisão de 1e-2.

# Utilize o Corolário 2.5 para estimar o número de iterações
# necessárias e compare com o valor real obtido.

```python
import math

def g(x):
    return x + 0.5 * math.sin(x/2)

p = 1
tol = 1e-2
erro = 1
iteracao = 0

while erro > tol:
    p_new = g(p)
    erro = abs(p_new - p)

    p = p_new
    iteracao += 1

print(f"Ponto fixo: {p:.4f}")
print(f"Iterações: {iteracao}")
```
