# Questão 11

Integral:

∫₁³ x² ln(x²+1) dx

h = 0,2

Resultados:

Trapézio composto = 15,794779

Simpson composto = 15,738642

Ponto médio composto = 15,710618

Valor exato ≈ 15,738674

A regra de Simpson apresentou o melhor resultado.

```python
import math

f = lambda x: x**2 * math.log(x**2 + 1)

a = 1
b = 3
h = 0.2

n = int((b-a)/h)

x = [a + i*h for i in range(n+1)]

trap = h * (
    0.5*f(x[0])
    + sum(f(xi) for xi in x[1:-1])
    + 0.5*f(x[-1])
)

simp = h/3 * (
    f(x[0])
    + f(x[-1])
    + 4*sum(f(x[i]) for i in range(1, n, 2))
    + 2*sum(f(x[i]) for i in range(2, n-1, 2))
)

mid = h * sum(
    f(a + (i+0.5)*h)
    for i in range(n)
)

print("Trapézio composto =", trap)
print("Simpson composto =", simp)
print("Ponto médio composto =", mid)
```
