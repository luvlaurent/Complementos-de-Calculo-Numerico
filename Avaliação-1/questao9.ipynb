Considere

f(x)=3x²−cos(x)

Objetivo: aproximar f''(1,3)

Diferença central de segunda ordem:

f''(x) ≈ [f(x+h)-2f(x)+f(x-h)]/h²

Para h=0,1:

f''(1,3) ≈ 6,267276

Para h=0,01:

f''(1,3) ≈ 6,267497

Valor exato:

f''(x)=6+cos(x)

f''(1,3)=6+cos(1,3)=6,267499

Comparação:

| h | Aproximação |
|---|---|
|0,1|6,267276|
|0,01|6,267497|

Quanto menor o passo, maior a precisão.

```python

h = 0.01

f129 = 13.78176
f130 = 14.04276
f131 = 14.30741

f2 = (f131 - 2*f130 + f129)/(h**2)

print("f''(1,3) ≈", f2)
```
