## Questão 1

Determine uma aproximação de \(\sqrt{2}\) correta até \(10^{-4}\) utilizando o algoritmo da bissecção.

*Sugestão:* considere \(f(x) = x^2 - 2\).

```python
# Questão 1 - Método da Bisseção

def bissecao(a, b, tol=1e-4):
    def f(x):
        return x**2 - 2

    if f(a) * f(b) >= 0:
        raise ValueError("Intervalo inválido.")

    while (b - a)/2 > tol:
        p = (a + b)/2

        if f(p) == 0:
            return p

        if f(a)*f(p) < 0:
            b = p
        else:
            a = p

    return (a + b)/2

raiz = bissecao(1, 2)
print("Aproximação =", raiz)
```
