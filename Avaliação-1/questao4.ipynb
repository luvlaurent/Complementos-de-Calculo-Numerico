## Questão 4

Utilize polinômios interpoladores de Lagrange de graus um, dois e três adequados para aproximar cada um dos seguintes:

- **a)** \(f(8,5)\) se  
  \(f(8,1) = 16,94410\),  
  \(f(8,3) = 17,56492\),  
  \(f(8,6) = 18,50515\),  
  \(f(8,7) = 18,82091\)

- **b)** \(f(0,75)\) se  
  \(f(0,6) = -0,17694460\),  
  \(f(0,7) = 0,01375227\),  
  \(f(0,8) = 0,22363362\),  
  \(f(1,0) = 0,65809197\)

```python
# Questão 4 - Lagrange

def lagrange(x, xs, ys):
    n = len(xs)
    p = 0

    for i in range(n):
        Li = 1

        for j in range(n):
            if i != j:
                Li *= (x - xs[j])/(xs[i] - xs[j])

        p += ys[i]*Li

    return p


# Item a
xs = [8.1, 8.3, 8.6, 8.7]
ys = [16.94410, 17.56492, 18.50515, 18.82091]

print("Grau 1:",
      lagrange(8.5, xs[1:3], ys[1:3]))

print("Grau 2:",
      lagrange(8.5, xs[:3], ys[:3]))

print("Grau 3:",
      lagrange(8.5, xs, ys))


# Item b
xs = [0.6, 0.7, 0.8, 1.0]
ys = [-0.17694460, 0.01375227, 0.22363362, 0.65809197]

print("Grau 1:",
      lagrange(0.75, xs[1:3], ys[1:3]))

print("Grau 2:",
      lagrange(0.75, xs[:3], ys[:3]))

print("Grau 3:",
      lagrange(0.75, xs, ys))
```
